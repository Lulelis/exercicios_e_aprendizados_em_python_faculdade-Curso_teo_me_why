#### ***044 — Condição de Pagamento***

#### Elabore um programa que calcule o valor a ser pago por um produto, considerando o seu **preço normal** e **condição de pagamento**:

  - À vista **dinheiro/cheque** → 10% de desconto  
  - À vista no **cartão** → 5% de desconto  
  - Em até **2x no cartão** → preço normal  
  - **3x ou mais no cartão** → 20% de juros

````python

def opcao_produto(preco, opcao):
    if opcao ==1:
        print("Pagamento a Vista!")
        valor = preco - 0.10 * preco
        return f"Pagamento a vista: {valor} com 10% de desconto"
    elif opcao ==2:
        print("A vista no cartão")
        valor = preco - 0.05 * preco
        return f"Pagamento final: {valor}"
    elif opcao ==3:
        print("Divindo o preco do produto em 2x")
        valor = preco//2
        return f"Valor divido em 2x: sai cada parcela a:{valor}"
    elif opcao ==4:
        print("Mais de 3x no cartão. Juros de 20%")
        num_parcelas = int(input("Digite o numero de parcelas"))
        while num_parcelas <3:
            print("Número de parcelas deve ser maior que 3")
            num_parcelas = int(input("Digite o numero de parcelas"))
        preco_final = preco + preco * 0.20
        parcela_individual = preco_final / num_parcelas
        for i in range(num_parcelas):
            preco_final -= parcela_individual
            print(f"Parcela {i+1}x: {preco_final:.2f} ")
        return f"Cada parcela após o juros ficará: {parcela_individual}"
````

````python
def analise_codigos(qtd_codigos):
    count_codigos_zem_zero  = 0
    maior_zero = float("-inf")
    for i in range(1, qtd_codigos+1):
        codigo = input(f"Digite o {i}º codigo: ").lower()
        codigos_zero = 0
        tem_zeros  =False
        for j in range(len(codigo)):
            if codigo[j] == "0":
                codigos_zero +=1
                tem_zeros = True
        if tem_zeros == False:
            count_codigos_zem_zero +=1
            print(f'Codigo: {codigo} Não tem zeros!')
        elif tem_zeros == True:
            print(f"Código: {codigo} tem zeros!")
        if j ==0 or codigos_zero > maior_zero:
            maior_zero = codigos_zero
    print(f"Quantidade de códigos digitados que NÃO possuem zero: {count_codigos_zem_zero}")
    if maior_zero < 0:
        print("Nenhum código que tenha zero foi digitado!")
    else:
        print(f'Maior quantidade de zeros digitada em um código: {maior_zero}')
            
````

````python

velocidade_maxima = int(input("Digite a velocidade máxima permitida na via: "))
while velocidade_maxima <0:
    print("Digite uma velocidade que seja MAIOR que zero!")
    velocidade_maxima = int(input("Digite a velocidade máxima permitida na via: "))
velocidade_registrada = int(input("Digite a velocidade  registrada na via!"))
while velocidade_registrada < 0:
    print("Digite uma velocidade que seja MAIOR que zero")
    velocidade_registrada = int(input("Digite a velocidade máxima permitida na via!"))

infracao_media = (0.2 * velocidade_maxima) + velocidade_maxima
infracao_gravissima = (0.5 * velocidade_maxima) + velocidade_maxima

if velocidade_registrada == velocidade_maxima or velocidade_registrada < velocidade_maxima:
    print("Sem Infração")
else:
    if velocidade_registrada <= infracao_media:
        print("Infração Média")
    if velocidade_registrada > infracao_media and velocidade_registrada <= infracao_gravissima:
        print("Infração grave")
    if velocidade_registrada > infracao_gravissima:
        print("Infração Gravíssima!")



````

````python

numero = int(input("Digite um número: "))
count = 0
maior = numero
qtd_numeros_multiplos3 = 0
count_negativos = 0

while numero !=0:
    if numero ==0:
        print("Programa encerrando!")
        break
    else:
        if numero <0:
            count_negativos +=1
        ### Ifs encadeados, por que se eu usar elif eu excluo uma condição: Exemplo: 
        ### 3 pode ser negativo e múltiplo de 3 ao mesmo tempo. Pode satisfazer mais de uma
        ### Condiçãp ao mesmo tempo
        if numero%3 ==0:
            qtd_numeros_multiplos3 +=1
        if  maior < numero:
            maior = numero
        count +=1
        numero = int(input("Digite um número: "))
if count > 0:
    print(f"Total de valores lidos {count}")
    print(f"Valores negativos {count_negativos}")
    print(f"O maior número é {maior}")
    print(f" quantidade de númores múltiplos de 3 são:{qtd_numeros_multiplos3} ")
else:
    print("Você digitou 0. Sem estatísticas válidas")
````
