#### **Exercício 1** Expedição de cargas

#### Uma transportadora utiliza caminhões idênticos para levar pacotes de um centro de distribuição até outra cidade. Todos os caminhões possuem a mesma capacidade máxima de carga.

#### Os pacotes chegam à área de embarque em uma ordem fixa e não podem ser reorganizados. Enquanto houver espaço, o próximo pacote é colocado no caminhão que está sendo carregado.

#### Quando o próximo pacote não cabe no caminhão atual, esse caminhão parte imediatamente e um novo caminhão começa a ser carregado com o pacote que ficou aguardando.

#### O processo continua até que todos os pacotes tenham sido processados. O último caminhão utilizado também deve ser contabilizado, mesmo que não esteja completamente cheio.

#### **Entrada**

#### A primeira entrada é um inteiro positivo C, representando a capacidade de cada caminhão.


#### Em seguida, são informados os pesos dos pacotes, um por vez. A sequência termina quando for informado o valor 0, que não representa um pacote.

#### Cada pacote possui peso maior que 0 e menor ou igual à capacidade do caminhão.

#### **Saída**:

-Exiba, em uma única linha e separados por espaços:

- a quantidade de caminhões utilizados;
- a soma de todo o espaço não utilizado nesses caminhões;
- a carga existente no último caminhão.

#### Exemplo

#### Entrada:

  - 10
  - 4
  - 3
  - 6
  - 2
  - 5
  - 0

#### Saída:

- 3 10 5

#### Nesse exemplo, o primeiro caminhão transporta 7 unidades de carga, o segundo transporta 8 e o terceiro transporta 5.

```python
def expedicao_cargas(capacidade_caminhao):
    peso_original_caminhao = capacidade_caminhao
    qtd_caminhoes = 0
    soma_espaco_nao_usado = 0
    carga_ultimo_caminhao = 0
    peso_pacote = int(input("Digite o peso do pacote: "))
    while peso_pacote < 0 or peso_pacote > capacidade_caminhao:
        print(f"O peso do pacote: {peso_pacote} deve ser maior que 0 e menor que a capcidade do caminhão!")
        peso_pacote = int(input("Digite o peso do pacote: "))
    while peso_pacote !=0:
        capacidade_caminhao -= peso_pacote
        qtd_caminhoes+=1
        if capacidade_caminhao !=0:
            soma_espaco_nao_usado+= capacidade_caminhao
        carga_ultimo_caminhao = capacidade_caminhao
        capacidade_caminhao = peso_original_caminhao
        peso_pacote = int(input("Digite o peso do pacote: "))
        while peso_pacote < 0 or peso_pacote > capacidade_caminhao:
            print(f"O peso do pacote: {peso_pacote} deve ser maior que 0 e menor que a capcidade do caminhão!")
            peso_pacote = int(input("Digite o peso do pacote: "))
    print(f"{qtd_caminhoes}, {soma_espaco_nao_usado}, {carga_ultimo_caminhao}")
```
