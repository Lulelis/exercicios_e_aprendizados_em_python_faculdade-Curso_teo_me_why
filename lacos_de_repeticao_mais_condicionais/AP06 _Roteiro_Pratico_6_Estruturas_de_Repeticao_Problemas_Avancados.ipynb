#### **Exercício 2** Comparando duas strings por índice

#### Leia duas palavras de mesmo tamanho.

#### Compare as duas palavras posição por posição, utilizando índices, e determine quantas posições possuem caracteres diferentes.

#### Ao final, mostre também quais posições são diferentes.

#### Exemplo:

- casa
- capa

#### As palavras diferem apenas na posição 2.

#### Importante: utilize texto[i]. Não percorra diretamente os caracteres com for letra in texto.

```python
def comparacao_por_indice(palavra_1, palavra_2):
    count_distintos = 0
    for i in range(len(palavra_1)):
        if palavra_1[i] != palavra_2[i]:
            count_distintos+=1
            print(f"Posição diferente:{palavra_1[i]}difere de: {palavra_2[i]}")
    print(f"Quantidade palavras distitnas: {count_distintos}") 
```

```python
palavra_1 = input("Digite a primeira palavra: ")
palavra_2 = input("Digite a segunda palavra: ")
while len(palavra_1) != len(palavra_2):
    print("Tamanhos distintos: ")
    palavra_1 = input("Digite a primeira palavra: ")
    palavra_2 = input("Digite a segunda palavra: ")
comparacao_por_indice(palavra_1, palavra_2)
```

#### **Exercício 3** Conversão manual de binário para decimal:

#### Leia um número binário como string, contendo apenas os caracteres 0 e 1.

#### Converta esse número para decimal sem utilizar funções prontas de conversão de base.

#### O programa deve percorrer a string por índice e construir o valor decimal progressivamente.

- *Dica: ao ler um novo dígito binário da esquerda para a direita, o valor acumulado pode ser atualizado por decimal = decimal * 2 + digito.
Exemplo: 1101 deve resultar em 13.*

```python
def verifica_entrada_se_binaria(entrada):
    numero_binario = 0
    num_binario = True
    while True:
        for i in range(len(entrada)):
            if entrada[i] !="0" and entrada[i] != "1":
                print(f"A entrada informada é diferente de de O ou 1: {entrada[i]}")
                num_binario = False
            else:
                num_binario = True
        if not num_binario:
            print("Digite a entrada novamente!")
            entrada = input("Digite o número binário: ").lower()
        elif num_binario:
            print("Entrada correta!")
            numero_binario = entrada
            break
    return numero_binario
```

```python
def conversor_binario(numero_binario):
    numero_binario = int(numero_binario)
    num_binario = 0
    count = 0
    while numero_binario !=0:
        resto = numero_binario%10
        num_binario += resto * (2 ** count)
        numero_binario = numero_binario//10
        count +=1
    return num_binario
```

```python
numero = input("Digite o número binário!").lower()
entrada_numero = verifica_entrada_se_binaria(numero)
print(f"O número informado: {entrada_numero}  convertido para binário: {conversor_binario(entrada_numero)}")
```

#### **Exercício 7 Maior bloco de caracteres iguais**

#### Leia uma string não vazia e determine qual é o maior bloco de caracteres iguais e consecutivos.

- Mostre:

  - o caractere que forma o maior bloco;
  - o tamanho desse bloco;
  - a posição em que o bloco começa.

#### Se houver empate, considere o primeiro bloco encontrado.

#### Exemplo: em aaabbccccdaa, o maior bloco é cccc, de tamanho 4, iniciando na posição 5.

```python
def maior_bloco(entrada):
    maior_bloco = 1
    atual_maior =  0
    posicao_maior = ""
    for i in range(len(entrada)-1):
        inicio_bloco = i
        if entrada[i] == entrada[i+1]:
            maior_bloco +=1
            if maior_bloco > atual_maior:
                atual_maior = maior_bloco
                posicao_maior = inicio_bloco
                
        else:
            maior_bloco = 1
    print(f"Maior bloco de caractere consecutivo: {atual_maior}\nPosição Maior Bloco: {posicao_maior}")
```

```python
entrada = input("Digite a sequência de caractere: ").lower()
maior_bloco(entrada)
```

#### **Exercício 8** Números primos e contagem de divisões

#### Leia um número inteiro positivo N.

#### Mostre todos os números primos entre 2 e N.

#### Para verificar se cada número é primo, teste possíveis divisores utilizando um laço interno.

#### O programa deve contar quantas operações de resto (%) foram realizadas durante toda a execução.

#### Assim que um divisor exato for encontrado, não é necessário continuar testando outros divisores daquele número.

#### Ao final, mostre:

- a quantidade de números primos encontrados;
- a quantidade total de operações de resto realizadas.

```python
def total_num_primos(numero):
    qtd_primos = 0
    qtd_operacoes = 0
    for i in range(1, numero+1):
        count_primos = 0
        for j in range(1,i+1):
            if i%j ==0:
                print(j)
                count_primos+=1
                qtd_operacoes+=1
        if count_primos ==2:
            qtd_primos+=1
    if qtd_primos ==0:
        print(f"Sem números primos!\nQuantidade total de Operações: {qtd_operacoes}")
    else:
        print(f"Quantidade de números primos entre o número informado ({numero}) quantidade: {qtd_primos}\nQuantidade Operacoes: {qtd_operacoes}")
```

```python
entrada = int(input("Digite um número: "))
total_num_primos(entrada)
```

```python
def verifica_se_tem_espaco(entrada):
    tem_espaco = False
    while True:
        for i in range(len(entrada)):
            if entrada[i] == " " or entrada[i] == "" or entrada[i] == "   ":
                tem_espaco=True
            else:
                tem_espaco = False
        if tem_espaco:
            print(f"A entrada digitata tem espaço!")
            entrada = input("Digite a entrada: ").lower()
        elif not tem_espaco:
            print("Não tem espaço!")
            break
    return entrada
    
```

```python
entrada_textual = input("Digite a entrada: ").lower()
entrada  = verifica_se_tem_espaco(entrada_textual)
count_caractere(entrada)
```
