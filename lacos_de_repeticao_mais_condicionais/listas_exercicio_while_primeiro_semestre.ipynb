#### **Exercício 5**: Quantidade de pares, ímpares, positivos e negativos

- Escreva um programa que leia números inteiros até que seja digitado 0

- Ao final, o programa deve mostrar:
  quantos números pares foram digitados
  quantos números ímpares foram digitados
  quantos números positivos foram digitados
  quantos números negativos foram digitados
- O zero apenas encerra a leitura.

```python
### Declaração das variáveis necessárias para entrada do programa

count_negativos = 0
count_positivos = 0
count_pares = 0
count_impares = 0
while True:
    entrada = int(input("Digite qualquer número real (positivo, negativo):"))
    if entrada ==0:
        break
    if entrada >0:
        count_positivos += 1  ### Como se quer a contagem, não a soma, não preciso declarar a variável igual ela mesma
    else:
        count_negativos +=1
    resto = entrada%2  ### Para números que são pares
    if resto ==0:
        count_pares +=1
    elif resto !=0:
        count_impares +=1

print(f"O resultado dos números digitados: números negativos {count_negativos}, números positivos: {count_positivos}, números pares: {count_pares}, números ímpares:{count_impares}")
```
