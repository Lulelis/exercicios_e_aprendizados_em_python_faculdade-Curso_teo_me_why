#### **Exercício 1** <u>Quadrados dos números de 1 até N</u>

#### Uma aplicação educativa deseja mostrar os quadrados dos primeiros números naturais.

#### Escreva um programa que leia um número inteiro positivo N e, para cada valor de 1 até N, exiba o número e seu quadrado.

#### Para N = 4, a saída deve seguir o formato:

- 1 - 1
- 2 - 4
- 3 - 9
- 4 - 16

```python
def mostra_num_quadrado(num):
    for i in range(1, num+1):
        print(f"{i} - {i**2}")
```

```python
num = int(input("Digite um número: "))
while num < 0:
    print("Digite um número inteiro positivo ")
    num = int(input("Digite um número: "))
mostra_num_quadrado(num)
```

#### **Exercício 2 Soma dos quadrados dos números pares**

#### Um sistema numérico precisa analisar os valores pares existentes entre 1 e um limite informado pelo usuário.

#### Escreva um programa que leia um inteiro positivo N e calcule:

- a quantidade de números pares entre 1 e N;
- a soma dos quadrados desses números pares.
- Para N = 6, os valores considerados são 2, 4 e 6 e a soma calculada é 2² + 4² + 6².

```python
def soma_quadrados(numero):
    count_pares = 0
    lista_pares = []
    soma_quadrados_pares = 0
    soma_quadrados_pares_lista = 0
    for i in range(1, numero+1):
        if i%2 ==0:
            count_pares +=1
            soma_quadrados_pares = soma_quadrados_pares + i **2
            lista_pares.append(i)
    if not lista_pares:
        print("Não tem números pares")
    else:
        for i in range(len(lista_pares)):
            print(f"Número: {lista_pares[i]}Posição {i} quadrado par: {lista_pares[i] **2}")
            soma_quadrados_pares_lista  += lista_pares[i] **2
        print(f'Fazendo com lista: {soma_quadrados_pares_lista}')
    if count_pares ==0:
        print("Número não tem par!")
    else:
        print(f"Quantidade de pares: {count_pares}\nSoma quadrado pares: {soma_quadrados_pares}")
```

```python
numero = int(input("Digite um número inteiro: "))
while numero < 0:
    print("Deve ser um número inteiro maior que zero!")
    numero = int(input("Digite um número inteiro: "))
soma_quadrados(numero)
```

#### **Exercício 3** <u>Cálculo de fatorial</u>

#### O fatorial de um número inteiro positivo N é o produto de todos os inteiros de 1 até N.

#### Por exemplo:

- 5! = 1 * 2 * 3 * 4 * 5 = 120

#### Escreva um programa que leia um inteiro N maior ou igual a 0 e calcule seu fatorial utilizando for.

- Considere que 0! = 1.

```python
def fatorial(num):
    if num ==0:
        return 1
    else:
        fatorial = 1
        for i in range(1, num+1):
            print(i)
            fatorial *=i
            print(i)
        return fatorial
```

```python
num = int(input("Digite o número para calcular o fatorial: "))
while num <0:
    print("Número inteiro positivo maior que 0")
    num = int(input("Digite o número para calcular o fatorial: "))
print(fatorial(num))
```

#### **Exercício 5** **Divisores e identificação de número primo**

#### Os divisores de um número inteiro positivo N são todos os valores que dividem N sem deixar resto.

#### Escreva um programa que leia um inteiro N maior que 1 e:

- exiba todos os divisores de N;
- conte quantos divisores foram encontrados;
- informe se N é primo.
- Um número é primo quando possui exatamente dois divisores positivos: 1 e ele mesmo

```python
def eh_primo(num):
    count_divisores = 0
    for i in range(1, num+1):
        if num %i  ==0:
            print(f"Número:{num} é divisível por: {i}")
            count_divisores +=1
        elif num%i !=0:
            continue
    if count_divisores ==2:
        print(f'Número digitado: {num} é primo!\nCount Divisores: {count_divisores}')
    else:
        print(f"{num} Não é primo")
```

```python
numero = int(input("Digite um núemro inteiro maior que 1: "))
eh_primo(numero)
```

#### Um sistema deseja analisar a composição de um texto informado pelo usuário.

#### Escreva um programa que leia uma string e, percorrendo seus caracteres, determine:

- quantas vogais existem;
- quantas consoantes existem;
- quantos dígitos numéricos existem;
- quantos espaços existem.

#### Considere como letras apenas os caracteres de A a Z e de a a z. Outros símbolos podem ser ignorados.

```python
entrada = "lucas"
count_vogais = 0
nova_palavra = ""
for i in entrada:
    print(i)
    if i =="a" or i=="e" or i =="i" or  i=="o" or i =="u":
        count_vogais +=1
    else:
        nova_palavra +=i

print(nova_palavra)
print(count_vogais)
```

```python
def analise_textual(entrada):
    count_vogais  =0
    count_consoantes = 0
    count_num = 0
    count_espacos = 0
    entrada = entrada.lower()
    for i in entrada:
        if i in ('aeiou'):
            count_vogais +=1
        elif i not in("aeiou") and i not in ("0123456789") and i not in (" ", ""):
            count_consoantes +=1
        if i in ('0123456789'):
            count_num+=1
        if i in (" ", ""):
            count_espacos +=1
    return (f'Quantidade Vogais: {count_vogais}\nQuantidade Consoantes: {count_consoantes}\nQuantidade espaços: {count_espacos}\nQuantidade Números: {count_num}')
```

```python
palavra = input("Digite a Palavra para se analisar ela textualmente: ")
print(analise_textual(palavra))
```

####  **Exercício 7 <u>Extração da parte numérica de um identificador**</u>

#### Um sistema recebe um identificador que pode conter letras, números, espaços e outros símbolos. Para uma etapa posterior do processamento, interessa apenas a parte numérica desse identificador.

#### Escreva um programa que leia o identificador como texto e construa uma nova string contendo somente os dígitos, preservando a ordem em que eles aparecem.

#### Ao final, exiba a parte numérica encontrada e a quantidade de dígitos utilizados.

#### - Exemplo:

- Entrada: AB-12 C9
- Parte numérica: 129

```python
def extrai_numero(entrada):
    parte_num  = ""
    qtd_num = 0
    for i in range(len(entrada)):
        if entrada[i] in ('0123456789'):
            parte_num += entrada[i]
            qtd_num+=1
    if qtd_num ==0:
        print("Entrada não tem número!")
    else:
        print(f"Parte Numérica: {parte_num}\nQuantidade Numérica: {qtd_num}")
```

```python
entrada = input("Digite a entrada para se verificar se tem número ou não: ")
extrai_numero(entrada)
```

#### Uma disciplina registrará as notas de uma atividade para uma quantidade conhecida de estudantes. As notas válidas estão no intervalo de 0 a 100.

#### Escreva um programa que leia a quantidade de estudantes. Para cada estudante:

- solicite a nota;
- se a nota estiver fora do intervalo de 0 a 100, continue solicitando até que um valor válido seja informado.

#### Ao final, apresente:

- a média da turma;
- a maior nota;
- a menor nota;
- quantos estudantes obtiveram nota maior ou igual a 60.

```python
def notas_validacao(n_estudantes):
    maior_nota = float("-inf")
    menor_nota = float("inf")
    soma_notas = 0
    count_60_acima = 0
    count_notas = 0
    for i in range(n_estudantes):
        notas = float(input(f"Digite a {i+1}ª nota: "))
        while notas < 0 or notas > 100:
            print("Digite uma nota dentro do intervalo válido!")
            notas = float(input(f"Digite a {i+1}ª nota: "))
        if notas >=60:
            print(f"A nota informada: {notas} é maior que 60")
            count_60_acima+=1
        if i ==1 or notas > maior_nota:
            maior_nota = notas
        if i ==1 or notas < menor_nota:
            menor_nota = notas
        soma_notas+=notas
        count_notas += 1
    media_notas = soma_notas/count_notas
    if count_notas ==0:
        print(f"Sem notas para mostrar :{count_notas}")
    else:
        print(f"Média de notas para o total de {count_notas}: {n_estudantes}:{media_notas:.2f}\nNúmero de notas acima de 60: {count_60_acima}\nMaior nota registrada: {maior_nota}\nMenor nota registrada: {menor_nota}")

            
            
```
