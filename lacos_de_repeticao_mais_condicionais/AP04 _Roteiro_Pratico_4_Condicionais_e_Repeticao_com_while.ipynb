### **Exercício 3** <u>Numeração dos pontos de inspeção</u>

### Uma equipe de inspeção numera sequencialmente os pontos verificados durante uma vistoria. O responsável informa quantos pontos existem no local.

#### Escreva um programa que leia um número inteiro positivo N e exiba, um por linha, os números de 1 até N.

#### Por exemplo, para N = 5, a saída deve ser

- 1
- 2
- 3
- 4
- 5

```python
def numeracao_inspecao(num):
    count= 1
    while count <=num:
        print(count)
        count +=1
```

```python
numero = int(input("Informe um número para se realizar a contagem: "))
numeracao_inspecao(numero)
```


#### **Exercício 4** <u>Soma e contagem de múltiplos de 3</u>

### Um sistema precisa analisar os identificadores numéricos de 1 até um limite informado pelo usuário. Para uma etapa do processamento, interessam apenas os valores divisíveis por 3.

#### Escreva um programa que leia um inteiro positivo N e percorra todos os valores de 1 até N. Ao final, apresente:

- a soma de todos os múltiplos de 3 encontrados;
- a quantidade de múltiplos de 3 encontrados.
- Para N = 10, os valores considerados são 3, 6 e 9.

```python
def funcao_soma_contagem_multiplos_de_3(num):
    soma_multiplos_3 = 0
    qtd_multiplos_3 = 0
    count = 1
    lista_multiplos_3 = []
    while count <=num:
        print(count)
        if count%3 ==0:
            soma_multiplos_3 +=count
            qtd_multiplos_3 +=1
            lista_multiplos_3.append(count)
        count +=1
    for i in range(len(lista_multiplos_3)):
        print(f'{i+1}º Termo: {lista_multiplos_3[i]}\nQuantidade: {len(lista_multiplos_3)}')
    print(f'Soma múltiplos de 3:{soma_multiplos_3}\nQuantidade Múltiplos 3:{qtd_multiplos_3}')
```

```python
num = int(input("Informe o número para ver os múltiplos até ele"))
```

```python
funcao_soma_contagem_multiplos_de_3(num)
```

#### O sensor do reservatório utilizado no Exercício 1 pode apresentar leituras inválidas por falha momentânea. Uma leitura válida deve estar entre 0 e 100.

#### Escreva um programa que solicite o percentual do reservatório e continue pedindo um novo valor enquanto a leitura estiver fora do intervalo válido.

#### Quando uma leitura válida for finalmente informada, o programa deve classificá-la como:

- abaixo de 25%: Nível baixo;
- de 25% até 80%: Nível adequado;
- acima de 80%: Nível alto

```python
def classificacao_reservatorio(entrada):
    if entrada < 25:
        print("Nível Abaixo!")
    elif entrada >= 25 and entrada <=80:
        print("Nível Adequado!")
    else:
        print("Nível Alto!")
```

```python
parada = input("Deseja iniciar o programa?").lower()
while parada not in ("nao", "n", "not", "no", "não"):
    nivel_agua = float(input("Digite o nível de água do reservatório: "))
    while nivel_agua <0 or nivel_agua >100:
        print("Digite um nível de água válido!")
        nivel_agua = float(input("Digite o nível de água do reservatório: "))
    classificacao_reservatorio(nivel_agua)
    parada = input("Deseja continuar o programa?").lower()
```

#### O valor -1 é apenas um sentinela e não deve participar dos cálculos.

#### Ao final, o programa deve apresentar:

- a quantidade de medições registradas;
- a velocidade média;
- a maior velocidade registrada;
- a menor velocidade registrada;
- quantas medições ficaram acima de 60 km/h.

#### O programa também deve funcionar caso o operador digite -1 logo na primeira leitura.

```python
def monitoramento_vendo():
    qtd_medicoes = 0
    soma_velocidade_vendo = 0
    maior_velocidade = float("-inf")
    menor_velocidade = float("inf")
    acima_60 = 0
    medicao = int(input("Digite a velocidade do vendo: "))
    while medicao !=-1:
        qtd_medicoes +=1
        soma_velocidade_vendo += medicao
        if medicao > maior_velocidade:
            maior_velocidade = medicao
        if medicao < menor_velocidade:
            menor_velocidade = medicao
        if medicao > 60.00:
            acima_60 +=1
        medicao = int(input("Digite a velocidade do vendo: OIu -1 para encerrar!"))
    if qtd_medicoes ==0:
        print("Não foram digitadas nenhumas medicões")
    else:
        print(f'Quantidade de medições: {qtd_medicoes}\nVelocidade Média: {soma_velocidade_vendo/qtd_medicoes:.2f}km\nMaior velocidade digitada: {maior_velocidade}\nMenor velocidade Digitada: {menor_velocidade}\nMedições maiores que 60: {acima_60}')
    
```


### **Desafio Missão de inspeção de um robô**

#### Um robô autônomo realiza uma missão de inspeção em um túnel. Antes de iniciar, o sistema recebe a quantidade inicial de energia da bateria, em unidades de energia.

#### Durante a missão, cada trecho do túnel exige uma determinada quantidade de energia. Os custos dos trechos são informados um por vez. O valor 0 indica que não existem mais trechos e que a rota foi concluída.

#### Um trecho só pode ser percorrido se a bateria atual possuir energia suficiente para pagar todo o seu custo. Se o custo informado for maior que a energia restante, o robô não percorre esse trecho e a missão é encerrada imediatamente por falta de energia.

#### Para cada trecho percorrido, o programa deve descontar seu custo da bateria e manter as informações necessárias para, ao final, apresentar:


- a quantidade de trechos concluídos;
- o total de energia consumida;
- a energia restante na bateria;
- o maior custo entre os trechos efetivamente percorridos;
- quantos trechos terminaram com a bateria em 20% ou menos da energia inicial;
- o motivo do encerramento: Rota concluída ou Energia insuficiente.

#### Considere que a energia inicial é positiva e que todos os custos informados são valores inteiros não negativos.

#### Exemplo: para energia inicial igual a 100 e custos 20, 35, 30 e 25, o robô percorre os três primeiros trechos. Depois deles, restam 15 unidades de energia, portanto o trecho de custo 25 não pode ser percorrido.

```python
def missao_robo():
    print("Missão Inspeção Robô!")
    qtd_inicial_energia = int(input("Digite a quantidade inicial de energia da bateria!"))
    qtd_trechos_concluidos = 0
    total_enegia_consumida = 0
    maior_custo = float("-inf")
    qtd_trechos_bateria_20 = 0
    custo_20_energia_inicial = qtd_inicial_energia * 20/100
    energia_consumida = int(input("Informe a energia consumida:  "))
    while qtd_inicial_energia!= 0 and energia_consumida !=0:
        if energia_consumida >  qtd_inicial_energia:
            print(f' Energia Insuficiente para ser consumida!\nEstá faltando:{energia_consumida - qtd_inicial_energia * (-1)}')   
        elif energia_consumida == qtd_inicial_energia:
            print(f"Rota Concluída Com Sucesso!\nBateria em: {qtd_inicial_energia}") 
        if qtd_inicial_energia <= custo_20_energia_inicial:
            qtd_trechos_bateria_20 +=1
        if energia_consumida > maior_custo:
            maior_custo = energia_consumida
        qtd_trechos_concluidos +=1
        total_enegia_consumida +=energia_consumida
        qtd_inicial_energia -= energia_consumida
        energia_consumida = int(input("Informe a energia consumida:  "))
    if qtd_trechos_concluidos ==0:
        print("Trecho encerrado! ou não houve rota para serem concluída!")
    else:
        print(f'Quantiade de Trechos Concluídos: {qtd_trechos_concluidos}\nEnergia Consumida : {total_enegia_consumida}\nEnergia Restante: {qtd_inicial_energia}\n Maior Custo: {maior_custo}\nQuantidade de Trechos com 20% da energia inicial: {qtd_trechos_bateria_20}')
```

```python
print(missao_robo())
```
