#### ***061 — Progressão Aritmética (PA)*** ***Com While***

#### Desenvolva um programa que leia o **primeiro termo** e a **razão** de uma **PA**.  
No final, mostre os **10 primeiros termos** dessa progressão.

```python
def progressao_aritmetica(primeiro_termo, razao):
    count = int(input("Quantos temos serão lidos? "))
    termo = primeiro_termo
    while count > 1:
        count -=1
        termo += razao
        print(f"Termo {count} --> {termo}")
```

#### **Exercício: 65**

#### Média, maior e menor valores

#### Crie um programa que leia **vários números inteiros** pelo teclado.  
#### No final da execução, mostre a **média entre todos** os valores e qual foi o **maior** e o **menor** valores lidos.  
#### O programa deve perguntar ao usuário se ele quer ou não **continuar** a digitar valores.


```python



def analisa_valores(lista):
    maior = float("-inf")
    menor = float("inf")
    soma_valores = 0
    qtd = 0
    for valor in range(len(lista)):
        soma_valores += lista[valor]
        qtd +=1
        if lista[valor] > maior:
            maior = lista[valor]
        if lista[valor] < menor:
            menor = lista[valor]
    media = soma_valores/qtd
    return media, maior, menor

    
```

#### No final, mostre **quantos números** foram digitados e qual foi a **soma** entre eles  
*(desconsiderando o flag)*.


```python
def funcao_soma_com_parada():
    soma = 0
    count = 0
    while True:
        num = int(input("Digite um número: ou digute 999 para parar:  "))
        if num ==999:
            break
        else:
            count+=1
            soma+=num
    return(f"Soma números digitados: {soma}\n Foram digitados: {count}")
```

#### Um professor deseja calcular a média das notas de uma atividade. Como a quantidade de alunos pode variar, o programa deve continuar lendo notas até que seja digitado o valor -1.

#### Escreva um programa que leia as notas dos alunos. Quando o valor -1 for informado, a leitura deve terminar.

#### Ao final, o programa deve exibir a quantidade de notas válidas e a média da turma. Se nenhuma nota válida for informada, o programa deve exibir “Nenhuma nota registrada”.

```python
def notas_com_while():
    count_notas_validas = 0
    soma_notas = 0
    notas_aluno = float(input("Informe a nota do Aluno: "))
    while notas_aluno !=-1.00:
        soma_notas += notas_aluno
        count_notas_validas +=1
        notas_aluno = float(input("Informe a nota do Aluno novamente ou -1 para encerrar! "))
    if count_notas_validas ==0:
        return ("Sem notas válidas digitadas!")
    else:
        media_notas = soma_notas/count_notas_validas
        return (f"Média Notas Alunos: {media_notas:.2f}\nQuantidade Notas Válidas:{count_notas_validas}")
```

#### Uma estufa automatizada registra leituras de temperatura ao longo do dia. Cada leitura é informada em graus Celsius. Quando o operador deseja encerrar o registro, ele digita -1.
#### Crie um programa que leia as temperaturas até que -1 seja digitado. Ao final, o programa deve informar: quantidade de leituras válidas, média das temperaturas, maior temperatura registrada e quantidade de leituras em estado de alerta.

#### Considere estado de alerta quando a temperatura for maior ou igual a 35°C. O valor -1 não deve ser contado como leitura válida.

```python
def registra_temperatura():
    qtd_temp_estado_alerta = 0
    qtd_leitura_validas = 0
    maior_temp = float("-inf")
    soma_temp = 0
    temperatura = float(input("Digite a temperatura: "))
    while temperatura != -1.00:
        qtd_leitura_validas +=1
        if temperatura >= 35.00:
            qtd_temp_estado_alerta +=1
        if temperatura > maior_temp:
            maior_temp = temperatura
        soma_temp +=temperatura
        temperatura = float(input("Digite a temperatura:ou -1 para encerrar! "))
    if qtd_leitura_validas ==0:
        print("Nenhum dado válido foi digitado, portanto, sem estatísticas!")
    else:
        print(f"Média Temperatura em Cº:{soma_temp/qtd_leitura_validas:.2f}\nQuantidade de Leituras Válidas: {qtd_leitura_validas}\nMaior Temperatura: {maior_temp}Cº\nQuantidade temperaturas acimad e 35ºC:{qtd_temp_estado_alerta}")
        
```

```python
def organizacao_livros(turnos= 1):
    meta = 0
    qtd_turnos = 0
    qtd_livros_repostos = 0
    meta_alcancada = True
    while meta < 120 and turnos <=5:
        qtd_turnos +=1
        livros = float(input(f"Digite a quantidade de livros repostos\nTurno:{turnos}"))
        meta +=livros
        qtd_livros_repostos += livros
        turnos +=1
    if qtd_livros_repostos ==0:
        return f'Quantidade reposta é zero!\nMeta Não alcançada: {meta}'    
    elif meta >=120:
        return f'Meta Alcançada!\nQuantidade Livros Repostos:{qtd_livros_repostos}\nMeta: {meta}\nTurnos Utilizados: {qtd_turnos}'
    else:
        return f'Meta Não Atingida!:{meta}\nTotal Livros repostos: {qtd_livros_repostos}\nQuantidade Turnos: {qtd_turnos}'
```
