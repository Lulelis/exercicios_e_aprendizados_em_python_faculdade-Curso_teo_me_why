#### 2. Faça um programa que receba 4 alturas usando um laço de repetição e realize a soma dessas alturas

```python
qtd_entradas = 1
soma_alturas = 0 ### guardando o valor da soma == 0 dentro da variável
while qtd_entradas <=4:
    alturas = float(input("Digite a primeira altura:"))
    soma_alturas += alturas
    qtd_entradas +=1
print("As alturas somadas são: ",soma_alturas)
```

```python
### Também:
soma = 0
qtd_entradas = 4
while qtd_entradas > 0:
    altura = float(input("Entre com a altura: "))
    soma += altura
    qtd_entradas-=1
print("A quantidde de alturas é : ", soma)
```

#### 3: Faça um programa que receba uma quantidade indefinida de valores correspondentes a "saldo em conta", mas, quando o usuário apertar "enter" sem digitar valor algum, o programa para de receber valores, e exibe a soma de todos os valores digitados anteriormente

```python
####  Primeiro capturar a entrad COMO string
saldo_em_conta = 0

while True: ### Laço infinito condição NUNCA vai mudar, pois é TRUE
    valor_entrada = input("Digite o valor presente na conta corrente: ")
    if valor_entrada == "": ### Verificando se o valor_entrada é uma string vazia
        break               ### se for, quebra o Loop ## Só pode existir dentro de Laços de Repetição 
    valor = float(valor_entrada) ### convertendo o valor para float
    saldo_em_conta += valor
print(f"O saldo na conta corrente é:{saldo_em_conta}")



```

## ***For***

> Percorrer Elemento, sequência, estruturas de dados no Python que são iteráveis - percorrendo de elemento a elemento

--> `Utilizado normalmente para percorrer sequências, estrutura de dados que são iteráveis, ou seja pode-se passar de elemento a elemento`

```python
nome = "LuElias"
for letra in nome: ### letra: variável temporária ### Nome variável declarada fora do for fica trocando de valor a cada rodada....Passando pelos
                   ###  i de índice: Elementos
    print(letra)
```

***Lógica:*** *Uma variável temporária declarada dentro do for que percorrerar outra variável de interesse*

> Percorre os elementos de um objeto

--> O próprio for já realiza essa troca

```python
nome = "LuElias"
for i in nome: ### letra: variável temporária ### Nome variável declarada fora do for
    print(i)
```

#### Escrea um programa que receba uma lista de números do usuário e conte quantas vezes um número específico aparece na lista. Solicite ao usuário um número e exiba a contagem

#### For vai percorrer cada elemento com a variável temporária no range ou na lista


```python

lista_numeros = [1,2,3,4,5,6,7,3,2,4,6,7,5,3,3,5,5,6]
count = 0
entrada = int(input("Qual número você deseja testa se aparece na lista?"))


#### For vai percorrer cada elemento com a variável temporária no range ou na lista

for i in lista_numeros:
    if i  == entrada:
        count +=1
    else:
        print(f"Número {entrada} não está na lista")
print("O número aparece: ", count)
```
