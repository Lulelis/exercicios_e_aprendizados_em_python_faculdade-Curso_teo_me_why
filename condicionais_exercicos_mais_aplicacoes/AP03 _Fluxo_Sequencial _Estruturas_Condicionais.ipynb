```python
### Exercício 2 & 5:

def resultado_divisao():
    num = int(input("Digite um número:"))
    divisao_inteira = num//2
    resto_divisao = num%2
    print(f"Divisão inteira (2): {divisao_inteira}\nResto Divisão (2):{resto_divisao}")
    if num > 0 and resto_divisao ==0:
        print(f"Número: {num} é Positivo e Par")
    elif num > 0 and resto_divisao !=0:
        print(f'Número: {num} é Positivo e ímpar!')
    elif num < 0 and resto_divisao  ==0:
        print(f'Número: {num} é Negativo e Par')
    elif num <0 and resto_divisao !=0:
        print(f'Número: {num} é Negativo e Ímpar')
    elif num ==0:
        print("Zero")
```

### 🎟️ Exercício 6 - Sistema de descontos para ingressos

#### Retome o cálculo do **_Exercício 3_**.  
Desta vez, o usuário **_não informará o percentual_**: o programa deverá escolhê-lo de acordo com a **_idade_**.


#### Leia:
- A **_idade_**

#### Regras de desconto:
- **_Menor que 12 anos_** → desconto de 50%  
- **_12 a 17 anos_** → desconto de 30%  
- **_18 a 59 anos_** → sem desconto  
- **_60 anos ou mais_** → desconto de 40%  


#### Apresente:
- **_Percentual de desconto_**  
- **_Valor do desconto_**

### 🧪 Testes sugeridos

```python
### Exercício 3 e 6:

def desconto_ingresso():
    valor_ingresso = float(input("Qual o valor do ingresso? "))
    pct_desconto = float(input("Qual o percentual de Desconto? (%)"))
    valor_final = valor_ingresso - (valor_ingresso * pct_desconto/100)
    print(f"Valor 1 com Desconto: {valor_final}")
    ### Continuando:

    idade = int(input("Qual a idade? "))
    if idade <12:
        valor_final = valor_ingresso - (valor_ingresso * 0.50)
        print(f"Valor final com 50% De Desconto:{valor_final:.2f}")
    elif idade >=12 and idade <=17:
        valor_final = valor_ingresso - (valor_ingresso * 0.30)
        print(f'O valor final com 30% de Ingresso: {valor_final:.2f}')
    elif idade > 17 and idade <=59:
        print(f"Sem desconto!Valor compra: {valor_ingresso}")
    else:
        valor_final = valor_ingresso - (valor_ingresso *0.40)
        print(f"Valor final com 40% de Desconto: {valor_final:.2f}")
```

#### 📦 Exercício 8 - Cálculo de frete de uma loja virtual

#### Leia o **_valor da compra_**, a **_distância em km_** e o **_tipo de entrega_**:  
**S** para **_expressa_** ou **N** para **_normal_**.

#### 🚚 Frete pela distância
#### • **_Até 5 km_**: R$ 8,00  
### • **_Mais de 5 km e até 15 km_**: R$ 15,00  
### • **_Acima de 15 km_**: R$ 25,00  

#### ⚙️ Regras adicionais

#### • Compras de **_R$ 200,00 ou mais_** recebem **_frete grátis_** somente na entrega **_normal_**  
### • Entrega **_expressa_** acrescenta **_R$ 12,00_** ao frete-base  
### • **_Frete grátis não se aplica à entrega expressa_**

#### 📤 Saída esperada

#### O programa deve apresentar:
- **_Valor da compra_**  
- **_Frete_**  
- **_Total do pedido_**


```python
def compra_calculo(valor_compra, distancia, tipo_entrega):
    valor_frete = 0
    if tipo_entrega =="S":
        valor_frete =12
        if distancia <=5:
            valor_frete +=8
            return f'Valor Compra:{valor_compra}\nValor Frete: {valor_frete}Valor Compra Total: {valor_compra + valor_frete}'
        elif distancia >5 and distancia <=15:
            valor_frete +=15
            return f'Valor Compra :{valor_compra}\nValor Frete: {valor_frete}\nValor Compra Total: {valor_compra + valor_frete}'
        elif distancia > 15:
            valor_frete +=25
            return f'Valor Compra :{valor_compra}\nValor Frete: {valor_frete}\nValor Compra Total: {valor_compra + valor_frete}'
    else:
        if valor_compra >=200:
            return f'Valor Compra:{valor_compra}\nFrete Grátis'
        elif distancia <=5:
            valor_frete = 8
            return f'Valor Compra: {valor_compra}\nValor Frete: {valor_frete}\nTotal Compra: {valor_compra + valor_frete}'
        elif distancia > 5 and distancia <=15:
            valor_frete = 15
            return f'Valor Compra: {valor_compra}\nValor Frete: {valor_frete}\nTotal Compra: {valor_compra + valor_frete}'
        elif distancia > 15:
            valor_frete = 25
            return f'Valor Compra: {valor_compra}\nValor Frete: {valor_frete}\nTotal Compra: {valor_compra + valor_frete}'
```

```python
valor_compra = float(input("Informe o valor da Compra: "))
distancia = float(input("Informe a distância de entrega:  "))
tipo_entrega = float(input("Informe o tipo de entrega: ")).upper()
while tipo_entrega not in ("S", "N"):
    print("Digite sim ou não!")
    tipo_entrega = input("Qual o tipo de Entrega? S/N").upper()
```

```python
print(compra_calculo(valor_compra, distancia, tipo_entrega))
```

### ⚡ Exercício 9 - Conta de energia com tarifa progressiva

#### Leia o **_consumo mensal em kWh_** e se o cliente participa da **_tarifa social_**: **S** ou **N**.

---

#### 📊 Faixas de consumo

#### • **_Primeiros 100 kWh_**: R$ 0,65/kWh  
#### • **_101 a 200 kWh_**: R$ 0,80/kWh (apenas sobre a parte acima de 100)  
#### • **_Acima de 200 kWh_**: R$ 0,95/kWh (apenas sobre a parte acima de 200)  

---

#### 💰 Regras adicionais

#### • **_Tarifa social_**: 15% de desconto se o consumo for de até **_200 kWh_**  
#### • **_Consumo acima de 300 kWh_**: taxa adicional de **_8%_**

- **_Consumo_**  
- **_Valor da energia_**

- **_Taxa adicional_**

#### 🧪 Testes sugeridos

- **_100 kWh_**  
- **_101 kWh_**

- **_201 kWh_**  
- **_300 kWh_**  
- **_301 kWh_**


```python
def consumo_kwh(entrada, tarifa_social):
    if tarifa_social =='S':
        if entrada <=100:
            pagamento = 100 * 0.65
            return f'{pagamento:.2f}'
        elif entrada  >100 and entrada <=200:
            restante = entrada - 100
            pagamento = 100* 0.65 + restante * 0.80
            pagamento_com_desconto = pagamento - (pagamento * 15/100)
            return f'Pagamento com Desconto: {pagamento_com_desconto}\nDesconto de: {pagamento * 15/100}'
        elif entrada > 200 and entrada < 300:
            pagamento = 100 * 0.65 + 100 * 0.80 + (entrada- 200) * 0.95
            return f'{pagamento:.2f}'
        elif entrada >= 300:
            pagamento = 100 * 0.65 + 100 * 0.80 + (entrada- 200) * 0.95 
            pagamento_com_taxa = pagamento + (pagamento *  8/100)
            return f'Pagamento com Taxa:{pagamento_com_taxa}'
    else:
        if entrada <=100:
            pagamento = 100 * 0.65
            return f'{pagamento:.2f}'
        elif entrada  >100 and entrada <=200:
            restante = entrada - 100
            pagamento = 100* 0.65 + restante * 0.80
            return f'Pagamento sem Desconto: {pagamento}'
        elif entrada >200 and entrada < 300:
            pagamento = 100 * 0.65 + 100 * 0.80 + (entrada- 200) * 0.95
            return f'{pagamento:.2f}'
        elif entrada >= 300:
            pagamento = 100 * 0.65 + 100 * 0.80 + (entrada- 200) * 0.95 
            pagamento_com_taxa = pagamento + (pagamento *  8/100)
            return f'Pagamento com Taxa:{pagamento_com_taxa}'
        


        


  
```

```python
consumo_mensal = float(input("Qual o consumo mensal? "))
tarifa_social = input("Tem tarifa social? ").upper()
while tarifa_social not in ("S", "N"):
    print("Escreva apenas sim S ou não N !")
    tarifa_social = input("Tem tarifa social? ").upper()

```

```python
print(consumo_kwh(consumo_mensal, tarifa_social))
```

#### 🚁 <u>Desafio Final da Corrida de Drones</u>

#### Durante uma competição de robótica, dois drones autônomos, **_Drone A_** e **_Drone B_**, disputam a etapa final de uma corrida.

#### Cada drone deve completar um percurso dividido em **_três trechos_**.  

- O tempo gasto em cada trecho é registrado em **_segundos_**.  
- Durante o percurso, o drone também pode receber **_penalidades_** por tocar em obstáculos.

---

## 📥 Entrada de Dados

### Para cada drone, o programa deve ler:
- **_Tempo do trecho 1_**  
- **_Tempo do trecho 2_**  
- **_Tempo do trecho 3_**  
- **_Quantidade de penalidades recebidas_**

### Cada penalidade acrescenta **_5 segundos_** ao tempo total do drone.

---

## 🧮 Cálculo do Tempo Final

### Fórmula:

- <u>tempo_final = trecho_1 + trecho_2 + trecho_3 + penalidades * 5 </u>
---

#### 🏆 Critérios para determinar o vencedor

### 1. Vence o drone com o **_menor tempo final_**  
### 2. Se os tempos finais forem iguais, vence o drone com **_menos penalidades_**  
### 3. Se ainda houver empate, vence o drone com o **_menor tempo no terceiro trecho_**  
### 4. Se todos os valores forem iguais → **_EMPATE_**

---

#### 📊 Classificação Geral da Prova

#### Considere o **_melhor tempo final_** entre os dois drones:
- **_Menos de 60 segundos_** → EXCELENTE  
- **_De 60 até 89 segundos_** → BOA  
- **_De 90 até 119 segundos_** → REGULAR  
- **_120 segundos ou mais_** → LENTA  

---

#### 📤 Saída do Programa

#### O programa deve apresentar:
- **_Tempo final do Drone A_**  
- **_Tempo final do Drone B_**  
- **_Resultado da corrida_**: DRONE A, DRONE B ou EMPATE  
- **_Classificação geral da prova_**

---

#### 📌 Exemplo

#### Drone A  
- Trecho 1: 20  
- Trecho 2: 18  
- Trecho 3: 19  
- Penalidades: 1  

#### Drone B  
- Trecho 1: 19  
- Trecho 2: 20  
- Trecho 3: 18  
- Penalidades: 1  

#### Resultado:
- Tempo final A: **_62_**  
- Tempo final B: **_62_**  
- Vencedor: **_DRONE B_**  
- Classificação da prova: **_BOA_**

---

#### ⚠️ Atenção

#### Os critérios de desempate devem ser avaliados **_na ordem apresentada_**.  

-Um critério posterior só deve ser considerado quando o anterior resultar em **_empate_**.


```python
def analisa_performance_drones(drone_a, drone_b, qtd_penalidades_a, qtd_penalidades_b):
    tempo_final_a = sum(drone_a) + qtd_penalidades_a * 5
    tempo_final_b = sum(drone_b) + qtd_penalidades_b * 5
    print(f"Tempo Final A:{tempo_final_a}\nTempo Final B:{tempo_final_b}")
    if tempo_final_a < tempo_final_b:
        print("Drone A possui menor tempo final, dessa forma ele venceu!")
    elif tempo_final_b < tempo_final_a:
        print("Drone B possui menor tempo final, dessa forma ele venceu!")
    elif tempo_final_a == tempo_final_b:
        if qtd_penalidades_a < qtd_penalidades_b:
            print(f"Drone A ganhou por apresentar menor quantidade de penalidades que o B!:{qtd_penalidades_a}")
        elif qtd_penalidades_b < qtd_penalidades_a:
            print(f'O Drone B ganhou por apresentar menor quantidade de penalidades que o A!:{qtd_penalidades_b}')
        elif qtd_penalidades_a == qtd_penalidades_b:
            if drone_a[2] < drone_b[2]:
                print("Empates de Penalides!Análise do menor tempo do 3º Trecho dos Drones, sendo o do drone a menor!")
            elif drone_b[2] < drone_a[2]:
                print("Empates de Penalides!Análise do menor tempo do 3º Trecho dos Drones, sendo o do drone b o menor!")
            elif drone_a[2] == drone_b[2]:
                print("Empate técnico em todas as categorias analisadas!")

    ### Classificaçãp menor tempo final
    menor_tempo_a = float("inf")
    for i in range(len(drone_a)):
        if drone_a[i] < menor_tempo_a:
            menor_tempo_a = drone_a[i]
    menor_tempo_b = float("inf")
    for i in range(len(drone_b)):
        if drone_b[i] < menor_tempo_b:
            menor_tempo_b = drone_b[i]
    ### Menor tempo entre os dois:
    menor_tempo_final = 0
    if tempo_final_a < tempo_final_b:
        menor_tempo_final = tempo_final_a
    elif tempo_final_b < tempo_final_a:
        menor_tempo_final = tempo_final_b
    elif tempo_final_a == tempo_final_b:
        menor_tempo_final = tempo_final_a
    if menor_tempo_final < 60:
        print("Excelente!")
    elif menor_tempo_final >=60 and menor_tempo_final <=89:
        print("Boa!")
    elif menor_tempo_final >=90 and menor_tempo_final <=119:
        print("Regular!")
    else:
        print("Lenta!")    


        
    
```

```python
### Entrada de dados:

drone_a = []
drone_b = []
tempo_trecho_1_a = int(input("Qual o tempo do trecho 1 do drone a?"))
drone_a.append(tempo_trecho_1_a)
tempo_trecho_2_a = int(input("Qual o tempo do trecho 2 do drone a?"))
drone_a.append(tempo_trecho_2_a)
tempo_trecho_3_a = int(input("Qual o tempo do trecho 3 do drone a?"))
drone_a.append(tempo_trecho_3_a)

for i in range(3):
    tempo_trecho = int(input(f"Digite o tempo do trecho {i+1} do Drone B: "))
    drone_b.append(tempo_trecho)
qtd_penalidades_a, qtd_penalidades_b = map(int,input("Digite as respectivas quantidades de penalidades: ").split())
```

```python
analisa_performance_drones(drone_a, drone_b,qtd_penalidades_a,qtd_penalidades_b )
```
