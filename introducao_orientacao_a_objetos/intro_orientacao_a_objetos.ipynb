#### Criando a calsse Ponto



```python
class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

- **self.x**: atributo x do objeto

- **self.y:** atributo y do objeto

#### **Adicionando métodos ao Ponto**:

```python
class Ponto:
    def __init__(self, x,y):
        self.x = x
        self.y = y

    def exibir(self):
        print("Ponto: ", self.x, self.y)

    def mover(self, dx, dy):
        self.x +=dx
        self.y +=dy
```

```python
p1 = Ponto(5,7)
p2 = Ponto(10,11)
```

```python
p1.exibir()
```

```python
p2.exibir()
```

```python
p1.mover(1,2)
```

```python
p2.mover(10,11)
```

```python
p1.exibir()
```

```python
p2.exibir()
```

#### **Entendendo o self:**

#### Ideia principal:


- self representa o próprio objeto:

- se chamamos p1.mover(), self é p1;

- se chamamos p2.mover(), self é p2

- def mover (self, dx, dy):
    - self.x += dx
    - self.y +=dy

#### 4. Playlist com músicas e usuário: Crie as classes Usuario, Musica e Playlist. A playlist deve usar as três formas de relacionamento estudadas: guardar um objeto Usuario, guardar uma lista de objetos Musica e possuir um método adicionar_musica(musica) que recebe uma música como parâmetro. Ao final, crie um método exibir_playlist() mostrando o dono da playlist e todas as músicas adicionadas.

```python
class usuario:
    def __init__(self, nome):
        self.nome = nome
    
    def exibir_usuario(self):
        print(f"O nome do usuário é {self.nome}")
```

```python
class musica:
    def __init__(self, nome_musica, artista, duracao):
        self.nome_musica = nome_musica
        self.artista = artista
        self. duracao  = duracao
    
    def exibir_musica(self):
        print(f"O nome da música é : {self.nome_musica}\n Do artista {self.artista}\n com duracao {self.duracao}")
        
```

```python
class playlist:
    def __init__(self, nome_playlist, usuario, lista_musica=None):
        self.nome_playlist = nome_playlist
        self.usuario = usuario  
        self.lista_musica = lista_musica if lista_musica is not None else []
    
    def adicionar_musicas(self, musica):
        self.lista_musica.append(musica)
        return f"Música {musica.nome_musica} adicioanada a playlist"
    
    def exibir_playlist(self):
        print(f"O usuário {self.usuario.nome} com playçist de nome {self.nome_playlist}, possui as músicas: ")
        for i in range(len(self.lista_musica)):
            musics = self.lista_musica[i]
            print(f"{musics.nome_musica} - {musics.artista} ({musics.duracao})")
```

#### 3. Pedido com produtos e cliente: Crie as classes Cliente, Produto e Pedido. O pedido deve usar as três formas de relacionamento estudadas: um método que recebe um objeto como parâmetro, um atributo que guarda um objeto e uma lista que guarda vários objetos. Por exemplo, o pedido pode ter um Cliente, guardar uma lista de Produto e possuir um método adicionar_produto(produto). Ao final, crie um método exibir_resumo() mostrando o cliente, os produtos e o valor total

```python
class Produto:
    def __init__(self, nome_produto, valor_produto):
        self.nome_produto = nome_produto
        self.valor_produto = valor_produto
        
```

```python
class Cliente:
    def __init__(self, nome_cliente):
        self.nome_cliente = nome_cliente

    def exibi_cliente(self):
        return f"Cliente {self.nome_cliente}"
```

```python
class Pedido:
    def __init__(self, cliente, nome_pedido, lista_pedidos = None):
        self.cliente = cliente
        self.nome_pedido = nome_pedido
        self.lista_pedidos = lista_pedidos if lista_pedidos is not None else []
    
    def adicionar_produto(self, produto):
        self.lista_pedidos.append(produto)
        return f"O pedido {self.nome_pedido} do produto {produto.nome_produto} de valor {produto.valor_produto} foi adicionado a lista"
    def exibir_resumo(self):
        valor_total = 0
        for total in self.lista_pedidos:
            print(f"Produto: {total.nome_produto} valor {total.valor_produto}")
            valor_total += total.valor_produto
        print(f"O valor gasto total é {valor_total}")
```
