#### **Adicionando métodos ao Ponto**:

```python
class Ponto:
    def __init__(self, x,y):
        self.x = x
        self.y = y

    def exibir(self):
        print("Ponto: ", self.x, self.y)

    def mover(self, move_x, move_y):
        self.x +=move_x
        self.y +=move_y
```

#### **Entendendo o self:**

#### Ideia principal:

#### Todo Método recebe self como primeiro parâmetro:

  - self representa o **próprio objeto** que chamou o método
      - em p1.mover(1,2), self é p1
      - e p2.mover(1,2), self é p2

  - self.x é o x **daquele** objeto

- def mover (self, dx, dy):
    - self.x += dx
    - self.y +=dy

#### Resumo:

Conceito| Ideia
---|---------|
Classe| Molde/*receita*-*PARA CRIAR OBJETOS*
Objeto| Elemento criado pelo molde
Atributo| Informação GUARDADA do objeto
Método| Ação que o objeto pode fazer
*self*| O próprio Objeto
*__init__*| Inicialização do Objeto (Construtor)

```python
class Personagem:
    def __init__(self, nome, vida, ataque, defesa, x,y):
        self.nome = nome
        self.vida = vida
        self.ataque = ataque
        self.defesa = defesa
        self.x =x
        self.y = y
        #Métodos

        def exibir_status(self):

            print(F"STATUS\nNome:{self.nome}\nVida:{self.vida}\nPosição:{self.x}:{self.y}")

        def mover(self,dx,dy):
            self.x +=dx
            self.y += dy
            print(f"Posição atual após movimentação de:{dx}:{dy}\n{self.x}:{self.y}")

        def estar_vivo(self):
            if self.vida >0:
                return True
            else:
                return False

        def receber_dano(self,valor):
            dano_final = valor - self.defesa
            vida = estar_vivo()
            if not vida:
                print("Personagem já está morto!Para de chutar cachorro morto!")
            else:
                self.vida -= dano_final
                if self.vida <=0:
                    self.vida = 0
                    print(f"MORREU! ☠️:{self.vida}")
                return self.vida
            
        ### Como referenciar outro objeto de mesma classe que ainda não foi criado dentro do método:?

        def atacar(self, outro):
            vida = estar_vivo()
            if not vida:
                print(f"Personagem derrotado!Não pode atacar a vida é :{self.vida}")
            else:
                print(f"Personagem:{self.nome} atacou {outro.nome}")
                outro.receber_dano(self.ataque)





            
```

O pedido deve usar as três formas de relacionamento estudadas: um método que recebe um objeto como parâmetro, um atributo que guarda um objeto e uma lista que guarda vários objetos. Por exemplo, o pedido pode ter um Cliente, guardar uma lista de Produto e possuir um método adicionar_produto(produto). Ao final, crie um método exibir_resumo() mostrando o cliente, os produtos e o valor total

```python
class Produto:
    def __init__(self, nome_produto, valor_produto):
        self.nome_produto = nome_produto
        self.valor_produto = valor_produto
        
```

```python
class Cliente:
    def __init__(self, nome_cliente):
        self.nome_cliente = nome_cliente

    def exibi_cliente(self):
        return f"Cliente {self.nome_cliente}"
```

```python
class Pedido:
    def __init__(self, cliente, nome_pedido, lista_pedidos = None):
        self.cliente = cliente
        self.nome_pedido = nome_pedido
        self.lista_pedidos = lista_pedidos if lista_pedidos is not None else []
    
    def adicionar_produto(self, produto):
        self.lista_pedidos.append(produto)
        return f"O pedido {self.nome_pedido} do produto {produto.nome_produto} de valor {produto.valor_produto} foi adicionado a lista"
    def exibir_resumo(self):
        valor_total = 0
        for total in self.lista_pedidos:
            print(f"Produto: {total.nome_produto} valor {total.valor_produto}")
            valor_total += total.valor_produto
        print(f"O valor gasto total é {valor_total}")
```

```python
class Disciplina:
    def __init__(self, cod_disciplina, nome, carga_horaria):
        self.cod_disciplina = cod_disciplina
        self.nome = nome
        self.carga_horaria = carga_horaria

    def exibir(self):
        print(f"Código:{self.cod_disciplina}\nNome Disciplina:{self.nome}\nCarga Horária:{self.carga_horaria}")
```

```python


class Matricula:
    def __init__(self, aluno, disciplina, nota_1, nota_2):
        ### Recebendo um objeto como parâmetro do construtor da classe matrícula
        self.aluno = aluno
        self.disciplina = disciplina
        self.nota_1 = nota_1
        self.nota_2 = nota_2
        self.lista_notas = [self.nota_1, self.nota_2]



    def calcular_media(self):
        media = sum(self.lista_notas) / len(self.lista_notas)
        return round(media,2)

    def obter_situacao(self):
        valor_media = self.calcular_media()
        if valor_media >= 70:
            return "Aprovado!"
        elif valor_media >=40 and valor_media < 70:
            return "Exame Especial!"
        elif valor_media <40:
            return "Reprovado!"
            
    def exibir_situacao(self):
        media = self.calcular_media()
        situacao = self.obter_situacao()
        print(f"Matrícula Aluno:{self.aluno.matricula}-Nome Aluno:{self.aluno.nome}\nCódigo da disciplina:{self.disciplina.cod_disciplina}-Nome Disciplina:{self.disciplina.nome}\nSituação:{situacao}-Valor Média:{media}")
```
