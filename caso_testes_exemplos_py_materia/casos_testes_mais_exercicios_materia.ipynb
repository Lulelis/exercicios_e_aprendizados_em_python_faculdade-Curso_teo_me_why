#### 🚴 Exercício 1 - Sistema de Bicicletas

#### Uma cidade implantou um sistema de **bicicletas compartilhadas**. Cada usuário informa o **tempo total de uso** da bicicleta em minutos,  se **possui plano mensal ativo** e se **já utilizou o sistema naquele mesmo dia**.

---

#### 🧩 Regras do sistema

- Usuários **com plano mensal** têm direito a até **45 minutos gratuitos por dia**.  
  Caso ultrapassem esse tempo, pagam **R$ 0,50 por minuto excedente**.  
- Usuários **sem plano mensal** pagam **R$ 1,00 por minuto** desde o primeiro minuto.  
- Nenhum usuário pode utilizar o sistema por mais de **180 minutos consecutivos**.  
  Caso isso ocorra, o uso é considerado **inválido** e o sistema deve **bloquear o encerramento** até que a situação seja regularizada.

---

#### ⚙️ O sistema deve determinar:

1. O **valor a ser pago**.  
2. Se o **uso é válido ou inválido**.

---

#### 💭 Desafio

- Como estruturar esse problema para que um **sistema automatizado** consiga decidir corretamente para qualquer usuário?


```python
def funcao_converte_entrada(entrada):
    entrada = entrada.strip().lower()
    if entrada in ('sim', 's', 'true', 'já', 'ja'):
        return True
    elif entrada in ('nao', 'n', 'não', 'false', 'nao sei', 'naosei', 'not'):
        return False
    else:
        return False  
```

```python
tempo = int(input("Qual o tempo total em minutos que se utilizou o sistema? "))
plano_mensal = str(input("Possui plano ativo? True or False")).lower()
utilizacao_dia = str(input("Já utilizou o sistema hoje? True or False")).lower()

plano_mensal_convertido = funcao_converte_entrada(plano_mensal)
utilizacao_dia_convertido = funcao_converte_entrada(utilizacao_dia)
valor_pago = 0
if tempo > 180 or utilizacao_dia_convertido == True:
    print("Acesso inválido!")
else:
    if plano_mensal_convertido == True:
        resto = tempo - 45
        valor_pago = resto * 0.50
        if resto <0:
            print("Usuário irá pagar R$0.00\nPois tem plano e usou dentro do limite de desconto do tempo")
        else:
            print(f"Usuário irá pagar:R${valor_pago:.2f}")
    elif plano_mensal_convertido == False:
        valor_pago = tempo * 1.00 
        print(f"Usuário sem plano irá pagar: {valor_pago:.2f}")
      
```

#### 📚 Exercício 2 - Sistema de Biblioteca

#### Uma biblioteca universitária permite o **empréstimo de livros físicos** aos alunos.  Cada aluno informa seu **tipo de vínculo** com a instituição (Graduação, Pós-Graduação ou Professor),  
a **quantidade de livros já emprestados** e se **possui ou não multas pendentes**.

---

#### 🧩 Regras do sistema

- Alunos de **Graduação** podem ter até **3 livros emprestados simultaneamente**.  
- Alunos de **Pós-Graduação**, até **5 livros**.  
- **Professores**, até **8 livros**.  
- Caso o aluno possua **multa pendente**, ele **não poderá realizar novos empréstimos**, independentemente da quantidade atual de livros.  
- O sistema deve **impedir qualquer novo empréstimo** caso a quantidade atual já esteja no limite permitido.

---

#### ⚙️ O sistema deve determinar:

1. Se o aluno **pode realizar um novo empréstimo**.  
2. Qual é o **limite máximo permitido** para aquele perfil.

---

#### 💭 Desafio

#### Como **modelar formalmente** esse problema antes de implementar qualquer algoritmo?


```python
def empresta_livros(tipo_vinculo, qtd_livros_emprestados, multas_pendentes, qtd_sera_emprestada):
    if tipo_vinculo =="graduacao" and multas_pendentes ==True:
        return 'Não será possível emprestar! Tem multas pendentes'
    else:
        if tipo_vinculo == "graduacao" and multas_pendentes ==False:
            saldo_disponivel = 3- qtd_livros_emprestados
            if saldo_disponivel <=0:
                return f'O saldo está indisponível, sem livros para emprestar{saldo_disponivel}!'
            else:
                return f'Será emprestado {qtd_sera_emprestada}\nCom Saldo disponível {saldo_disponivel - qtd_sera_emprestada}'
        elif tipo_vinculo =="pos graduacao" and multas_pendentes == True:
            return 'Não será possível emprestar! Tem multa'
        elif tipo_vinculo == "pos graduacao" and multas_pendentes ==False:
            saldo_disponivel = 5 - qtd_livros_emprestados
            if saldo_disponivel <=0:
                return 'Saldo indisponível para empréstimo!'
            else:
                return f'Será emprestado: {qtd_sera_emprestada}\nSaldo Disponível:{saldo_disponivel-qtd_sera_emprestada}'
        elif tipo_vinculo =="professor":
            saldo_disponivel = 8 - qtd_livros_emprestados
            if saldo_disponivel <=0:
                return 'Saldo Indisponível para empréstimos!'
            else:
                return f'Será emprestado: {qtd_sera_emprestada}\nSaldo Disponível:{saldo_disponivel- qtd_sera_emprestada}'
            
            
        
    
        
        
```

```python
print("Programa que simula empréstimos")
tipo_vinculo = input("Qual o tipo de vínculo: ").lower().strip()
while tipo_vinculo not in ("professor", "posgraduacao", "graduacao"):
    print("Apenas: Professor, posgraduacao e graduacao")
    tipo_vinculo = input("Qual o tipo de vínculo: ").lower().strip()
qtd_livros_ja_emprestados = int(input("Qual a quantidade de livros já emprestados? "))
multas_pendentes = input("Usuário tem multas pendentes? ")
qtd_sera_emprestada = int(input("Qual a quantidade que será emprestada? "))
multas_pendentes_convertidas = funcao_converte_entrada(multas_pendentes)
print(empresta_livros(tipo_vinculo, qtd_livros_ja_emprestados, multas_pendentes_convertidas, qtd_sera_emprestada))

```

#### 🚚 Exercício 3 - Plataforma de Delivery

#### Uma plataforma de delivery calcula a **taxa de entrega** com base na **distância**, no **valor do pedido** e no **horário da compra**.

---

#### 🧩 Regras do sistema

- Pedidos **acima de R$ 120** têm **entrega gratuita**, desde que a distância seja de até **8 km**.  
- Para pedidos **abaixo de R$ 120**, a taxa base é de **R$ 5,00** mais **R$ 1,50 por quilômetro**.  
- Entre **18h e 22h**, é aplicado um **adicional fixo de R$ 3,00** na taxa final.  
- A plataforma **não aceita pedidos** com distância superior a **20 km**.

---

#### ⚙️ O sistema deve determinar:

1. O **valor final da taxa de entrega**.  
2. Se o **pedido é aceito ou rejeitado**.

---

#### 💭 Desafio

- Como estruturar esse problema utilizando o **procedimento definido na disciplina**?


```python
def calcular_taxa_pedido(distancia_km, valor_pedido,horario_compra):
    if distancia_km > 20.00:
        print("Pedido Inválido!")
    else:
        if distancia_km <=8.00 and valor_pedido >=120.00:
            print("Pedido Válido!")
            print("Tem Entrega gratuita!")
            return f'O valor final da compra é: {valor_pedido:.2f}'

        elif valor_pedido < 120.00:
            print("Pedido Aceito!")
            if horario_compra >=18 and horario_compra <=22:
                valor_final = valor_pedido + 3 + 5 + (1.50 * distancia_km)
                return f'O valor final é: {valor_final:.2f}'
            else:
                print("Sem acréscimos!")
                valor_final = valor_pedido + 5 + (1.50 * distancia_km)
                return f'O valor final será de: {valor_final}'
            
```

```python
distancia_km = float(input("Digite a distância do pedido: "))
valor_pedido = float(input("Digite o valor do pedido: "))
horario_compra = int(input("Qual o horário de compra do pedido: "))
print(calcular_taxa_pedido(distancia_km, valor_pedido, horario_compra))
```
