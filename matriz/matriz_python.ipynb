#### Desafio 1: Médias por Jogador

#### Imagine que estamos desenvolvendo um sistema para analisar o desempenho de jogadores em um campeonato. As pontuações dos jogadores foram armazenadas em uma matriz, onde cada linha representa um jogador e cada coluna representa uma partida. Crie um programa que percorra essa matriz e calcule a média de pontuação de cada jogador.

#### Ao final, exiba o resultado no seguinte formato:
- ○ Media do jogador 1: X pontos
- ○ Media do jogador 2: X pontos

```python
def monta_matriz(jogador, partida):
    matriz_pontuacoes = []
    for i in range(jogador):
        jogadores = []
        for j in range(partida):
            pontuacoes = float(input(f"Insira as pontuações:{i}{j} "))
            jogadores.append(pontuacoes)
        matriz_pontuacoes.append(jogadores)
    return matriz_pontuacoes
```

```python
jogadores = int(input("QUANTOS JOGADORES? "))
while jogadores < 0:
    print("Não pode ser menor que zero!")
    jogadores = int(input("QUANTOS JOGADORES? "))
partidas = int(input("Quantas partidas efetuadas?"))
while partidas < 0:
    print("Não pode ser menor que zero!")
    partidas = int(input("QUANTas Partidas? "))
matriz = monta_matriz(jogadores, partidas)
for i in range(len(matriz)):
    print(matriz[i])
```

```python
def calcula_media(matriz):
    soma_media_geral = 0
    media_jogadores = []
    for i in range(len(matriz)):
        soma = 0
        media = 0
        for j in range(len(matriz[i])):
            soma += matriz[i][j]
        media = soma/len(matriz[i])
        soma_media_geral+=media
        media_jogadores.append(media)
        print(f"MÉDIA jogador {i}:{media}")
    return soma_media_geral, media_jogadores
```

```python
media_geral, media_jogadores = calcula_media(matriz)
print(f"A média geral é:{media_geral}\nLista da média de cada jogador:{media_jogadores}")
```

#### **9.** Crie uma matriz identidade (1 na diagonal, 0 no resto).

```python
def matriz_identidade(n):
    matriz_id = []
    for i in range(n):
        colunas =[]
        for j in range(n):
            if i ==j:
                valor =1
                colunas.append(valor)
            else:
                valor = 0
                colunas.append(valor)
        matriz_id.append(colunas)

    return matriz_id
```

#### Crie um programa onde o usuário passo digitar sete valores numéricos e cadastre-os em uma lista única que mantenha separado os valores pares e ímpares. No final, mostre os valores pares e ímpares em ordem crescente

```python
lista_vazia = []
for i in range(0,7):
    numeros = float(input("Digite o valor para se colocar na lista: "))
    lista_vazia.append(numeros)
print(lista_vazia)
```

```python
lista_par = []
lista_impar = []

for i in range(len(lista_vazia)):
    if lista_vazia[i] % 2 == 0:
        lista_par.append(lista_vazia[i])
    else:
        lista_impar.append(lista_vazia[i])
```

```python
print(lista_par)
#lista_impar
```

```python
nova_lista = []
nova_lista.append(lista_par)
```
