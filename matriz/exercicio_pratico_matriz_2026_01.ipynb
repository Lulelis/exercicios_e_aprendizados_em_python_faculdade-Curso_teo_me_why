#### Exercício_01: **Fazenda de Matriz**

#### Uma fazenda registrou níveis de umidade do solo em uma matriz quadrada. A diagonal secundária representa pontos de controle usados para medir a umidade mínima esperada. A parte interna da matriz representa uma área de cultivo sensível. O agrônomo deseja saber quantos pontos internos estão abaixo do menor valor observado nos pontos de controle. Sua tarefa é:

- (a) encontrar o menor valor da diagonal secundária;
- (b) contar quantos valores da parte interna da matriz são menores que esse menor valor.

#### *Observação: A parte interna é formada pelas posições que não pertencem às bordas da matriz.*

- #### **Entrada**:

- A primeira linha contém um inteiro  *N*, representando a dimensão da matriz quadrada.
- As próximas  *N linhas* contêm  *N números inteiros* cada, separados por espaço.

- #### **Saída**:

- Imprima um único número inteiro, correspondente à quantidade de valores internos que são menores que o menor valor da diagonal secundária.

- Restrições: *3 < N < 10*

-  4
-  9 8 7 20
-  6 2 30 5
-  4 20 3 6
-  10 7 8 9

#### **Exemplo de Saída**: *2*

```python
def definir_matriz_quadrada():
    n_linhas_n_colunas = int(input("Como a matriz é quadrada, o número de linhas e colunas é igual! Digite um valor maior que 3 e menor que 10!"))
    while n_linhas_n_colunas <=3 or n_linhas_n_colunas >=10:
        print("O número informado deve ser maior que 3 e menor que 10")
        n_linhas_n_colunas = int(input("Como a matriz é quadrada, o número de linhas e colunas é igual! Digite um valor maior que 3 e menor que 10!"))
    matriz_plantio = []
    for i in range(n_linhas_n_colunas):
        linhas = []
        for j in range(n_linhas_n_colunas):
            valor_matriz = int(input(f"Digite o valor presente na {i}ª linha e {j}º coluna: "))
            linhas.append(valor_matriz)
        matriz_plantio.append(linhas)
    return matriz_plantio
        
```

```python
matriz_plantio = definir_matriz_quadrada()
```

```python
def metricas_plantio(matriz):
    n_linhas_matriz = len(matriz)
    n_colunas_matriz = len(matriz[0])
    menor_valor_diagonal_secundaria = float("inf")
    for i in range(len(matriz)):
        valor = matriz[i][n_colunas_matriz-1-i]
        if valor < menor_valor_diagonal_secundaria:
            menor_valor_diagonal_secundaria= valor
    print(f"O Menor valor da diagonal secundária é: {menor_valor_diagonal_secundaria}")
    qtd_valores_menores = 0
    for i in range(len(matriz)):
        for j in range(len(matriz[i])):
            if i!=0 and j!=0 and i!=(n_linhas_matriz-1) and j !=(n_colunas_matriz-1):
                if matriz[i][j] < menor_valor_diagonal_secundaria:
                    qtd_valores_menores+=1
    return qtd_valores_menores
                    
```

```python
quantidade_valores_menores = metricas_plantio(matriz_plantio)
print(f"Quantidade de valores internos da matriz menores que a diagonal secundária: {quantidade_valores_menores}")
```

#### **Questão 2**:

#### A organização de um evento registrou os check-ins dos participantes. Cada registro contém o nome de um participante e um valor associado ao horário do seu check-in. Sua tarefa é imprimir o nome do participante que apareceu mais vezes nos registros.

#### **Entrada:**

- A primeira linha contém um inteiro Q, representando a quantidade de registros.

- As próximas  Q linhas contêm um nome de participante e um número inteiro, separados por espaço.

- A entrada deve ser obrigatoriamente armazenada em uma matriz.

#### **Saída:**

#### Imprima o nome do participante que apareceu mais vezes (o que mais fez check-ins).

#### Restrições: *1 < Q < 100*

#### **Exemplo de Entrada:**

- 7
- João 10
- Bárbara 9
- João 20
- Caio 8
- Bárbara 15
- Diego 11
- João 5

#### Exemplo de Saída: **João**

```python
def definir_matriz_participacao():
    matriz_participacao = []
    n_linhas = int(input("Digite o número de linhas da matriz: "))
    while n_linhas <=1 or n_linhas >=100:
        print("O número de linhas deve ser maior que 1 & menor que 100")
        n_linhas = int(input("Digite o número de linhas da matriz: "))
    n_colunas = int(input("Digite o número de colunas: "))
    while n_colunas <=1 or n_colunas >=100:
        print("O número de colunas deve ser maior que 1 & menor que 100")
        n_colunas = int(input("Digite o número de colunas da matriz: "))
    for i in range(n_linhas):
        linhas = []
        for j in range(n_colunas):
            valor_matriz = input(f"Digite o nome {i}º  ou a quantidade de participação {j}º:")
            if j ==1:
                print("Quantidade de participação!")
                valor_matriz = float(valor_matriz)
                linhas.append(valor_matriz)
            else:
                linhas.append(valor_matriz)
        matriz_participacao.append(linhas)
    return matriz_participacao

```

```python
matriz_participacao_evento = definir_matriz_participacao()
```

```python
def  analise_matriz_participantes(matriz):
    nome_com_maior_participacao =""
    maior_quantidade_participacao = float("-inf")
    dicionario_participacao = dict()
    for i in range(len(matriz)):
        dicionario_participacao[matriz[i][0]] = 0
    for i in range(len(matriz)):
        for chave, valor in dicionario_participacao.items():
            if chave == matriz[i][0]:
                dicionario_participacao[chave] += 1
    print(dicionario_participacao)
    for chave, valor in dicionario_participacao.items():
        if valor > maior_quantidade_participacao:
            maior_quantidade_participacao = valor
            nome_com_maior_participacao = chave
    return nome_com_maior_participacao, maior_quantidade_participacao
```

```python
nome_maior_participacao, maior_quantidade =  analise_matriz_participantes(matriz_participacao_evento)
print(f"Participante com maior quantidade de participação evento: {nome_maior_participacao}\nQuantidade que participou: {maior_quantidade}")
```
