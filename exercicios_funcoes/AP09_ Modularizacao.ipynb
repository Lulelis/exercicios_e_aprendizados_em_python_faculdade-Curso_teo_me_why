### **Exercício 4 Cálculo do valor final com desconto**:

#### Crie uma função chamada calcular_valor_final(valor_produto, desconto) que receba:

- o valor original de um produto;
- o percentual de desconto concedido ao cliente.
- A função deve retornar o valor final do produto após a aplicação do desconto.

#### No programa principal:

- leia o valor do produto;
- leia o percentual de desconto;
- exiba o valor final calculado.

```python
def calcular_valor_final(valor_produto, desconto):
    valor_final_desconto = valor_produto - (valor_produto * desconto/100)
    return valor_final_desconto
```

```python
valor_produto = float(input("Digite o valor do produto: "))
while valor_produto < 0:
    print("Não existe valor de produto negativo!")
    valor_produto = float(input("Digite o valor do produto: "))
desconto = float(input(f"Qual será o desconto % para o produto: {valor_produto}"))
while desconto < 0:
    print("Não existe desconto negativo!")
    desconto = float(input(f"Qual será o desconto % para o produto: {valor_produto}"))
valor_final = calcular_valor_final(valor_produto, desconto)
print(f"O valor do produto: {valor_produto} após desconto de:{desconto}%\nValor final: {valor_final:.2f} ")
```

#### **Exercício 6 Verificação de nota válida**:

#### Crie uma função chamada nota_valida(nota) que receba uma nota e retorne:

- True, se a nota estiver entre 0 e 100, inclusive;
- False, caso contrário.

#### No programa principal:

- leia uma nota;
- exiba o resultado retornado.

```python
def nota_valida(nota):
    nota_bool = True
    if nota >=0 and nota <=100:
        print("Nota Válida!")
        return nota_bool
    else:
        print("Nota Inválida!")
        nota_bool = False
        return nota_bool
```

### **Exercício 11 Cálculo da inscrição com taxa padrão**

#### Crie uma função chamada calcular_inscricao(valor_base, taxa=10) que receba:

- o valor base da inscrição;
- a taxa percentual de serviço, que por padrão será 10.
- A função deve retornar o valor total da inscrição.

#### No programa principal:

- leia o valor base;
- chame a função uma vez sem informar a taxa;
- chame a função outra vez informando uma taxa diferente;
- exiba os dois resultados.

```python
def calcular_inscricao(valor_base, taxa=10):
    valor_total_inscricao = valor_base + (valor_base * taxa/100)
    return valor_total_inscricao
```

### **Exercício 12 Conversão do tempo de prova**

#### Crie uma função chamada converter_tempo(total_segundos) que receba o tempo total de prova em segundos e retorne três valores:

- horas;
- minutos;
- segundos restantes.

#### No programa principal:

- leia o tempo total em segundos;
- chame a função;
- exiba os valores retornados

```python
def converter_tempo(tempo_total):
    horas = tempo_total//3600 ### Valor inteiro das horas
    resto_segundos = tempo_total%3600 ### resta da divisão por segundos em horas
    minutos = resto_segundos//60 ### pegadno o resto e convertendo para minutos em divisão inteira
    segundos = resto_segundos%60 ### Resto da Divisão do restos segundos para segundos de fato
    return horas, minutos, segundos
```

```python
tempo_total_segundos = int(input("Informe o tempo total em segundos:"))
while tempo_total_segundos < 0:
    print("Deve ser maior que zero")
    tempo_total_segundos = int(input("Informe o tempo total em segundos:"))
horas, minutos, segundos = converter_tempo(tempo_total_segundos)
print(f"Tempo total de segundos informado: {tempo_total_segundos}\n{horas}Hora(s),{minutos}Minuto(s),{segundos}Segundo(s)")
```

### **Exercício 14 Faixa de números de peito**

#### Em uma corrida, a organização deseja identificar rapidamente quantos números de peito pares existem em uma faixa numérica, pois eles serão destinados a uma categoria específica.

#### Crie duas funções:


#### a) eh_par(numero)

- Deve retornar True se o número for par e False caso contrário.

#### b) contar_pares_faixa(inicio, fim)

- Deve percorrer todos os números do intervalo de inicio até fim, inclusive, e contar quantos deles são pares.

#### Importante: dentro da função contar_pares_faixa, a verificação deve ser feita chamando a função eh_par(numero).

#### No programa principal:

- leia o início e o fim da faixa;
- exiba a quantidade de números pares.

```python
def eh_par(numero):
    par = True
    if numero%2 ==0:
        return par
    elif numero%2 !=0:
        par = False
        return par
```

```python
numero = int(input("Digite um número"))
print(eh_par(numero))
```

```python
def contar_pares(inicio, fim):
    count_pares = 0
    for i in range(inicio, fim+1):
        par = eh_par(i)
        if par:
            count_pares+=1
        elif not par:
            continue
    return count_pares
```

```python
inicio_faixa = int(input("Digite a primeira faixa: "))
fim_faixa = int(input("Digite o fim da faixa: "))
qtd_pares = contar_pares(inicio_faixa, fim_faixa)
print(f"Quantidade de faixas pares:{qtd_pares} entre: {inicio_faixa} & {fim_faixa}")
```
