#### Em sistemas simples de segurança, é comum aplicar transformações básicas em textos antes de armazená-los, como forma de dificultar a leitura direta das informações. Uma dessas estratégias consiste em substituir cada letra de uma palavra pela próxima letra do alfabeto, gerando uma versão modificada do texto original. Nosso objetivo é simular esse comportamento:

- ○ receber uma palavra
- ○ transformar cada caractere na próxima letra do alfabeto
- ○ gerar uma nova palavra com o resultado dessa transformação
- ○ exibir o texto transformado

```python
texto = "casa"
resultado = ""

for i in texto:
    if i =="z":
        resultado += "a"
    else:
        resultado += chr(ord(i)+ 1)
print(resultado)
```

#### O sistema continua evoluindo. Agora, em vez de trabalhar com apenas um valor, ele precisa lidar com mais de uma informação ao mesmo tempo. O usuário irá informar dois números, e o sistema deve calcular o triplo de cada um deles e, ao final, exibir a soma desses resultados. Como podemos resolver esse problema sem repetir a mesma lógica de cálculo várias vezes?

```python
def triplo_numero(numero):
    resultado = numero * 3
    return resultado
```

```python
entrada_1 = int(input("informe um número: "))
entrada_2 = int(input("informe um número: "))

resultado_entrada_1 = triplo_numero(entrada_1)
resultado_entrada_2 = triplo_numero(entrada_2)

print(f"A soma dos dois triplos é: {resultado_entrada_1+resultado_entrada_2}")
```

#### Agora que o sistema já consegue criptografar textos individuais, surgiu uma nova necessidade.Além de transformar separadamente o nome e a senha de um usuário, o programa tambémprecisa gerar uma assinatura criptografada a partir da combinação dessas informações.Para isso, é possível reutilizar a lógica de criptografia já construída, aplicando-a tanto aos dadosindividuais quanto à junção dos resultados. Nosso objetivo é simular esse comportamento:

○ receber dois textos (nome e senha)
○ gerar as versões criptografadas de cada um deles
○ combinar os textos criptografados
○ aplicar novamente a criptografia sobre essa combinação
○ exibir os resultados obtidos


```python
def criptografar_2(texto):
    resultado_1 =""
    for i in texto:
        if i =="z":
            resultado_1 +="a"
        else:
            resultado_1 += chr(ord(i)+ 1)
    return resultado_1
```

```python
def criptografando(texto,senha):
    resultado_1 = ""
    resultado_2 = ""
    resultado_3 = ""
    assinatura_final = ""

    for i in texto:
        if i =="z":
            resultado_1 +="a"
        else:
            resultado_1 += chr(ord(i)+ 1)
    for j in senha:
        if j =="z":
            resultado_2 +="a"
        else:
            resultado_2 =  resultado_2 + chr(ord(j)+ 1)
    resultado_3 = resultado_1 + resultado_2

    ### Aplicando novamente a criptografia sob a combinação:
    for k in resultado_3:
        if k =="z":
            assinatura_final +="a"
        else:
            assinatura_final += chr(ord(k)+ 1)

    return assinatura_final
```

```python
assinatura = criptografando("Leandrinha","Lucas")
print(assinatura)
```

#### Exercício 12 Conversão do tempo de prova

Crie uma função chamada converter_tempo(total_segundos) que receba o tempo total de prova em segundos e retorne três valores:

- horas;
- minutos;
- segundos restantes.

No programa principal:

- leia o tempo total em segundos;
- chame a função;
- exiba os valores retornados.

```python
def converter_tempo(total_segundos):
   

    ### cada hora tem 3600 segundos. Para se descobrir quantas horas cabem dentro do total, se usa  DIVISÃO INTEIRA COM //
    horas = total_segundos // 3600

    ### Recendo o que sobrou  após tirar as horas:
    resto = total_segundos%3600

    ### Minutos: Cada minuto tem 60 segundos  . Então, dentro do resto, se usa novamente a divisão inteira  
    minutos = resto //60

    #### Segundos finais: o que sobrar depois de tirar os munutos:
    segundos =  resto%60
    return f"{horas} horas, {minutos} minutos,e {segundos} segundos"
```

```python
segundos = int(input("Informe o tempo total em segundos: "))

converter_tempo(segundos)
```

##### Em uma corrida, a organização deseja identificar rapidamente quantos números de peito pares existem em uma faixa numérica, pois eles serão destinados a uma categoria específica.

- Crie duas funções:

- a) eh_par(numero)

*Deve retornar True se o número for par e False caso contrário.*

- b) contar_pares_faixa(inicio, fim)

*Deve percorrer todos os números do intervalo de inicio até fim, inclusive, e contar quantos deles são pares.*

- Importante: dentro da função contar_pares_faixa, a verificação deve ser feita chamando a função eh_par(numero).

No programa principal:

leia o início e o fim da faixa;

**exiba a quantidade de números pares.**

```python

def eh_par(numero):
    if numero %2 ==0:
        return True
    else:
        return False
```

```python
def contar_pares_faixa(inicio, fim):
    count = 0
    for i in range(inicio, fim+1):
        if eh_par(i):
            count +=1
    return count
```

```python
numero = int(input("Digite um número para verificar se é par"))
numero_inicio = int(input("Digite o primeiro número do intervalo: "))
numero_fim = int(input("Digite o segundo número do intervalo: "))

print(eh_par(numero))
print(contar_pares_faixa(numero_inicio, numero_fim))
```

#### Em um sistema que armazena informações de usuários, as datas são inseridas manualmente. Antes de qualquer processamento mais complexo, o sistema precisa garantir que a data foi informada no formato correto. Por exemplo, uma data válida deve seguir o padrão: DD/MM/AAAA Se o formato estiver incorreto, o sistema não deve continuar o processamento. Como podemos construir uma função que verifique se uma data está nesse formato antes de utilizá-la?

```python
#### Entrada:
#- Uma String representando uma data

#### Processamento:
# Verificar se a string possui exatamente 10 caracteres
# Verificar se existem '/' nas posições corretas
# Verificar se os demais caracteres são números

#### Saída:

#- Verdadeiro (formato válido) 00/00/0000
#- Falso (formato inválido)


### Recebe uma string
data = input("Insira uma data: ")

def verifica_data(data):

    ### Verificar regras específicas e retorna True Ou False
    ### Represent auma responsabilidade clara no sistema
    ### Primeiro passo para construir uma biblioteca de funções
    if len(data) != 10:
        return False
    if data[2] != '/' or data[5] != '/':
        return False
    for i in range(len(data)):
        if i != 2 and i !=5:
            if data[i] < '0' or data[i] > '9':
                return False
    return True
```

```python
data_informada = input("Insira uma data: ")

verifica_data(data_informada)
```
