#### Imagine que estamos desenvolvendo um jogo. Um jogador realizou 5 jogadas, e em cada jogada ele fez uma pontuação. Crie um programa que:

- ○ leia as pontuações das 5 jogadas
- ○ armazene esses valores
- ○ calcule a média das pontuações
- ○ exiba:
■ uma lista com as pontuações acima da média
■ uma lista com as pontuações abaixo da média

```python
### Entrada: Pontuação de Cada jogada

### Processamento: armazenar os valores, calcular a média dos valores armazenados e separar os valores em dois grupos

### Saída: Lista de Valores acima da média e lista de valores abaixo da média

lista_pontuacoes = []
for i in range(1,5+1):
    pontuacoes = float(input("Escreve a pontuação do jogador"))
    lista_pontuacoes.append(pontuacoes)
soma = sum(lista_pontuacoes)
media = soma / len(lista_pontuacoes)
lista_pontuacoes_abaixo = []
lista_pontuacoes_acima = []
for k  in lista_pontuacoes:
    if k > media:
        lista_pontuacoes_acima.append(k)
    elif k < media:
        lista_pontuacoes_abaixo.append(k)
print(lista_pontuacoes)
print(lista_pontuacoes_acima)
print(lista_pontuacoes_abaixo)
        
```

### Pensando em uma estrutura de Matriz: ***Listas de Listas***

Pensando sempre assim:
matriz[i] [j]

- **i** --> "Qual linha?"
- **j** --> "Qual coluna"?

Exemplo

#### Matriz[1][2] --> Linha 1, Coluna 2 

#### Acessando os valores assim:



```python
pontuacoes= [
    [10,20, 30, 40,50],
    [15, 25, 35, 45, 55],
    [5, 10, 15, 20, 25]
]

print(pontuacoes[0])
```

```python
print(pontuacoes[0][0])
```

```python
### Elemento 25 Linha x Coluna
print(pontuacoes[2][4])
```

```python
### Elemento 15: linha x Coluna
print(pontuacoes[2][2])
```

#### Percorrendo a estrutura:

- Para percorrer uma lista, fazemos?:

```python
for i in pontuacoes:
    print(i)
```

--> Percorrendo a coleção de valores!

--> Percorrendo jogador por jogador. Ou seja, lista por lista!

#### Para acessar cada pontuação separado: ***FOR Aninhado***

```python
for i in pontuacoes:
    for j in i: ### nota-se que o segundo for dentro do for sempre vai percorrer  a primeira variável temporária do for anterior
        print(j)
```

#### Percorrendo índices:

```python
for i in range(len(pontuacoes)):
    for j in range(len(pontuacoes[i])):
        print(pontuacoes[i][j])
```
