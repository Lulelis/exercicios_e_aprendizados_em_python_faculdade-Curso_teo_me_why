#### Imagine que estamos desenvolvendo um jogo. Um jogador realizou 5 jogadas, e em cada jogada ele fez uma pontuação. Crie um programa que:

- ○ leia as pontuações das 5 jogadas
- ○ armazene esses valores
- ○ calcule a média das pontuações
- ○ exiba:
■ uma lista com as pontuações acima da média
■ uma lista com as pontuações abaixo da média

```python
### Entrada: Pontuação de Cada jogada

### Processamento: armazenar os valores, calcular a média dos valores armazenados e separar os valores em dois grupos

### Saída: Lista de Valores acima da média e lista de valores abaixo da média

lista_pontuacoes = []
for i in range(1,5+1):
    pontuacoes = float(input("Escreve a pontuação do jogador"))
    lista_pontuacoes.append(pontuacoes)
soma = sum(lista_pontuacoes)
media = soma / len(lista_pontuacoes)
lista_pontuacoes_abaixo = []
lista_pontuacoes_acima = []
for k  in lista_pontuacoes:
    if k > media:
        lista_pontuacoes_acima.append(k)
    elif k < media:
        lista_pontuacoes_abaixo.append(k)
print(lista_pontuacoes)
print(lista_pontuacoes_acima)
print(lista_pontuacoes_abaixo)
        
```

#### Construindo uma matriz 3 x 5: 3 Linhas e 5 Colunas

#### **Listas Multimdimensionais**

- É possível compor listas, nas quais cada elemento é uma nova lista por si só

   - Listas de Listas
   - Matrize
   - Multidimensional

#### **Matrizes**
- Matriz é uma arranjo Multidimensional

   - Bidimensional *MxN* onde ***M*** é o *número de Linhas* e ***N*** o *número de Colunas*

   - Dois ou mais índices:
      
      - Geralmente ***i*** e ***j*** para *linhas* e *colunas*, respectivamente

### Matriz 3 x 3

```python
matriz_3x3 = []

for i in range(3):
    linhas_e_colunas= []
    for j in range(3):
        numeros =  int(input("Insira os números  para compor a matriz: "))
        linhas_e_colunas.append(numeros)
    matriz_3x3.append(linhas_e_colunas)
print(matriz_3x3)
```

--> Imprimindo apenas os elementos da primeira linha:

```python
### Imprimindo os elementos da primeira linha:

for i in matriz_3x3[0]:
    print(i)
```

```python
for i in matriz_3x3[0]:
    for j in matriz_3x3[0]:
        print(j)
```

```python
for i in matriz_3x3:
    for j in i:
        print(j, end="")
```

---> Todos os elementos da Matriz?

```python
for linha in matriz_3x3:
    for elem in linha:
        print(elem, end="")
```

--> Posição de um elemento de uma matriz depende de uma mais de um índice: Bidemensional: 2 ÍNDICES

```python
for i in matriz_3x3:
    print(i)
```

#### --> Acessando elemento a elemento da matriz:

--> 2 índcies para matriz Bidimencional

```python
### Penúltimo

print(matriz_3x3[2][1])
```
