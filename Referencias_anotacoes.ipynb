### Alterar vs Sobrescrever

✅ Alterar Conteúdo --> **Alteração Direta**

- lista[0] = 99

Altera o valor em uma posição específica.
Modifica a estrutura original (afeta o escopo
global).

❌ Sobrescrever Variável --> Nova Atribuição

- lista = lista_original.copy()

Atribui uma nova estrutura à variável.
Cria uma nova lista isolada apenas no escopo local
(quebra a ponte com o global).



```python
lista_original = [23,2843,274]

```

```python
lista = lista_original.copy()
```

```python
lista[0] = 200
```

### <u>**Resumo: Mutabilidade e Parâmetros**</u>

#### **Tipos Imutáveis**

- int, float, bool, tuple

#### Passagem por Valor:

--> A função trabalha com uma **cópia** isolada

--> O dado original está **seguro** contr a alterações internar

> **Segurança de Dados**

#### **Tipos Mutáveis**

- list, dict, set

#### ***Passagem por referência***

--> A função recebe o **endereço de memória**.

--> O conteúdo original **será modificado** se você alterar índices ou chaves.

> **Altera o original**

#### **2.** Crie um procedimento que receba uma lista de preços de produtos. O procedimento deve percorrer a lista através de seus índices e aplicar um desconto de 5% em cada valor. No programa principal, imprima a lista antes e depois da chamada da função para comprovar a modificação in-place (passagem por referência).


```python
produtos = []
n = int(input("Quantos produtos estarão na lista? "))
for i in range(n):
    preco = float(input(f"Digite o preço do produto {i}: "))
    produtos.append(preco)
print(produtos)

lista_copia_produtos = produtos.copy()
```

```python
lista_copia_produtos
```

```python
def desconto(lista):
    desconto = 0.05
    for i in range(len(lista)):
        lista[i] = lista[i] - (desconto * lista[i])
```

```python
desconto(produtos)
```

```python
produtos
```

```python
lista_copia_produtos
```

#### ✅ Procedimento (sem return) — a função desconto(lista) modifica os valores diretamente dentro da lista recebida.

#### ✅ Passagem por referência — como listas são mutáveis, a função altera o conteúdo original (produtos), e isso é comprovado quando você imprime antes e depois.

#### ✅ Cópia preservada — lista_copia_produtos = produtos.copy() mostra que apenas a lista original foi modificada, enquanto a cópia continua igual.

#### 4: Crie um procedimento que receba um dicionário contendo os dados de um jogador (nome, nível, pontuação) e um número inteiro representando os pontos que ele acabou de ganhar. O procedimento deve somar os novos pontos à pontuação atual no dicionário. Imprima o dicionário no programa principal após a chamada para validar a alteração referencial.

#### > Nesse exercício, o ponto central é mostrar mutabilidade e passagem por referência usando um dicionário.

```python
jogador = dict()

for i in range(1):
    nome_jogador = input("Digite o nome do jogador: ")
    jogador["nome"] = nome_jogador
    nivel = int(input(f"Digite o nível do jogador {nome_jogador}"))
    jogador["nivel"] = nivel
    pts = float(input("Digite quantos pontos o jogador marcou:"))
    jogador["pontos"] = pts
print(jogador)
```

```python

def soma_pontos(dicio, pontos):

    for chave, valor in dicio.items():
        if chave == "pontos":
            dicio["pontos"] +=pontos
    print(f"O dicionário atulizado: {dicio}")
```

```python
pontos = int(input("Informe a quantidade pontos para se atualizar: "))
soma_pontos(jogador, pontos)
```

#### O parâmetro dicio não é uma cópia: ele aponta para o mesmo espaço de memória do dicionário original (jogador).

#### Como dicionários são mutáveis, qualquer alteração feita dentro da função afeta o objeto original.

#### Por isso, quando você soma os pontos, a chave "pontos" do dicionário é atualizada tanto dentro quanto fora da função.

> #### 🧠 **Em termos de referência** O Python não cria uma nova estrutura, ele apenas passa o endereço de memória do dicionário. Então, dicio["pontos"] += pontos modifica o mesmo objeto que jogador está usando.
