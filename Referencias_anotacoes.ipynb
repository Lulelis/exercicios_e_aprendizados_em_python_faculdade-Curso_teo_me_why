### **Nomes e Valores**

- #### Uma variável não é uma "caixa" que contém o valor.
- #### Ela é um nome que aponta para um valor guardado na memória.
- #### Dizemos que a variável guarda uma referência para esse valor.
                          - - palavra ----->  -"sol"

### **Exemplo com Números:**

#### a = 10

#### b= a - <u>*Aqui, a & b = 10*</u>

#### b = b + 5 - aqui, depois dessa expressão b = b + 5 | a = 10 e b = 15/ foi criado uma nova referência para o valor de b
   - b + 5 cria um novo valor, e b passa a apontar para ele. O valor de a nunca foi tocado






### **Exemplo com Listas:**

#### - a = ["sol", "abc"]
#### - b = a

#### Realizando: b.append("fim")

#### a & B **APONTAM** para o mesmo valor na memória: ["sol", "abc", "fim"]
    - a & b apontam para a mesma lista. Modificar a lista por um none afeta o que se vê pelo outro

### **Atribuir é Diferente de Modificar**:

#### Existem duas operações muito diferentes:


| **Operação** | **Exemplo** | **O que acontece**|
|:-----------|:----------:|:------------:|
| Atribuir     | b = b + 5 | *o nome passa a apontar para outro valor*|
| Atribuir     | b = [7,8]| *O nome passa a apontar para outra lista*/ outra lista é criada| 
| Modificar     | b.append("fim") | **a própria lista é alterada**|
| Modificar  | b[0] = "x" | **A própria lista é alterada**|

### <u>**Resumo: Mutabilidade e Parâmetros**</u>

#### **Tipos Imutáveis**

- int, float, bool, tuple

#### Passagem por Valor:

--> A função trabalha com uma **cópia** isolada

--> O dado original está **seguro** contr a alterações internar

> **Segurança de Dados**

#### **Tipos Mutáveis**

- list, dict, set

#### ***Passagem por referência***

--> A função recebe o **endereço de memória**.

--> O conteúdo original **será modificado** se você alterar índices ou chaves.

> **Altera o original**

#### **2.** Crie um procedimento que receba uma lista de preços de produtos. O procedimento deve percorrer a lista através de seus índices e aplicar um desconto de 5% em cada valor. No programa principal, imprima a lista antes e depois da chamada da função para comprovar a modificação in-place (passagem por referência).


```python
produtos = []
n = int(input("Quantos produtos estarão na lista? "))
for i in range(n):
    preco = float(input(f"Digite o preço do produto {i}: "))
    produtos.append(preco)
print(produtos)

lista_copia_produtos = produtos.copy()
```

```python
lista_copia_produtos
```

```python
def desconto(lista):
    desconto = 0.05
    for i in range(len(lista)):
        lista[i] = lista[i] - (desconto * lista[i])
```

#### ✅ Procedimento (sem return) — a função desconto(lista) modifica os valores diretamente dentro da lista recebida.

#### 4: Crie um procedimento que receba um dicionário contendo os dados de um jogador (nome, nível, pontuação) e um número inteiro representando os pontos que ele acabou de ganhar. O procedimento deve somar os novos pontos à pontuação atual no dicionário. Imprima o dicionário no programa principal após a chamada para validar a alteração referencial.

#### > Nesse exercício, o ponto central é mostrar mutabilidade e passagem por referência usando um dicionário.

```python
jogador = dict()

for i in range(1):
    nome_jogador = input("Digite o nome do jogador: ")
    jogador["nome"] = nome_jogador
    nivel = int(input(f"Digite o nível do jogador {nome_jogador}"))
    jogador["nivel"] = nivel
    pts = float(input("Digite quantos pontos o jogador marcou:"))
    jogador["pontos"] = pts
print(jogador)
```

```python

def soma_pontos(dicio, pontos):

    for chave, valor in dicio.items():
        if chave == "pontos":
            dicio["pontos"] +=pontos
    print(f"O dicionário atulizado: {dicio}")
```

```python
pontos = int(input("Informe a quantidade pontos para se atualizar: "))
soma_pontos(jogador, pontos)
```

#### O parâmetro dicio não é uma cópia: ele aponta para o mesmo espaço de memória do dicionário original (jogador).

#### Como dicionários são mutáveis, qualquer alteração feita dentro da função afeta o objeto original.

#### Por isso, quando você soma os pontos, a chave "pontos" do dicionário é atualizada tanto dentro quanto fora da função.

> #### 🧠 **Em termos de referência** O Python não cria uma nova estrutura, ele apenas passa o endereço de memória do dicionário. Então, dicio["pontos"] += pontos modifica o mesmo objeto que jogador está usando.

#### 8. **Desafio da Cópia Acidental:** Crie um procedimento que receba uma lista e altere o primeiro valor dela para 0. No programa principal, crie uma lista preenchida. Acione o procedimento, mas em vez de passar a lista normalmente, passe uma cópia dela (ex: minha_lista[:]). Imprima a lista original no terminal e justifique o resultado observado. Exercícios de Fixação.

```python
lista = [1,3,4,5,6,7,8,9,0]
```

```python
def modifica_lista(lista):
    for i in range(len(lista)):
        lista[0] = 0
    print(f"A lista Alterada  é: {lista}")
```

```python
minha_lista = lista.copy()
```

```python
modifica_lista(minha_lista)
```

#### Justificativa: 

--> `lista` & `minha_lista` têm valores iguais, mas endereços de memória diferentes, por conta da cópia que foi gerada. O **python cria uma nova referência** para um novo objeto com o mesmo conteúdo. Nesse sentido, como foi criada uma cópia (minha_lista) independente da lista original (lista), o py gerou um novo objeto na memória para `minha_lista`, assim, a modificação feita dentro do procedimento afetou apenas a cópia, e não o objeto original

#### Caso não passasse minha_lista e sim lista para o procedimento:

--> Quando você passa a lista original para a função, o Python não cria uma cópia; ele apenas compartilha o mesmo endereço de memória. Como listas são mutáveis, qualquer alteração feita dentro da função (como lista[0] = 0) modifica o conteúdo diretamente nesse endereço.

--> Por isso, o valor é alterado permanentemente — tanto dentro quanto fora da função.


--> **Tipos mutáveis (como list, dict, set)** → passagem por referência → o conteúdo original é modificado.

--> **Tipos imutáveis (como int, float, str, tuple)** → passagem por valor → o conteúdo original permanece intacto.

Você está raciocinando exatamente como o material da PUC Minas mostra: entender não só o “como”, mas o porquê da mutabilidade.

#### Exercício 2. Implemente uma função dobrar_todos que recebe uma lista de números e devolve uma nova lista com cada valor dobrado, sem alterar a original.

#### Exemplo:

- ○ entrada: [1, 2, 3]
- ○ saída: [2, 4, 6]

```python
def dobrar_numeros(lista):
    lista_original = list(lista)
    nova_lista = []
    for i in range(len(lista_original)):
        dobrar = lista_original[i] * 2
        nova_lista.append(dobrar)
    return nova_lista
```

#### **Exercício 4.** Implemente uma função remover_invalidas que recebe uma lista de mensagens e devolve uma nova lista apenas com as que têm somente letras minúsculas. Reaproveite so_minusculas. Exemplo:

- ○ entrada: ["sol", "Sol", "ab c", "abc"]
- ○ saída: ["sol", "abc"]

```python
def so_minusculas(lista):
    nova_lista = []
    for i in range(len(lista)):
        for j in lista[i]:
            if j < "a" or j > "z":
                lista[i] = lista[i].lower()
        nova_lista.append(lista[i])
    return nova_lista
```

```python
entrada = ["sol", "Sol", "abC"]
```

```python
nova_lista = so_minusculas(entrada)
```
