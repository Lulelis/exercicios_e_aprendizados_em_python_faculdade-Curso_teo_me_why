#### Uma função recursiva é uma função que chama ela mesma para resolver uma versão menor do problema


#### --> Ela é composta por 2 partes principais:
       -1: Caso BASE: Condição que encerra a recursão
       -2: Caso Recursivo: Chamada da Própria Função

### **Pilhas de Execução**: 

--> A execução acontece em 2 Etapas:

  - Ida: Empilhando as chamadas
  - Volta: Resolvendo e Desempilhando

  

```python
def fatorial(n):
    if n ==1:
        return 1
    else:
        return n * fatorial (n-1)
        
```

#### Se chamamos a função fatorial(3), como resolvemos a seguinte linha?

--> ***return n * fatorial (n-1)***

#### Sabemos o valor de fatorial(2)? AINDA não!

#### Então, EMPILHAMOS a chamada de ***fatorial(2)***

```python
print(fatorial(3))
```

#### Versão iterativa vs recursiva


#### **Nem sempre a versão recursiva é melhor!**

- ● a versão iterativa usa laço;
- ● a versão recursiva usa chamadas de função;
      - ○ Existe um limite de chamadas!
- ● nem sempre recursão é mais eficiente;

- ● recursão pode ser mais elegante quando o problema tem estrutura naturalmente recursiva.

#### **Se faltar o caso base?**

#### A função nunca para de chamar a si mesmo:
  - Cada chamada empilha um novo quadro
  - Nenhum quadro é retirado
  - A pilha cresce até estourar

### - Recapitulando:

#### - 1. Uma função recursiva chama ela mesma para um problema menor.
#### - 2. O caso base encerra; o caso recursivo continua.
#### - 3. Cada chamada vira um quadro na pilha: a última a entrar é a primeira a sair.
#### - 4. Na ida as chamadas empilham; na volta os resultados voltam.
#### - 5. Sem caso base, a pilha cresce até estourar.

- ○ o problema diminui a cada chamada?
- ○ o que acontece na ida e o que acontece na volta?

#### A sequência de Fibonacci é aquela famosa sequência de número onde cada termo é a soma dos dois anteriores:

0,1,1,2,3,5,813,21,34...

#### Implementando um código recursivo em python para calcular o termo N da sequência de Fibonacci

```python
def fibonaci(n):
    ### Casos base: 
    if n ==0:
        return 0
    elif n ==1:
        return 1
    ### Caso recursivo (Dupla Chamada)    
    else:
        return fibonaci(n-1) + fibonaci(n-2)
```

#### 1. Crie uma função recursiva chamada soma_ate(n) que receba um número inteiro positivo n e retorne a soma de todos os números de 1 até n.

```python
def soma_ate(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return n + soma_ate(n-1)
```

```python
print(soma_ate(3))
```

#### **2.** Crie uma função recursiva chamada potencia(base, expoente) que receba dois números inteiros, base e expoente, e retorne o valor de base elevado a expoente.

```python
def potencia(base, expoente):
    if expoente ==0:
        return 1
    elif expoente ==1:
        return base
    else:
        return base * potencia(base, expoente-1)
```

```python
print(potencia(3,6))
```

#### 3. Crie uma função recursiva chamada contar_digitos(n) que receba um número inteiro positivo n e retorne a quantidade de dígitos desse número

```python
def contar_digito(n):
    if n==0:
        return 1
    elif n==1:
        return 1
    else:
        return n + contar_digito(n)
```

```python
contar_digito(4)
```

#### Um aplicativo de alfabetização mostra quantas vogais existem em cada palavra digitada pela criança.

#### Escreva uma função recursiva contar_vogais(texto, i) que devolva quantas vogais existem no texto a partir da posição i. Na primeira chamada, i vale 0. Por exemplo:

```python
texto = "banana"
len(texto)
```

```python
def contar_vogais(texto,i):
    if i ==  (len(texto)):
        return 0
    elif texto[i] not in ("a", "e", "i", "o", "u"):
        return contar_vogais(texto,i+1 )
    elif texto[i] in ('a', 'e', 'i', 'o', 'u'):
        return  1+  contar_vogais(texto,i+1 )
```

```python
print(contar_vogais("banana", 0))
```

#### Uma biblioteca guarda em uma lista os códigos dos livros disponíveis. Antes de fazer uma reserva, o sistema precisa saber se o código procurado está na lista.

#### Escreva uma função recursiva existe(lista, valor, i) que devolva True se valor aparece na lista a partir da posição i, e False caso contrário. Por exemplo:



<div style="text-align: center;">
  <img src="ex_lista.png" width="900" height="400">
</div>


```python
def existe(lista, valor, i):
    if i== len(lista):
        return False
    elif lista[i] == valor:
        return True
    else:
        return existe(lista, valor, i+1) 
```

```python
lista = []
n= int(input("Quantos elementos deverá estar contido na listA? "))
while n < 0:
    print("Deve ser maior que zero!")
    n= int(input("Quantos elementos deverá estar contido na listA? "))

for i in range(n):
    valor = int(input(f"Digite o {i+1} valor: "))
    lista.append(valor)

```

```python
valor_procurado = int(input("Qual valor você procurará na listA? "))
```

```python
print(existe(lista, valor_procurado, 0))
```
