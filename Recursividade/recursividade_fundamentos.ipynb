##  🎲<u>**Recursividade Fundamentos**</u>



#### Uma função recursiva é uma função que chama ela mesma para resolver uma versão menor do problema


#### --> Ela é composta por 2 partes principais:
       -1: Caso BASE: Condição que encerra a recursão
       -2: Caso Recursivo: Chamada da Própria Função

### **Pilhas de Execução**: 

--> A execução acontece em 2 Etapas:

  - Ida: Empilhando as chamadas
  - Volta: Resolvendo e Desempilhando

  

```python
def fatorial(n):
    if n ==1:
        return 1
    else:
        return n * fatorial (n-1)
        
```

#### Se chamamos a função fatorial(3), como resolvemos a seguinte linha?

--> ***return n * fatorial (n-1)***

#### Sabemos o valor de fatorial(2)? AINDA não!

#### Então, EMPILHAMOS a chamada de ***fatorial(2)***

```python
print(fatorial(3))
```

#### - **Recursividade:**

--> A função vai simplificando o problema em passos até que ele seja facilmente resolvível

#### Implementando um código recursivo em python para calcular o termo N da sequência de Fibonacci

```python
def fibonaci(n):
    ### Casos base: 
    if n ==0:
        return 0
    elif n ==1:
        return 1
    ### Caso recursivo (Dupla Chamada)    
    else:
        return fibonaci(n-1) + fibonaci(n-2)
```

#### 1. Crie uma função recursiva chamada soma_ate(n) que receba um número inteiro positivo n e retorne a soma de todos os números de 1 até n.

```python
def soma_ate(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return n + soma_ate(n-1)
```

```python
print(soma_ate(99))
```

#### **2.** Crie uma função recursiva chamada potencia(base, expoente) que receba dois números inteiros, base e expoente, e retorne o valor de base elevado a expoente.

```python
def potencia(base, expoente):
    if expoente ==0:
        return 1
    elif expoente ==1:
        return base
    else:
        return base * potencia(base, expoente-1)
```

```python
print(potencia(3,6))
```

#### 3. Crie uma função recursiva chamada contar_digitos(n) que receba um número inteiro positivo n e retorne a quantidade de dígitos desse número

```python
def contar_digito(n):
    if n==0:
        return 1
    elif n==1:
        return 1
    else:
        return n + contar_digito(n)
```

```python
contar_digito(4)
```

#### > Uma função recursiva resolve um problema usando uma versão menor do próprio problema

--> Para construir uma função recursiva, precisamos identificar:

#### - **1- Caso base**
        - --> Quando a função deve parar? 

#### - **2- Problema Menor**
        - --> COmo reduzir o problema a cada chamada?

#### - **3 - Combinação da resposta?**
        --> Como usar a resposta chamada recursiva para montar a resposta final? 

#### - Introdução:

- Estrutura geral de uma função recursiva:

- def funcao(problema):
    - if caso_base:
          - return resposta_simples
- **return <u>combinação da resposta com</u> funcao(problema_menor)**
