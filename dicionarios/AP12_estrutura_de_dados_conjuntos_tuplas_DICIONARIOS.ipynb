### **Exercício 1 Monitoramento de consumo de energia**

#### Uma residência inteligente registra diariamente o consumo de energia elétrica, em kWh, para identificar possíveis desperdícios.

#### Desenvolva um programa que leia a quantidade N de dias e, em seguida, o consumo registrado em cada dia, armazenando todos os valores em uma lista.

#### O programa deve:

- calcular o consumo médio diário;
- identificar o maior consumo registrado;
- informar em quais dias o consumo foi superior à média;
- exibir a quantidade de dias com consumo acima da média.
- Os dias são numerados de 1 até N, na ordem em que os consumos são informados. Considere N positivo e os consumos não negativos.

- Exemplo:

- N = 5
- Consumos: 4, 8, 3, 7, 8

#### Média: 6.00
#### Maior consumo: 8
#### Dias acima da média: 2, 4, 5
#### Quantidade: 3

```python
def lista_consumo(n_dias):
    lista_consumo_kwh = []
    for i in range(1, n_dias+1):
        valor_consumo = float(input(f"Informe o valor consumido para o dia:{i}"))
        while valor_consumo < 0:
            print(f"Valor:`{valor_consumo} não pode ser menor que zero!")
            valor_consumo = float(input(f"Informe o valor consumido para o dia:{i}"))
        lista_consumo_kwh.append(valor_consumo)
    return lista_consumo_kwh
```

```python
def estatisticas_consumo(lista):
    media_consumo = sum(lista)/len(lista)
    dias_superior_media = []
    maior_consumo = float("-inf")
    for i in range(len(lista)):
        if lista[i] > media_consumo:
            print(f"Dia:{i} superior a media:{lista[i]} > media:{media_consumo}")
            dias_superior_media.append(i+1)
        if lista[i] > maior_consumo:
            maior_consumo = lista[i]
    qtd_dias_acima_media = len(dias_superior_media)
    return media_consumo, maior_consumo, dias_superior_media, qtd_dias_acima_media
```

#### **Exercício 3 Identificação de sequências crescentes**

#### Um sistema de análise de desempenho registra, em uma lista, as pontuações obtidas por um jogador ao longo de várias partidas.

#### O sistema deseja identificar o maior período de evolução contínua, ou seja, a maior sequência de partidas consecutivas em que cada pontuação seja estritamente maior que a anterior.

#### Escreva um programa que leia um inteiro positivo N, seguido das N pontuações, armazenando-as em uma lista.

#### O programa deve identificar:

- o comprimento da maior sequência crescente consecutiva;
- a posição da partida em que essa sequência começa;
- os valores que compõem essa sequência.

#### Se houver duas sequências com o mesmo comprimento máximo, considere a que aparece primeiro. As partidas são numeradas a partir de 1.

#### Atenção: os elementos da sequência devem ocupar posições consecutivas na lista. Não é permitido reorganizar os valores.

#### Exemplo:

- Pontuações:
- 7, 8, 9, 2, 4, 6, 8, 1

- Maior sequência: [2, 4, 6, 8]
- Comprimento: 4
- Posição inicial: 4

```python
def maior_sequencia(lista):
    pos_anterior = 0
    pos_menor = ""
    maior_sequencia_encontrada = []
    lista_maior_seq = []
    for i in range(len(lista)-1):
        if lista[i] < lista[i+1]:
            lista_maior_seq.append(lista[i])
            pos_menor = i-1
        else:
            lista_maior_seq.append(lista[i])
            if len(maior_sequencia_encontrada) < len(lista_maior_seq):
                maior_sequencia_encontrada = lista_maior_seq
            lista_maior_seq = []
            pos_anterior = pos_menor
            pos_menor = ""
    if len(maior_sequencia_encontrada) < len(lista_maior_seq):
        maior_sequencia_encontrada = lista_maior_seq
        pos_anterior = pos_menor
    qtd_maior = len(maior_sequencia_encontrada)
    return maior_sequencia_encontrada, qtd_maior, pos_anterior
```

```python
def lista_medicoes():
    n_sequencias = int(input("Quantas sequências serão analisadas? "))
    lista_sequencias = []
    for i in range(1, n_sequencias+1):
        valor = float(input(f"Informe o valor da : {i}ª sequência"))
        while valor < 0:
            print(f"Valor:`{valor} não pode ser menor que zero!")
            valor = float(input(f"Informe o valor da : {i}ª sequência"))
        lista_sequencias.append(valor)
    return lista_sequencias
```

```python
lista_identificacao_sequencias = lista_medicoes()
```

```python
maior_seq, comprimento, pos_inicial_maior_seq = maior_sequencia(lista_identificacao_sequencias)
print(f"Maior sequência:{maior_seq}\nComprimento:{comprimento}\nPosição Inicial:{pos_inicial_maior_seq}")
```

#### <u>**Exercício 4 Controle de acesso a um evento** </u>

#### A organização de um evento universitário utiliza códigos de credenciamento para controlar a entrada dos participantes.

#### Antes da abertura dos portões, o sistema recebe os códigos de todos os participantes autorizados. Durante o evento, cada pessoa apresenta seu código para tentar entrar.

#### Uma entrada será autorizada somente quando:

- o código informado estiver entre os códigos autorizados;
- o participante ainda não tiver entrado no evento.

#### Escreva um programa que leia inicialmente N códigos autorizados distintos e, depois, uma quantidade M de tentativas de entrada.

#### Utilize conjuntos para controlar os participantes autorizados e aqueles que já entraram.

#### Para cada tentativa, o programa deve exibir uma das seguintes mensagens:

- ACESSO LIBERADO, quando a entrada for permitida;
- DUPLICADO, quando o participante já tiver entrado;
- NAO AUTORIZADO, quando o código não existir no credenciamento.

#### Ao final, informe:

- quantos participantes conseguiram entrar;
- quantas tentativas foram duplicadas;
- quantas tentativas utilizaram códigos não autorizados;
- quantos participantes autorizados ainda não entraram.

#### Considere que os códigos são strings e que N e M são inteiros não negativos.

```python
def conjunto_codigos(n_codigos):
    conjunto_codigos_evento = set()
    for i in range(1,n_codigos+1):
        codigo_informado = input(f"Digite o {i} código para se credenciar no evento: ").lower()
        while codigo_informado in conjunto_codigos_evento:
            print(f"O Código informado está dentro do conjunto! {codigo_informado}")
            codigo_informado = input(f"Digite o {i} código para se credenciar no evento: ").lower()
        conjunto_codigos_evento.add(codigo_informado)
    return conjunto_codigos_evento
```

```python
def controle_participantes_evento(conjunto_codigos_evento, tentativas):
    acesso_liberado = 0
    duplicados = 0
    nao_autorizado = 0
    controle_acesso_evento = []
    for i in range(1,tentativas+1):
        codigo_informado = input(f"Informe o {i} código para se ter controle ao evento!").lower()
        if codigo_informado not in conjunto_codigos_evento:
            print("O Código informado não está cadastrado no evento!")
            nao_autorizado+=1
        else:
            controle_acesso_evento.append(codigo_informado)
            contagem = controle_acesso_evento.count(codigo_informado)
            if codigo_informado in conjunto_codigos_evento and contagem ==1:
                print(f"Código informado: {codigo_informado} está cadastrado no evento e possui acesso liberado ao evento!: {contagem}")
                acesso_liberado +=1
            elif codigo_informado in conjunto_codigos_evento and contagem >1:
                print(f"Código informado: {codigo_informado} já está no evento!\nPois {contagem} é maior que 1!")
                duplicados +=1
    conjunto_participante_evento = set(controle_acesso_evento)
    qtd_partiticpantes_autorizados = len(conjunto_codigos_evento)-  len(conjunto_participante_evento)
    participantes_autorizados_mas_nao_entraram = conjunto_codigos_evento - conjunto_participante_evento
    return acesso_liberado, duplicados, nao_autorizado, qtd_partiticpantes_autorizados, participantes_autorizados_mas_nao_entraram
```

```python
n_codigos = int(input("Quantos códigos deverão ser cadastrados? "))
while n_codigos < 0:
    print("Deve ser maior que zero!")
    n_codigos = int(input("Quantos códigos deverão ser cadastrados? "))
conjunto_cod_evento = conjunto_codigos(n_codigos)
```

```python
for i in conjunto_cod_evento:
    print(i)
```

```python
tentativas = int(input("Quantas tentativas serão feitas? "))
while tentativas < 0:
    print("Deve ser maior que zero!")
    tentativas = int(input("Tentativas deverão ser maior que zero! "))
qtd_acesso_liberado, duplicados, nao_autorizado, qtd_participantes_autorizados_nao_entraram, participantes_autorizados_nao_entraram = controle_participantes_evento(conjunto_cod_evento,tentativas)
print(f"Quantidade de acessos liberados:{qtd_acesso_liberado}\nDuplicados:{duplicados}\nNão autorizados:{nao_autorizado}\nQuantidade participantes autorizados ainda não entraram:{qtd_participantes_autorizados_nao_entraram}")
```

#### <u>**Exercício 7 Controle de estoque e processamento de vendas** </u>

#### Uma pequena loja precisa controlar seu estoque de produtos durante um dia de funcionamento.

#### Cada produto possui um código único, uma quantidade disponível e um preço unitário. Essas informações devem ser armazenadas de forma que o sistema consiga localizar e atualizar rapidamente os dados de um produto a partir de seu código.

#### Inicialmente, o programa deve ler a quantidade N de produtos e os dados de cada produto.

#### Em seguida, o sistema receberá uma sequência de M operações. Cada operação possui:

- o código do produto;
- o tipo da operação;
- a quantidade de unidades envolvidas.

#### Existem dois tipos de operação:

- V: venda de unidades, reduzindo o estoque;
- R: reposição de unidades, aumentando o estoque.
- O programa deve respeitar as seguintes regras:

#### uma venda somente poderá ser realizada quando houver estoque suficiente;
#### uma venda recusada não deve alterar o estoque nem o faturamento;
#### operações com códigos inexistentes devem ser recusadas;
#### o faturamento deve considerar apenas as vendas realizadas com sucesso.

#### Ao final, o programa deve exibir:

- o faturamento total das vendas;
- a quantidade restante de cada produto;
- o valor financeiro do estoque restante.
- Os produtos devem ser apresentados na ordem em que foram cadastrados. Considere que os códigos cadastrados são distintos, as quantidades iniciais são não negativas, os preços também são não negativos, as operações utilizam apenas V ou R e suas quantidades são positivas.

```python
def cadastro_produtos(n_produtos):
    dicio_estoque = dict()
    for i in range(1, n_produtos+1):
        dicio_operacoes = {}
        codigo_produto = input(f"Informe o código do {i}º produto: ").upper()
        quantidade = int(input(f"Informe a quantidade do produto: {codigo_produto}"))
        while quantidade < 0:
            print("Não existe quantidade negativa!")
            quantidade = int(input(f"Informe a quantidade do produto: {codigo_produto}"))
        preco = float(input(f"Informe o preço do produto: {codigo_produto}"))
        while preco < 0:
            print("Não existe preço negativo!")
            preco = float(input(f"Informe o preço do produto: {codigo_produto}"))
        dicio_operacoes["QUANTIDADE"] = quantidade
        dicio_operacoes["PRECO"] = preco
        dicio_estoque[codigo_produto] = dicio_operacoes
    return dicio_estoque
```

```python
def manipulacao_estoque_dicionario(m_operacoes, dicio_estoque):
    fat_total = 0
    valor_estoque_final = 0
    quantidade_estoque_final= 0
    for i in range(1,m_operacoes+1):
        cod_produto = input("Digite o código do produto no dicionário para se realizar a operação: ").upper()
        tipo_operacao = input("Qual o tipo de operação se deseja realizar?\nV:Venda de unidades, reduzindo o estoque\nR:Reposição de unidades aumentando o estoque?").upper()
        while tipo_operacao not in("V", "R"):
            print("Apenas V OU R")
            tipo_operacao = input("Qual o tipo de operação se deseja realizar?\nV:Venda de unidades, reduzindo o estoque\nR:Reposição de unidades aumentando o estoque?").upper()
        quantidade_unidades = int(input("Quantas unidades envolvidas durante essa operação? "))
        while quantidade_unidades < 0:
            print("Não existe quantidade de unidades negativas!")
            quantidade_unidades = int(input("Quantas unidades envolvidas durante essa operação? "))
        if cod_produto not in dicio_estoque:
            print(f"O código informado:{cod_produto} não está no dicionário!")
            continue
        else:
            print(f"Código informado existe:{cod_produto}\nTipo de operação selecionada:{tipo_operacao}")
            if tipo_operacao == "V":
                print("Operação de Venda")
                for chaves in dicio_estoque:
                    dicio_itens = dicio_estoque[chaves]
                    if cod_produto == chaves:
                        for chaves, valores in dicio_itens.items():
                            if chaves == "QUANTIDADE":
                                if valores < quantidade_unidades:
                                    print(f"Não é possível realizar a operação poiss qauntidade em estoque:{valores} é menor do que a quantidade informada:{quantidade_unidades}")
                                elif valores >= quantidade_unidades:
                                    dicio_itens["QUANTIDADE"] -= quantidade_unidades
                                    valor_vendido = dicio_itens["PRECO"] * quantidade_unidades
                                    print(f"Valor vendido:{valor_vendido}")
                                    fat_total += valor_vendido
            if tipo_operacao == "R":
                print("Operação de Reposição!")
                for chave in dicio_estoque:
                    dicio_itens = dicio_estoque[chave]
                    if chave == cod_produto:
                        for chaves, valores in dicio_itens.items():
                            if chaves == "QUANTIDADE":
                                dicio_itens["QUANTIDADE"] += quantidade_unidades
    for chaves in dicio_estoque:
        dicio_itens = dicio_estoque[chaves]
        valor_estoque = 0
        for chave, valor in dicio_itens.items():
            print(f"{chave}:{valor}Unidades")
            valor_estoque = dicio_itens["QUANTIDADE"] * dicio_itens["PRECO"]
        valor_estoque_final+= valor_estoque
    return fat_total, valor_estoque_final

        

```
