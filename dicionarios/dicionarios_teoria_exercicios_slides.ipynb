#### 1. Uma loja deseja analisar os preços de seus produtos utilizando um dicionário. Crie um programa que armazene produtos e preços, calcule a média dos valores e exiba o produto mais caro, o produto mais barato e todos os produtos acima da média.




```python
dicio_loja = {
    "Produtos": [],
    "Preco": []
}
```

```python

for i in range(5):
    produtos = input("Digite o Produto para ser armazenado: ")
    dicio_loja["Produtos"].append(produtos)
    precos = float(input("Digite o preço do produto: "))
    dicio_loja["Preco"].append(precos)

print(dicio_loja)
```

```python
print(dicio_loja)
```

```python
def analisa_prorduto(dicio_loja):
    produto_mais_caro = ""
    soma = 0
    produto_mais_barato = ""
    maior_valor =  0   ### NÃO DEVE SER LISTAS
    menor_valor = float("inf") ### MENOR VALOR POSSÍVEL
    lista_acima_media = []
    
    for chave, valor in zip(dicio_loja["Produtos"], dicio_loja["Preco"]):
        if valor > maior_valor:
            maior_valor = valor
            produto_mais_caro = chave
        elif valor < menor_valor:
            menor_valor = valor
            produto_mais_barato = chave
    for i in range(len(dicio_loja["Preco"])):
        soma += dicio_loja["Preco"][i]
    media = soma/len(dicio_loja["Preco"])

    for i in range(len(dicio_loja["Produtos"])):
        if dicio_loja["Preco"] [i] > media:
            lista_acima_media.append(dicio_loja["Produtos"][i])
    return produto_mais_caro, produto_mais_barato, lista_acima_media
```

```python
produto_mais_caro, produto_mais_barato, lista = analisa_prorduto(dicio_loja)
print(f"O produto mais caro é: {produto_mais_caro}")
print(f"O produto mais barato é: {produto_mais_barato}")
print(f"A lista de produtos acima da média é: : {lista}")
```

#### 2. Um sistema de estoque armazena produtos e suas quantidades utilizando dicionários. Crie um programa que exiba a quantidade total de itens armazenados, o produto com maior estoque e o produto com menor estoque.

```python
dicio_estoque= {"Produtos":[]}
```

```python
for i in range(10):
    produtos = int(input(f"Digite a quantidade de produtos para se armazendar em {dicio_estoque["Produtos"]}"))
    dicio_estoque["Produtos"].append(produtos)
```

```python
dicio_estoque
```

```python
def analisa_quantidade(dicio_estoque):
    maior_quantidade = 0
    menor_quantidade = float("inf")

    for i in range(len(dicio_estoque["Produtos"])):
        if dicio_estoque["Produtos"][i] > maior_quantidade:
            maior_quantidade = dicio_estoque["Produtos"][i]
        elif dicio_estoque["Produtos"][i] < menor_quantidade:
            menor_quantidade = dicio_estoque["Produtos"][i]
    return maior_quantidade, menor_quantidade
```

```python
maior, menor = analisa_quantidade(dicio_estoque)
print("O Produto que tem mais quantidade é: ", maior)
print("O Produto que tem menor quantidade é: ", menor)
```

#### 3. Crie um programa que leia 10 notas utilizando uma lista, calcule a média da turma e exiba a maior nota, a menor nota, os valores acima da média e os valores abaixo da média.


```python
notas = []

for i in range(10):
    notas_alunos = float(input("Insira o valor da nota dos alunos: "))
    notas.append(notas_alunos)
print(notas)
```

```python
def funcao_nota(notas):
    maior_nota_aluno = notas[0]
    menor_nota_aluno = float("inf")
    soma = 0
    for i in range(len(notas)):
        if notas[i] > maior_nota_aluno:
            maior_nota_aluno = notas[i]
        elif notas[i] < menor_nota_aluno:
            menor_nota_aluno = notas[i]
        soma += notas[i]
    media = soma/ len(notas)
    return media, maior_nota_aluno, menor_nota_aluno
```

```python
media, maior, menor = funcao_nota(notas)
print("A média dos 10 alunos foi: ", media)
print("A menor nota foi : ", menor)
print("A maior nota foi: ", maior)
```


#### 4. Uma empresa deseja eliminar emails duplicados de sua base de dados. Crie um programa que leia vários emails, utilize conjuntos para remover repetições e exiba todos os emails únicos cadastrados.

```python
emails = []
for i in range(7):
    emails_alunos = input("Digite o email de fulano: ")
    emails.append(emails_alunos)
print(emails)
```

```python
emails_conjunto = set(emails)

def remove_duplicatas(emails_conjunto):
    repeticoes = []

    for i in range(len(emails_conjunto)-1):
        if emails_conjunto[i] == emails_conjunto[i+1]:
            repeticoes.append(emails_conjunto[i])
            emails_conjunto.remove(emails_conjunto[i])
    return repeticoes, emails_conjunto
```

```python
repeticoes, emails_conjunto = remove_duplicatas(emails_conjunto)
print("Repeticoes: ", repeticoes, emails_conjunto)
```
