### 📑📚 Exercícios de Fixação- ***Dicionários***

#### ⏳ Agora, é sua vez!:

#### Imagine um sistema simples de personagens de um jogo. Crie um programa que:
#### crie um dicionário representando um personagem e armazene:

- ■ nome
- ■ vida
- ■ nível
- ■ exiba os dados do personagem
    - ○ altere o valor da vida
    - ○ aumente o nível do personagem
    - ○ adicione um novo equipamento ao dicionário
    - ○ exiba os dados atualizados

```python
dicio_jogo ={}
```

```python
dicio_jogo["Nome"] = input("Digite o nome do personagem do jogo: ")
dicio_jogo["Vida"] = int(input(f"Digite o nível do Personagem: {dicio_jogo["Nome"]} "))
dicio_jogo["Nível"] = int(input("Digite a vida do personagem do jogo: "))
```

```python
dicio_jogo["Vida"] = 220
dicio_jogo["Nível"] = 4
dicio_jogo["Equipamento"] = "Arco e Flecha de Fogo"
print(dicio_jogo)
```

#### Explorando a Solução:

#### Observe que:
- ○ acessamos informações pelas chaves
- ○ alteramos valores existentes
- ○ adicionamos novos dados dinamicamente

#### Nesse modelo:
- ○ cada informação possui um identificador
- ○ o código fica mais legível
- ○ novas associações podem ser criadas ao longo do programa

#### ***👉 Dicionários são estruturas flexíveis para representar dados relacionados.***
