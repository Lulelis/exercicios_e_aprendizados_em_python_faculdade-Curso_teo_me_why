```python
dicio_loja = {
    "Produtos": [],
    "Preco": []
}
```

```python

for i in range(5):
    produtos = input("Digite o Produto para ser armazenado: ")
    dicio_loja["Produtos"].append(produtos)
    precos = float(input("Digite o preço do produto: "))
    dicio_loja["Preco"].append(precos)

print(dicio_loja)
```

```python
print(dicio_loja)
```

```python
def analisa_prorduto(dicio_loja):
    produto_mais_caro = ""
    soma = 0
    produto_mais_barato = ""
    maior_valor =  0   ### NÃO DEVE SER LISTAS
    menor_valor = float("inf") ### MENOR VALOR POSSÍVEL
    lista_acima_media = []
    
    for chave, valor in zip(dicio_loja["Produtos"], dicio_loja["Preco"]):
        if valor > maior_valor:
            maior_valor = valor
            produto_mais_caro = chave
        elif valor < menor_valor:
            menor_valor = valor
            produto_mais_barato = chave
    for i in range(len(dicio_loja["Preco"])):
        soma += dicio_loja["Preco"][i]
    media = soma/len(dicio_loja["Preco"])

    for i in range(len(dicio_loja["Produtos"])):
        if dicio_loja["Preco"] [i] > media:
            lista_acima_media.append(dicio_loja["Produtos"][i])
    return produto_mais_caro, produto_mais_barato, lista_acima_media
```

```python
def funcao_nota(notas):
    maior_nota_aluno = notas[0]
    menor_nota_aluno = float("inf")
    soma = 0
    for i in range(len(notas)):
        if notas[i] > maior_nota_aluno:
            maior_nota_aluno = notas[i]
        elif notas[i] < menor_nota_aluno:
            menor_nota_aluno = notas[i]
        soma += notas[i]
    media = soma/ len(notas)
    return media, maior_nota_aluno, menor_nota_aluno
```

```python
media, maior, menor = funcao_nota(notas)
print("A média dos 10 alunos foi: ", media)
print("A menor nota foi : ", menor)
print("A maior nota foi: ", maior)
```

--> Modificando a questão já que irei utilizar dicionários para: Armazenar quantas temperaturas o usuário irá digitar, e colocar cada temperatura para o mês em específico para quantos graus estava fazendo naquele dia.

```python
lista_temperatura = []

n = int(input("Quantas temperaturas serão armazenadas no dicionário? "))
for i in range(n):
    dicio_temp = {}
    mes = input("Qual o mês informado?")
    dia_mes = int(input("Qual o dia do mês informado? "))
    valor_temperatura = float(input(f"Digite o valor aferido de temperatura para o {dia_mes} do mês {mes}:  "))
    dicio_temp["Mês"] = mes
    dicio_temp["Dia_do_Mês"] = dia_mes
    dicio_temp["Temperatura"] = valor_temperatura
    lista_temperatura.append(dicio_temp)

print(lista_temperatura)

```

```python
def analisa_temp(lista):
    maior_temp = float("-inf")
    menor_temp = float("inf")
    dia_mais_quente = float("-inf")
    mes_mais_quente = ""
    mes_mais_frio = ""
    soma = 0

    for valores in lista:
        mes = valores["Mês"]
        dia_mes = valores["Dia_do_Mês"]
        temperatura = valores["Temperatura"]
        if maior_temp < temperatura:
            maior_temp = temperatura
            dia_mais_quente = dia_mes
            mes_mais_quente = mes
        elif menor_temp > temperatura:
            menor_temp = temperatura
            mes_mais_frio =  mes
        soma += temperatura
    media_temperatura = soma / len(lista)
    return maior_temp, menor_temp, dia_mais_quente, mes_mais_quente, mes_mais_frio, media_temperatura
```

```python
def conta_palavra(lista):
    dicio_palavras = {}
    for i in range(len(lista)):
        if lista[i] not in dicio_palavras:
            dicio_palavras[lista[i]] = 1
        elif lista[i] in dicio_palavras:
            dicio_palavras[lista[i] ] +=1
    letra_mais_frequente = 0
    qual_letra = 0
    for chave, valor in dicio_palavras.items():
        if valor > letra_mais_frequente:
            letra_mais_frequente = valor
            qual_letra = chave
    qtd_letras_diferentes = len(dicio_palavras)
    return dicio_palavras, qtd_letras_diferentes, qual_letra, letra_mais_frequente
```

```python
print(conta_palavra(lista_palavra))
```

```python
def analista_temperaturas(dicio):
    dia_mais_quente = []
    dia_mais_frio = []
    maior_temp = float("-inf")
    menor_temp = float("inf")
    dicio_resultados_dias = dict()
    dicio_resultados_medias = dict()

    for chave, valor in dicio.items():
        soma_temp = 0
        count_temp = 0
       
        for temp in valor:
            soma_temp += temp
            count_temp +=1
            if temp > maior_temp:
                maior_temp = temp
                dia_mais_quente = [chave]
            elif temp == maior_temp:
                dia_mais_quente.append(chave)
            if temp < menor_temp:
                menor_temp = temp
                dia_mais_frio = [chave]
            elif temp == menor_temp:
                dia_mais_frio.append(chave)
        media_temp = soma_temp/ count_temp
        dicio_resultados_medias[chave] = media_temp
    dicio_resultados_dias["Dia mais quente"] = dia_mais_quente
    dicio_resultados_dias["Dia mais frio"] = dia_mais_frio

    maior_media = float("-inf")
    menor_media = float("inf")
    for chave, valor in dicio_resultados_medias.items():
        if valor > maior_media:
            maior_media = valor
        if valor < menor_media:
            menor_media = valor
    dicio_resultados_medias["Maior_Media_Temperatura"] = maior_media
    dicio_resultados_medias["Menor_Media_Temperatura"] = menor_media


    return dicio_resultados_medias, dicio_resultados_dias
```

```python
dicio_medias, dicio_resultados_dias = analista_temperaturas(dicio_temp)
print("O dicionário com as médias é:", dicio_medias)
print("O dicionário com os resultados são:", dicio_resultados_dias)
```
