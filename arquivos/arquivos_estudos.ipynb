### **O Ciclo de Vida de Arquivos: Básico** 🗂️

#### Para manipular qualquer arquivo no seu código, o Python sempre segue uma regra de três passos:
- ***○ Abrir (open):*** Cria uma ponte entre o seu script Python e o arquivo no sistema operacional.
- ***○ Processar (Ler/Escrever):*** Transfere os dados através dessa ponte.
- ***○ Fechar (close):*** Corta a ponte, salvando as alterações e liberando a memória
do sistema operacional.

### Abrindo no modo de escrita ('w'):

arquivo = open("exemplo.txt", "w", encoding = "utf-8")

### Escrevendo algo

arquivo.write("Olá Mundo!")

### Fechar é obrigatório :

arquivo.close()

### **Abrindo Arquivos**

#### -A Função open() e seus Modos
- ○ 'r' (Read): Modo de leitura. O arquivo precisa já existir, ou o Python gera um erro (FileNotFoundError).

- ○ 'w' (Write): Modo de escrita. Se o arquivo já existir, ele apaga todo o conteúdo e começa do zero. Se não existir, ele cria um arquivo novo.

- ○ 'a' (Append): Modo de adição. Ele abre o arquivo e posiciona o "cursor" no final. Adiciona-se ao conteúdo existente, sem apagar nada.

### **Fechando os arquivos**

#### A Função close() é de uso obrigatório

- ○ No modo escrita, ao não fechar um arquivo a escrita não acontece no disco.
Perde-se dados.
- ○ Adicionalmente, ele fica bloqueado para outros acessos.
- ○ No modo leitura, não fechar um arquivo consome uma unidade do limite
máximo de arquivos que um único programa pode manter aberto (por
exemplo, 1024 no linux).

### Boa prática: Uso do ***With***

#### Gerenciador de contexto with 

- ○ Se ocorrer algum erro antes da chamada à função close(), os problemas do não fechamento podem ocorrer 

- o with garante que o arquivo será fechado automaticamente assim que o bloco de código terminar, mesmo que ocorra um erro no meio do caminho.

```python

lista_placa= []
with open("placas.txt", "w", encoding="utf-8") as lista_tupla:
    for i in range(4):
        placa=input("Ensira as placas: ")
        modelo = input("Digite o Modelo de carro")
        lista_temp = []
        lista_temp.append(placa)
        lista_temp.append(modelo)
        tupla = tuple(lista_temp)
        lista_placa.append(tupla)
    lista_tupla.write(str(lista_placa)+"\n")
```

```python
print(lista_placa)
```

#### Abrindo com "R"

```python
lista_recuperada = []
with open ("placas.txt", "r") as lista_tupla:
    for i in lista_tupla:
            lista_recuperada.append(i.strip())
print(lista_recuperada)
```

### Forma mais comum e usual para ***Escrever Arquivos em python*** | **Leitura**

```python
nome_arquivo_2 = "letra_nusica_2.txt"

letra2 = input("Digite a segunda letra!")
with open(nome_arquivo_2, mode="w", encoding="utf-8") as letra_2:
    letra_2.write(letra2)
```

### Forma mais comum e usual para ***Escrever Arquivos em python sem SOBREESCREVER*** | **"A"** | Escrevendo sem sobreescrever o que já exite


```python
### Escrevendo a partir do último caractere
nome_arquivo_2 = "letra_nusica_2.txt"

letra2 = input("Digite a segunda letra!"+ "\n")
with open(nome_arquivo_2, mode="a", encoding="utf-8") as letra_2:
    letra_2.write(letra2)
```

### **Lendo Arquivos CSV com WITH and Open**

```python
arquivo_csv = "heart.csv"

with open(arquivo_csv, encoding="utf-8") as planilha_csv:
    data = planilha_csv.readlines()
print(data)
```

--> Cada linha da planilha recebe o nome de: registro, objeto, tupla

```python
for i in range(len(data)):
    print(i)
    print(data[i])
```

#### > Passando as informações das Linhas da planilha para um dicionário aonde cada chave receberá uma lista de valores das informações respectivas das linhas

```python
chaves = data[0].strip("\n").split(",")
chaves
```

```python
dados_heart = dict()
```

```python
for i in chaves:
    dados_heart[i] = []
dados_heart
```

#### Percorrendo as demais linhas e adicionando os valores nas listas de valores do dicionário?:

```python
for i in data[1:]: ### Percorrendo todos as linhas a partir da segunda linha pq a primeira é o cabeçalho da planilha
    valores  = i.strip("\n").split(",")
    for j in range(len(valores)):
        dados_heart[chaves[j]].append(valores[j])
```

```python
dados_heart
```

```python
for chave, valor in dados_heart.items():
    print(chave, "-->", valor)
```
