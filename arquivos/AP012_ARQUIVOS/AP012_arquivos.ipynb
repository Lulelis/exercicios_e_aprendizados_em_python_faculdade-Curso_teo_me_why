#### <u>**Exercício 1** Criando e recuperando os primeiros contatos</u>

#### A primeira versão da agenda deverá permitir registrar contatos e recuperar os dados posteriormente, a partir de um arquivo de texto.

#### Escreva um programa que leia uma quantidade N de contatos e, para cada contato, solicite nome, telefone e e-mail.

#### Todos os registros deverão ser gravados no arquivo agenda.txt, utilizando um contato por linha e separando os três campos por ponto e vírgula.

#### Depois de concluir a gravação, o programa deverá abrir novamente o arquivo para leitura e apresentar todos os contatos cadastrados.

#### Considere N positivo e todos os campos preenchidos, sem ponto e vírgula em seu conteúdo.

#### - Exemplo de conteúdo do arquivo:

- Ana Silva;31988881111;ana@exemplo.test
- Bruno Costa;31988882222;bruno@exemplo.test
- Carla Lima;31988883333;carla@exemplo.test

#### **Atenção:** nesta primeira versão, o programa deverá criar o arquivo do zero. Nas próximas questões, você deverá preservar os contatos já registrados.

```python
def verifica_caractere(palavra):
    tem_ponto_e_virgula = False
    for i in range(len(palavra)):
        for j in palavra[i]:
            if j == ";":
                tem_ponto_e_virgula = True
    return tem_ponto_e_virgula
```
