#### Todos os registros deverão ser gravados no arquivo agenda.txt, utilizando um contato por linha e separando os três campos por ponto e vírgula.

#### Depois de concluir a gravação, o programa deverá abrir novamente o arquivo para leitura e apresentar todos os contatos cadastrados.

#### Considere N positivo e todos os campos preenchidos, sem ponto e vírgula em seu conteúdo.

#### - Exemplo de conteúdo do arquivo:

- Ana Silva;31988881111;ana@exemplo.test

#### **Atenção:** nesta primeira versão, o programa deverá criar o arquivo do zero. Nas próximas questões, você deverá preservar os contatos já registrados.

```python
def verifica_caractere(palavra):
    tem_ponto_e_virgula = False
    for i in range(len(palavra)):
        for j in palavra[i]:
            if j == ";":
                tem_ponto_e_virgula = True
    return tem_ponto_e_virgula
```

#### **<u>Exercício 3 Alteração e exclusão de contatos**</u>

#### Até o momento, a agenda permite criar, consultar e acrescentar registros. Entretanto, os dados de um contato podem mudar ou precisar ser removidos.

#### Em um arquivo de texto, não podemos simplesmente alterar um campo e esperar que todos os demais registros sejam reorganizados automaticamente.

#### Desenvolva um programa que carregue os contatos do arquivo agenda.txt para um dicionário e permita executar uma das seguintes operações:

- A - Alterar: localizar um telefone cadastrado e atualizar o nome e o e-mail associados a ele.
- E - Excluir: localizar um telefone cadastrado e remover completamente o contato.
- O telefone será utilizado para identificar o contato e não poderá ser modificado na operação de alteração.

#### Se o telefone informado não existir, o programa deverá apresentar uma mensagem e manter o arquivo inalterado.

#### Após uma operação realizada com sucesso, todas as informações da agenda deverão ser gravadas novamente no arquivo, refletindo a alteração ou exclusão.

#### Ao final, exiba a quantidade de contatos que permaneceram cadastrados.

```python
def alterar_arquivo(dicio_contatos, nome_arquivo):
    lista_tupla = []
    for chave_dicio, dados in dicio_contatos.items():
        lista_tupla.append((chave_dicio,dados["nome_contato"], dados["email"] ))
    print(f"Dados desempacotados na lista de tupla: {lista_tupla}")
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo_agenda:
        for chave_tel, nome,  email in lista_tupla:
                arquivo_agenda.write(f"{nome};{chave_tel};{email}\n")
    print("Dados salvos com sucesso no arquivo!")
```

```python
def excluir_contato_arquivo(dicio_contatos, nome_arquivo):
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo_agenda:
        for chave_dicio, dados in dicio_contatos.items():
            arquivo_agenda.write(f'{dados["nome_contato"]};{chave_dicio};{dados["email"]}\n')
    print("Dados excluídos do arquivo com sucesso!")
```

```python
### Função para Alteração de Contatos:
def alterar_arquivo(dicio_contatos, nome_arquivo):
    lista_tupla = []
    for chave_dicio, dados in dicio_contatos.items():
        lista_tupla.append((chave_dicio,dados["nome_contato"], dados["email"] ))
    print(f"Dados desempacotados na lista de tupla: {lista_tupla}")
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo_agenda:
        for chave_tel, nome,  email in lista_tupla:
                arquivo_agenda.write(f"{nome};{chave_tel};{email}\n")
    print("Dados salvos com sucesso no arquivo!")
```

```python
### Função para  Exclusão de Contatos:
def excluir_contato_arquivo(dicio_contatos, nome_arquivo):
    with open(nome_arquivo, "w", encoding="utf-8") as arquivo_agenda:
        for chave_dicio, dados in dicio_contatos.items():
            arquivo_agenda.write(f'{dados["nome_contato"]};{chave_dicio};{dados["email"]}\n')
    print("Dados excluídos do arquivo com sucesso!")
```
