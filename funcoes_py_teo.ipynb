#### Ex: Criando uma função para juros compostos

```python
def juros_compostos(capital_inicial: float,taxa_juros: float,tempo: int):
    #### Adicionando docstring

    """
    Juros Compostos Servem para Calcular o Retorno Financeiro a partir de um aporte deve
    se considerar o valor a taxa de juros atual e o tempo (em anos) para Cálculo do valor
    a ser retornado
    aporte:
      um número inteiro que represente o valor me R$
    Taxa: 
    um número Float entre 0 E 1 que representa o valor taxa de juros

    anos:
    Um número inteiro >=1 que representa o tempo que o investimento terá liquidez
 
    """
    calculo_taxa= (1 + taxa_juros) ** tempo
    Montante = capital_inicial * calculo_taxa

    return Montante
    
```

> Docstring --> Definição e espeficicação de dados da Função

```python
capital_inicial  = float(input("Intforme o capital Inicial: "))
taxa = float(input("Informe a taxa "))
tempo = int(input("Informe o período temporal"))

### A ordem altera a forma como se está executando uma função, ou seja depende dos parâmetros
round(juros_compostos(capital_inicial, taxa, tempo),2)

#### Taxa de juros e Capital Inicial Variam
```

Exercício 2.2- Criando uma função para verificar se o um número é par ou ímpar; Exibir o resultado de acordo com a saída:
- O número x é ímpar
- O número x é par

```python

numero = int(input("Digite um número: "))

if numero %2 ==0:
    print("O número é par")
else:
    print("O número é ímpar")
```

```python
def par_ou_impar(numero:int):

    if numero %2 ==0:
        return "O número é par!"
    else:
        return "O número é impar"
    
```

```python
numero = int(input("Digite um número! "))
par_ou_impar(numero)
```

#### Função de soma simples

```python
def soma(num_1:float, num_2:float):
    soma = num_1 + num_2
    return soma
```

#### Utilizando a função de soma para DOIS números dentro de outra função

```python
def media(num_1:int,num_2:int, contagem_numeros:int):

    return soma(num_1, num_2)/contagem_numeros
```
