####  <u>**🔹 1. Números pares e média** </u>

#### Enunciado de prova:  

#### Um programa deve ler uma sequência de números inteiros digitados pelo usuário. A entrada termina quando o valor -1 for informado.

- Ao final, o programa deve:

-calcular e mostrar a média dos números digitados (sem incluir o -1),

#### mostrar quantos números pares foram informados.

#### Atenção:

#### Se nenhum número for digitado antes do -1, o programa deve avisar que não há dados para calcular a média.

### O valor -1 não deve ser considerado nos cálculos.

#### Usando funções + listas:

```python


def calcular_media(lista):
    soma = 0
    for i in lista:
        soma +=i
    media = soma / len(lista)
    return media

def numeros_pares(lista):
    count_pares = 0
    for i in lista:
        if i % 2==0:
            count_pares +=1
    return count_pares

numeros = int(input("Digite uma sequência de Números inteiros"))
lista_numeros = []
while numeros != -1:
    
    lista_numeros.append(numeros)
    numeros = int(input("Digite uma sequência de Números inteiros"))
print(lista_numeros)
if lista_numeros ==[]:
    print("Lista Vazia. Não tem como calcular nada!")
else:

    media_lista_numeros = calcular_media(lista_numeros)
    print(f"A média dos números informados é {media_lista_numeros}")
    contagem_pares = numeros_pares(lista_numeros)
    print("A contagem de pares existentes é: ", contagem_pares)
```
