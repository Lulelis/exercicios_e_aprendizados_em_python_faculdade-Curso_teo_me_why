#### <u>**3.**</u> Crie um programa que leia 20 números inteiros e os armazene em uma lista. Crie uma função eliminar_duplicatas(lista) que retorne uma nova lista contendo os mesmos elementos, mas mantendo apenas a primeira ocorrência de cada número (preservando a ordem original). Observação: Não utilize conjuntos (sets) para resolver a questão.

```python
lista_num_inteiro =[]

for i in range(20):
    numeros = int(input("Digite um número inteiro para se armazernar na lista: "))
    lista_num_inteiro.append(numeros)
print(lista_num_inteiro)
```

```python
def elemina_duplicatas(lista):
    nova_lista_inteiro = []
    for i in range(len(lista)):
        if lista[i] not in nova_lista_inteiro:
            nova_lista_inteiro.append(lista[i])
    return nova_lista_inteiro
```

```python
nova_lista_sem_duplicata = elemina_duplicatas(lista_num_inteiro)
print(nova_lista_sem_duplicata)
```

#### **26.** Desenvolva um programa que leia os preços de N produtos em um carrinho de compras. Crie uma função que aplique um desconto progressivo na lista: 5% para itens acima de R$ 100, e 10% para itens acima de R$ 500. A função deve retornar a lista de preços atualizada e o valor total economizado.


```python
lista_produtos = []
for i in range(10):
    valor = float(input("Digite o valor para compor a lista:"))
    lista_produtos.append(valor)
print(lista_produtos)
```

```python
def economiza(lista):
    lista_valores_economizados =[]
    for i in range(len(lista)):
        ### '10% de Desconto: pois é acumulativo
        if lista[i] > 500:
            valor_original = lista[i]
            lista[i] = valor_original - 0.10 * lista[i]
            lista_valores_economizados.append(valor_original - lista[i])
        elif lista[i] > 100:
            valor_original = lista[i]
            lista[i] = valor_original - 0.05 * lista[i]
            lista_valores_economizados.append(valor_original - lista[i])
    valor_economizado = sum(lista_valores_economizados)
    return  lista, valor_economizado

```

```python
economiza(lista_produtos)
```

#### **28.** Crie uma função que receba duas listas ordenadas de inteiros (tamanhos diferentes) e retorne uma terceira lista, também ordenada, contendo a mescla dos elementos das duas listas, sem usar a função sort() do Python.


```python
lista = [1,2,3,4,5,6,7,8,9,22,34,567,6060]

```

```python
def sorteadora(lista):
    aux= 0
    for i in range(len(lista)-1):
        if lista[i+1] <lista[i]:
            aux = lista[i+1]
            lista[i+1] = lista[i]
            lista[i] = aux
    return lista

```

```python
sorteadora(lista)
```

```python


lista_1 = [-1, -3, -90, 0, 2.3, 39.5, 34, 3, 5, 4,3]
lista_2 = [2, 3, 4,-90, -23,-3,-1, 0]

def ordernar_listas(lista_1,lista_2):
    nova_lista_ordenada = []
    sorteadora(lista_1)
    sorteadora(lista_2)
    for i in range(len(lista_1)):
        nova_lista_ordenada.append(lista_1[i])
    sorteadora(nova_lista_ordenada)
    for i in range(len(lista_2)):
        nova_lista_ordenada.append(lista_2[i])
    sorteadora(nova_lista_ordenada)
    return nova_lista_ordenada
```

#### **38.** Ao manipular laços while e for para percorrer Listas e Matrizes, avalie as assertivas:

- I. A construção for linha in M: itera sobre cada elemento numérico individual
de M de forma linear.

- II. Para alterar o valor interno de uma célula e refletir essa alteração na estrutura
global, deve-se realizar o percurso utilizando índices numéricos.

- III. O uso da função embutida len(M) em uma matriz quadrada N × N retorna
a quantidade total de células (N × N).

#### É correto o que se afirma em:
#### A) Apenas II.
#### B) Apenas I e II.
#### C) Apenas I e III.
#### D) Apenas II e III.
#### E) I, II e III.

#### **Resposta- Letra A**- *Apenas a II*

#### Justificatica:

- I. *“A construção for linha in M: itera sobre cada elemento numérico individual de M de forma linear.”*

Na prática: Se M é uma matriz representada como lista de listas, o for linha in M percorre cada linha (ou seja, cada sublista), não cada número individual.

Para acessar cada número, seria necessário um segundo for dentro desse loop.

#### Conclusão: Essa assertiva está falsa porque não percorre cada elemento numérico individual, apenas cada linha.

- II. *“Para alterar o valor interno de uma célula e refletir essa alteração na estrutura global, deve-se realizar o percurso utilizando índices numéricos.”*
Correto: Se você quer mudar um valor específico dentro da matriz, precisa acessar por índice, tipo M[i][j].

Usar apenas for linha in M não permite alterar diretamente uma célula específica sem índices.

#### Conclusão: Essa assertiva está verdadeira.

- III. “O uso da função embutida len(M) em uma matriz quadrada N × N retorna a quantidade total de células (N × N).”
Na prática: len(M) retorna apenas o número de linhas da matriz, ou seja, N.

Para obter o total de células, seria necessário N * N.

#### Conclusão: Essa assertiva está falsa.
