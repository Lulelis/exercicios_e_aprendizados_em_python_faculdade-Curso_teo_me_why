Escreva uma função que receba essa lista e calcule a distância total percorrida. Para simplificar, considere a distância entre os pontos (x1, y1) e (x2, y2) como (x2 − x1)2 + (y2 − y1)2.

#### lista de tuplas = [(latitude_1, longitude_1), (latidude_2, longitude_2)] é ***DIRERENTE*** de Tuplas de listas = ([latitude_1, longitude_1], [latitude_2, longitude_2])

```python
import math
```

```python
def calcula_distancia(lista_coordenadas):
    distancia_latitudes =  (lista_coordenadas[0][0] - lista_coordenadas[1][0])
    distancia_longitudes = lista_coordenadas[0][1] - lista_coordenadas[1][1]
    distancia_entre_pontos = math.pow(distancia_latitudes,2) + math.pow(distancia_longitudes,2)
    calculo_distancia = math.sqrt(distancia_entre_pontos)
    return calculo_distancia
```

```python
latitude_1 = input("Informe a coordenada na posição latitude 1: ")
latitude_1 = float(latitude_1)
longitude_1 = input("Informe a coordenada na posição longitude_1: ")
longitude_1 = float(longitude_1)
latitude_2 = input("Informe a coordenada na posição latitude 2: ")
latitude_2 = float(latitude_2)
longitude_2= input("Informe a coordenada na posição longitude 2: ")
longitude_2 = float(longitude_2)
coordenada_1 = (latitude_1, longitude_1)
coordenada_2 = (latitude_2, longitude_2)
lista_coordenadas= [coordenada_1, coordenada_2]
calculo_distancias = calcula_distancia(lista_coordenadas)
print(f"O cálculo da distância é {calculo_distancias}")
```

```python
def descarta_notas(notas_dos_atletas):
    lista_notas_jurados = list(notas_dos_atletas)
    maior_nota = lista_notas_jurados[0]
    menor_nota = lista_notas_jurados[0]
    for i in range(len(lista_notas_jurados)):
        if lista_notas_jurados[i] > maior_nota:
            maior_nota = lista_notas_jurados[i]
        elif lista_notas_jurados[i] < menor_nota:
            menor_nota = lista_notas_jurados[i]

    lista_notas_jurados.remove(maior_nota)
    lista_notas_jurados.remove(menor_nota)

    soma_notas = 0
    for i in lista_notas_jurados:
        soma_notas +=i
    media_notas = soma_notas/len(lista_notas_jurados)
    return media_notas
```

```python



def calcular_prazo(entrega_prevista, entrega_real):
    nova_lista_classificacao = []
    atrasadas =  []

    for previsto, real in zip(entrega_prevista, entrega_real):
        if previsto > real:
            nova_lista_classificacao.append("Adiantada")
        elif previsto < real:
            nova_lista_classificacao.append("Atrasado")
            atrasadas.append(real-previsto)
        else:
            nova_lista_classificacao.append("No Prazo")
    soma = 0
    for i in range(len(atrasadas)):
        soma += atrasadas[i]
    media = soma/ len(atrasadas)
    return nova_lista_classificacao, media
```

#### <u>**42.** Considere um programa desenvolvido para remover valores repetidos de uma lista, utilizando a conversão para conjunto:</u>

- 1 numeros = [4, 2, 4, 8, 2, 10, 8]
- 2
- 3 sem_repeticao = set(numeros)
- 4 resultado = list(sem_repeticao)
- 5
- 6 print(len(resultado))

#### Qual a saída deste bloco de código?
- A) 7
- B) 4
- C) 3
- D) [4, 2, 8, 10]
- E) O código lançará um erro, pois listas não podem ser convertidas em conjuntos.
15

#### **Resposta** --> **Letra B-)** 4

É possível converter uma lista em conjunto set() e depois converter novamente de conjunto para lista (*list*)

- primeira saída linha 3 = set ={2,4,8,10}
- segunda saída linha 4 = list[2,4,8,10]
- terceira saída linha 6 = 4


#### Letra A (7)

- Errada porque o número 7 seria o tamanho da lista original, antes de remover duplicatas.
O código imprime o tamanho depois da conversão para conjunto.

####  Letra C (3)
- Errada porque há quatro valores únicos (2, 4, 8, 10), não três.

####  Letra D ([4, 2, 8, 10])
- Errada porque o código imprime len(resultado), ou seja, o tamanho da lista, não o conteúdo.
Se fosse print(resultado), aí sim o resultado seria uma lista parecida com essa (a ordem pode variar, pois conjuntos não mantêm ordem).

####  Letra E (erro)
- Errada porque listas podem ser convertidas em conjuntos sem problema algum.
O Python só lança erro se você tentar converter algo não iterável ou com elementos não “hasháveis” (como outras listas dentro da lista).

#### **A conversão de volta para lista mantém apenas os valores únicos.**
