Um trecho crescente consecutivo é formado por números em que cada elemento é estritamente maior que o elemento imediatamente anterior.

#### Desenvolva um programa que determine o tamanho do maior trecho crescente consecutivo existente na sequência.

#### Por exemplo:
#### Entrada:
- 3
- 5
- 7
- 2
- 4
- 6
- 8
- 1
- 0

#### Saida:
#### Maior trecho crescente: 4
#### Nesse exemplo, o maior trecho é formado por:

- 2, 4, 6, 8

#### Caso apenas um número seja informado antes da sentinela, o maior trecho terá tamanho 1.

```python
def leitura_ciclo(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada > anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
        anterior = entrada
    print(f"Maior trecho da sequência: {maior_sequencia}")
```

- • o maior número de repetições consecutivas encontrado;

#### Por exemplo, para a sequência:

```python
def sequencia_iguais(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    numero_maior_sequencia = entrada
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada == anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
            numero_maior_sequencia = entrada
        anterior = entrada
    print(f"Maior trecho da sequência de números repetidos:{maior_sequencia}\nNúmero que se repete: {numero_maior_sequencia}")
```

#### Enquanto B for maior que zero, se B for ímpar, o valor atual de A é acrescentado ao resultado. Em seguida, A é duplicado e B é substituído por sua divisão inteira por 2.

#### Por exemplo, para calcular 13 × 11, os pares de valores processados são (13, 11), (26, 5), (52, 2) e (104, 1). Como os valores 11, 5 e 1 são ímpares, o resultado é obtido pela soma 13 + 26 + 104 = 143.

#### Escreva um programa que leia dois números inteiros, que podem ser positivos, negativos ou zero, e calcule seu produto utilizando esse procedimento. O programa deve tratar adequadamente o sinal do resultado e informar também quantas etapas foram necessárias para concluir o processamento. O operador de multiplicação * não pode ser utilizado durante o cálculo

```python
def processador_embarcado_calculo(valor_a, valor_b):
    etapas = 0
    resultado = 0
    while valor_b > 0:
        if valor_b %2 !=0:
            resultado += valor_a
        valor_a = valor_a + valor_a
        valor_b = valor_b//2
        etapas +=1
    print(f"Foram necessárias: {etapas} etapas")
    return resultado
```
