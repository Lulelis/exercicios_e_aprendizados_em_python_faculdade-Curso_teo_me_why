#### 14. Peça ao usuário que informe uma sequência de números inteiros. A leitura deve ser encerrada quando o valor 0 for digitado. O valor 0 serve apenas como sentinela e não deve fazer parte da sequência. Um trecho crescente consecutivo é formado por números em que cada elemento é estritamente maior que o elemento imediatamente anterior.

#### Desenvolva um programa que determine o tamanho do maior trecho crescente consecutivo existente na sequência.

#### Por exemplo:
#### Entrada:
- 3
- 5
- 7
- 2
- 4
- 6
- 8
- 1
- 0

#### Saida:
#### Maior trecho crescente: 4
#### Nesse exemplo, o maior trecho é formado por:

- 2, 4, 6, 8

#### Caso apenas um número seja informado antes da sentinela, o maior trecho terá tamanho 1.

```python
def leitura_ciclo(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada > anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
        anterior = entrada
    print(f"Maior trecho da sequência: {maior_sequencia}")
```

```python
entrada = int(input("Digite o primeiro número da sequência!"))
print(leitura_ciclo(entrada))
```

#### **15.** Um equipamento registra continuamente códigos numéricos inteiros. A leitura termina quando o valor 0 é informado.

#### Deseja-se identificar repetições consecutivas de um mesmo código. Desenvolva um programa que determine:
- • o maior número de repetições consecutivas encontrado;
- • qual foi o número que formou essa maior sequência.

#### Por exemplo, para a sequência:

- 4, 4, 7, 7, 7, 2, 2, 5, 5, 5, 5, 3, 0 o resultado deve ser:

- Numero: 5
- Maior sequencia: 4

#### Se houver empate entre duas sequências de mesmo tamanho, considere aquela que apareceu primeiro.

```python
def sequencia_iguais(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    numero_maior_sequencia = entrada
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada == anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
            numero_maior_sequencia = entrada
        anterior = entrada
    print(f"Maior trecho da sequência de números repetidos:{maior_sequencia}\nNúmero que se repete: {numero_maior_sequencia}")
```

```python
entrada = int(input("Digite o primeiro número da sequência!"))
print(sequencia_iguais(entrada))
```

#### No primeiro intervalo foi registrada uma requisição e, no segundo, também uma requisição. A partir do terceiro intervalo, a quantidade registrada é sempre igual à soma das quantidades observadas nos dois intervalos imediatamente anteriores. Dessa forma, os primeiros valores são

- 1, 1, 2, 3, 5, 8, 13, 21, . . .

#### Escreva um programa que leia um número inteiro positivo N, representando a quantidade de intervalos que devem ser analisados, e a capacidade máxima de requisições que o servidor consegue processar em um único intervalo. O programa deve gerar os N primeiros valores dessa sequência e informar o valor correspondente ao N-ésimo intervalo, a soma das requisições registradas em todos os intervalos e o primeiro intervalo em que a quantidade de requisições ultrapassou a capacidade do servidor.

#### Caso isso não aconteça nos N intervalos analisados, o programa deve informar que a capacidade não foi ultrapassada. Os casos N = 1 e N = 2 também devem ser tratados corretamente.

```python
def infraestrutura_servidor(qtd_intervalos, num_requisicoes):
    soma_intervalo= 0
    intervalos_limites = ""
    count = 0
    count_valores =0
    for i in range(1, qtd_intervalos+1):
        count +=1
        count_valores += i
        print(f"Intervalo: {i}: count valores: {count_valores}")
        soma_intervalo +=count_valores
        if soma_intervalo > num_requisicoes:
            intervalos_limites = i
    if intervalos_limites == "" or count <=2:
        print("Não ultrapassou a capacidade de requisições do servidores")
    else:
        print(f"Soma dos intervalos: {soma_intervalo}\nIntervalo Limite: {i}")
```

```python
quantidade_intervalos = int(input("Digite a quantidade de Intervalos: "))
while quantidade_intervalos < 0:
    print("Deve ser maior que zero")
    quantidade_intervalos = int(input("Digite a quantidade de Intervalos: "))
capacidade_requisicoes = int(input("Digite a capacidade máxima de requisições: "))
while capacidade_requisicoes < 0:
    print("Deve ser maior que zero!")
    capacidade_requisicoes = int(input("Digite a capacidade máxima de requisições: "))
infraestrutura_servidor(quantidade_intervalos, capacidade_requisicoes)

    
```
