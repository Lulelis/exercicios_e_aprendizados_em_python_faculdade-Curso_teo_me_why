O valor 0 serve apenas como sentinela e não deve fazer parte da sequência. Um trecho crescente consecutivo é formado por números em que cada elemento é estritamente maior que o elemento imediatamente anterior.

#### Desenvolva um programa que determine o tamanho do maior trecho crescente consecutivo existente na sequência.

#### Por exemplo:
#### Entrada:
- 3
- 5
- 7
- 2
- 4
- 6
- 8
- 1
- 0

#### Saida:
#### Maior trecho crescente: 4
#### Nesse exemplo, o maior trecho é formado por:

- 2, 4, 6, 8

#### Caso apenas um número seja informado antes da sentinela, o maior trecho terá tamanho 1.

```python
def leitura_ciclo(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada > anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
        anterior = entrada
    print(f"Maior trecho da sequência: {maior_sequencia}")
```

```python
entrada = int(input("Digite o primeiro número da sequência!"))
print(leitura_ciclo(entrada))
```

#### Deseja-se identificar repetições consecutivas de um mesmo código. Desenvolva um programa que determine:
- • o maior número de repetições consecutivas encontrado;
- • qual foi o número que formou essa maior sequência.

#### Por exemplo, para a sequência:

- 4, 4, 7, 7, 7, 2, 2, 5, 5, 5, 5, 3, 0 o resultado deve ser:

- Numero: 5

```python
def sequencia_iguais(entrada):
    atual_sequencia = 1
    maior_sequencia = 1
    numero_maior_sequencia = entrada
    anterior = entrada
    while entrada !=0:
        print("Digite o próximo número da sequência: ")
        entrada = int(input("Digite o próximo número da sequência!"))
        print(f"Número Digitado: {entrada} anterior: {anterior}")
        if entrada == anterior:
            atual_sequencia +=1
        else:
            atual_sequencia =1
        if atual_sequencia > maior_sequencia:
            maior_sequencia = atual_sequencia
            numero_maior_sequencia = entrada
        anterior = entrada
    print(f"Maior trecho da sequência de números repetidos:{maior_sequencia}\nNúmero que se repete: {numero_maior_sequencia}")
```

Para analisar mudanças bruscas no comportamento observado, os pesquisadores desejam identificar os picos locais da série. Uma medição é consideradaum pico local quando sua temperatura é estritamente maior que a medição imediatamente anterior e também estritamente maior que a medição imediatamente posterior. Na sequência:

#### > 18, 21, 19, 22, 27, 25, 20,

#### por exemplo, as temperaturas 21 e 27 são picos locais. A primeira e a última medição nunca podem ser classificadas dessa forma, pois não possuem dois vizinhos. Ao final, informe quantos picos locais foram encontrados e, caso exista pelo menos um, apresente também o valor do maior pico e sua posição, considerando que a primeira medição ocupa a posição 1. Se nenhum pico for encontrado, o programa deverá informar essa situação.

```python
def pico_estacao_metereologica(n_temperaturas):
    maior_valor = float("-inf")
    picos_encontrados = 0
    for i in range(1, n_temperaturas+1):
        temperaturas = int(input(f"Digite a {i} temperatura: "))
        anterior = temperaturas
        i-=1
        atual = int(input(f"Digite a próxima {i}temperatura"))
        i-=1
        proxima = int(input(f"Digite a próxima {i} temperatura: "))
        if anterior < atual and atual > proxima:
            picos_encontrados +=1
            if atual > maior_valor or i  ==1:
                maior_valor = atual
    if picos_encontrados ==0:
        print("Nenhum pico foi encontrado!")
    else:
        print(f"Número de picos encontrados: {picos_encontrados}\nMaior valor no pico: {maior_valor}")
```

#### **29.** Uma sonda espacial utiliza um procedimento numérico para reduzir códigos inteiros antes de transmiti-los para a estação de controle. O processo começa com um número inteiro positivo N. Sempre que o valor atual for par, ele é substituído por sua metade. Quando for ímpar, ele é substituído por três vezes seu valor acrescido de uma unidade. O procedimento é repetido até que o valor 1 seja alcançado. Por exemplo, começando em 6, são produzidos os valores

- 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1.

#### Escreva um programa que leia o valor inicial e execute esse procedimento. Ao final, informe quantas transformações foram necessárias até chegar a 1, qual foi o maior valor alcançado durante todo o processo, quantas transformações foram realizadas quando o valor corrente era par e quantas foram realizadas quando ele era ímpar. O valor inicial também deve ser considerado na determinação do maior valor alcançado.

```python
def sonda(entrada):
    count_transformacoes = 0
    valor = entrada
    maior_valor = entrada
    count_par = 0
    count_impar = 0
    while entrada !=1:
        if entrada %2 ==0:
            count_transformacoes +=1
            count_par +=1
            entrada  = entrada //2
            if maior_valor < entrada:
                maior_valor = entrada
        elif entrada %2 !=0:
            count_impar +=1
            count_transformacoes +=1
            entrada = (entrada *3) +1
            if maior_valor < entrada:
                maior_valor = entrada
    return (f'Transformações totais: {count_transformacoes}\nMaior Valor: {maior_valor}\nQuantidade de Pares: {count_par}\nQuantidade de ímpar: {count_impar}')

        
```

```python
entrada = int(input("Digite o número: "))
sonda(entrada)
```
