#### <u>**18.** </u> Implemente uma classe chamada ContaBancaria. Cada objeto deve possuir os seguintes atributos privados:

- • número da conta;
- • nome do titular;
- • saldo;
- • limite de crédito;
- • histórico de operações.

#### O construtor deve receber o número da conta, o nome do titular e o limite de crédito.
#### Toda conta deve ser criada com saldo inicial igual a zero e histórico vazio.

#### Implemente os seguintes métodos:
- • depositar(valor): adiciona um valor positivo ao saldo;

- sacar(valor): realiza o saque apenas quando o valor for positivo e não ultrapassar a soma do saldo com o limite;

- • transferir(destino, valor): retira o valor da conta atual e o deposita em outro objeto da classe ContaBancaria;
- • consultar_saldo(): retorna o saldo atual;
- • consultar_saldo_disponivel(): retorna a soma do saldo com o limite;
- • exibir_extrato(): apresenta todas as operações realizadas na conta.

#### Cada operação válida deve ser registrada no histórico, indicando seu tipo e valor. Operações inválidas não devem alterar o saldo nem o histórico.
#### No programa principal:
- • crie pelo menos três contas;
- • realize depósitos, saques e transferências entre elas;
- • tente executar pelo menos uma operação inválida;
- • exiba o extrato e o saldo final de cada conta;
- • identifique a conta com maior saldo disponível.

#### Os atributos não devem ser acessados ou modificados diretamente fora da classe.

```python

class Conta_Bancaria:
    def __init__(self, num_conta, nome_titular, limite_credito):
        self.num_conta = num_conta
        self.nome_titular = nome_titular
        self.limite_credito = limite_credito
        self.saldo =  0
        self.hist_operacoes = []
    
    def depositar(self,valor):
        if valor >0:
            self.saldo += valor
            self.hist_operacoes.append(f"Depósito de R${valor} reais")
    
    def sacar(self, valor):
        soma_saldo = self.saldo + self.limite_credito
        if valor > 0 and valor <= soma_saldo:
            self.saldo -= valor
            self.hist_operacoes.append(f"Saque de R${valor} reais realizado na conta")

    def transferir(self, destino, valor):
        if valor >=0 and valor <= self.saldo:
            self.saldo -= valor
            destino.saldo += valor
            self.hist_operacoes.append(f"Valor de {valor} retirado da conta {self.num_conta} para a conta {destino.num_conta}")
            destino.hist_operacoes.append(f" Recebimento de RS{valor} da conta {self.num_conta}")
            
    def consultar_saldo(self):
        ####print(f"O saldo da conta {self.num_conta} é: {self.saldo:.2f}")
        return self.saldo
    
    def consultar_saldo_valor(self):
        return self.saldo
    
    def consultar_saldo_disponivel(self):
        soma_saldo = self.saldo + self.limite_credito
        print(f"O saldo disponível da conta {self.num_conta} é {soma_saldo:.2f}")
    
    def exibir_extrato(self):
        print("O histórico de movimentações é: ")
        for i in range(len(self.hist_operacoes)):
            print(f"{self.hist_operacoes[i]}")



       
```

```python
class Produto:
    def __init__(self, cod, descricao, preco_uni, qtd_estoque):
        self.cod = cod
        self.descricao = descricao
        self.preco_uni = preco_uni
        self.qtd_estoque = qtd_estoque
    
    def consultar_dados(self):
        dados_produto = dict()
        dados_produto["Codigo Produto"] = self.cod
        dados_produto["Descricao"] = self.descricao
        dados_produto["Preço_Unitario"] = self.preco_uni
        dados_produto["Quantidade_Estoque"] = self.qtd_estoque
        return dados_produto
    
    def acrescentar_unidades(self,valor):
        unidade_acrescentada = valor
        self.qtd_estoque += unidade_acrescentada
        return self.qtd_estoque
    
    def retirar_unidades_estoque(self, valor):
        if self.qtd_estoque > 0 and self.qtd_estoque > valor:
            self.qtd_estoque -= valor
            return True
        else:
            return False
```

```python
class Item_Pedido:
    def __init__(self, produto, quantidade):
        self.produto = produto
        self.quantidade = quantidade

    def  subtotal(self):
        subtotal_item = self.produto.preco_uni * self.quantidade
        return subtotal_item
```

```python

class pedido:
    def __init__(self, num_pedido, nome_cliente):
        self.num_pedido = num_pedido
        self.nome_cliente = nome_cliente
        self.lista_item_pedidos =  []
        self.situacao_pedido = "Aberto"
    
    def adicinar_item(self, produto, quantidade):
        novo_item = Item_Pedido(produto, quantidade)
        self.lista_item_pedidos.append(novo_item)
    
    def remover_item(self, cod_produto):
    
        ### duvida para remover o item pelo cod_produto:
        for itens in self.lista_item_pedidos:

            if itens.produto.cod == cod_produto:
                self.lista_item_pedidos.remove(itens)

    def calcular_total(self):
        soma = 0

        ### acessar dentro da lista a quantidade e multiplicar pelo preço unitário:
        for itens in self.lista_item_pedidos:
            soma += itens.quantidade * itens.produto.preco_uni
        return soma

    def finalizar(self):
        ### verificando se todos os pedidos tem estoque o suficiente:
        for itens in self.lista_item_pedidos:
            if itens.produto.qtd_estoque < itens.quantidade:
                return "Fora de Estoque! Não pode finalizar"
            
        for itens in self.lista_item_pedidos:
                itens.produto.qtd_estoque -= itens.quantidade
        if self.situacao_pedido == "Aberto":
            self.situacao_pedido = "Fechado"

    
    def exibir_resumo(self):
        total = self.calcular_total()
        estado = self.finalizar()
        print(f"O cliente : {self.nome_cliente} com número de pedido {self.num_pedido} pediu os itens\n, pagou um total de {total}\n, com estado do pedido {estado}")
        for itens in self.lista_item_pedidos:
            print(f"Itens pedidos: {itens.produto.descricao} - {itens.quantidade} unid")
    
```

#### Crie as seguintes classes derivadas:

- • Vendedor, que possui um valor total de vendas e recebe uma comissão percentual sobre esse valor;

- • Desenvolvedor, que possui uma quantidade de horas extras e recebe um valor adicional por hora.

### Cada classe derivada deve sobrescrever o método calcular_salario() de acordo com sua regra.

#### No programa principal:

- • crie uma única lista contendo objetos das diferentes classes;
- • percorra a lista e exiba os dados e o salário calculado de cada funcionário;
- • calcule o total da folha de pagamento;

#### O cálculo realizado no percurso da lista deve utilizar o mesmo método calcular_salario(),independentemente da classe concreta do objeto

```python
### começando por partes

class Funcionario:
    def __init__(self, matricula, nome, salario_base):
        self.matricula = matricula
        self.nome = nome
        self.salario_base = salario_base

    def calcular_salario(self):
        return self.salario_base

    def exibir_dados(self):
        print(f"Matrícula: {self.matricula}")
        print(f"Nome: {self.nome}")
        print(f"Salário base: {self.salario_base}")
```

```python
class Gerente(Funcionario):
    def __init__(self, matricula, nome, salario_base, valor_gratificacao):
        super().__init__(matricula, nome, salario_base)
        self.valor_gratificacao = valor_gratificacao

    def calcular_salario(self):
        salario_alterado = super().calcular_salario() + self.valor_gratificacao
        return salario_alterado
```

```python
class Vendedor(Funcionario):
    def __init__(self, matricula, nome, salario_base, valor_total_vendas, pct_comissao):
        super().__init__(matricula, nome, salario_base)
        self.valor_total_vendas = valor_total_vendas
        self.pct_comissao = pct_comissao

    def calcular_salario(self):
        salario_alterado = super().calcular_salario() + (self.valor_total_vendas * self.pct_comissao)
        return salario_alterado
```

```python
class Desenvolvedor(Funcionario):
    def __init__(self, matricula, nome, salario_base, qtd_horas_extras, valor_hora):
        super().__init__(matricula, nome, salario_base)
        self.qtd_horas_extras = qtd_horas_extras
        self.valor_hora = valor_hora

    def calcular_salario(self):
        salario_alterado = super().calcular_salario() + (self.qtd_horas_extras * self.valor_hora)
        return salario_alterado
```

```python
def adicionar_funcionario(lista):
    qtd = int(input("Quantos funcionários serão adicionados à lista? "))
    salario_base = 2500
    for i in range(qtd):
        matricula = int(input("Digite a matrícula do funcionário: "))
        nome = input("Digite o nome do funcionário: ")
        cargo = input("Qual o cargo do funcionário? ").lower() 

        if cargo == "desenvolvedor":
            qtd_horas_extras = float(input("Quantas horas extras ele fez? "))
            valor_hora = float(input("Qual o valor por hora extra? "))
            lista.append(Desenvolvedor(matricula, nome, salario_base, qtd_horas_extras, valor_hora))

        elif cargo == "vendedor":
            valor_total_vendido = float(input("Qual o valor total que o vendedor obteve? "))
            pct_comissao = float(input("Qual a porcentagem de comissão ele obteve (em %)? ")) / 100
            lista.append(Vendedor(matricula, nome, salario_base, valor_total_vendido, pct_comissao))

        elif cargo == "gerente":
            valor_gratificacao = float(input("Qual o valor da gratificação que o gerente obteve? "))
            lista.append(Gerente(matricula, nome, salario_base, valor_gratificacao))
        else:
            lista.append(Funcionario(matricula, nome,salario_base))

    return lista
```
