#### <u>Exercício 1 Classificação de números em intervalos</u>

#### Escreva um programa que leia dois números inteiros A e B.

#### Para cada número i no intervalo de A até B, o programa deve:

- mostrar o número i;
- mostrar todos os números de 1 até i;
- indicar, para cada valor, se ele é: 
    - múltiplo de 3;
    - múltiplo de 5;
    - múltiplo de ambos;
    - ou nenhum dos dois.

#### Exemplo para i = 6:

- Número 6:
- 1 - nenhum
- 2 - nenhum
- 3 - múltiplo de 3
- 4 - nenhum
- 5 - múltiplo de 5
- 6 - múltiplo de 3

#### Ao final, identifique quais padrões de lógica se repetem ao longo da solução.

```python
A= int(input("Digite o número A: "))
B= int(input("Digite o número B: "))

for i in range(A,B):
    print(i)
    for j in range(1,i+1):
        print(j)
        if j %3 ==0 and j %5 ==0:
            print(f"{j} é múltiplo de 3 e 5")
        elif j %3==0:
            print(f"{j} é múltiplo de 3")
        elif j%5 ==0:
            print(f" {j} É múltiplo de 5")
        else:
            print("Nenhum dos dois!")
```

#### Padrões de Lógica que se repetem:

- **Laços aninhados:** o for externo percorre o intervalo principal, e o interno percorre de 1 até o valor atual.

- **Validação com condições exclusivas:** Se usa if/elif/else para classificar cada número;

- **Dependência entre laços:** O segundo laço depende do valor do primeiro (i)

#### <u>**Exercício 2 Tabela de multiplicação com análise do resultado**</u>

#### Escreva um programa que leia dois números inteiros positivos: linhas e colunas.

#### O programa deve gerar uma tabela em que cada posição contenha o produto entre o número da linha e o número da coluna.

#### Além de mostrar o valor calculado, para cada resultado o programa deve informar:

- se ele é par ou ímpar;
- se ele é maior que 20;
- se ele pertence à faixa:

    - pequeno: até 10;
    - médio: de 11 a 20;
    - grande: acima de 20.

#### Exemplo:

- linha 3 e coluna 4 → produto = 12 → par, médio;
- linha 3 e coluna 5 → produto = 15 → ímpar, médio.

#### Ao final, identifique quais padrões de lógica se repetem ao longo da solução.

```python
linha = int(input("Digite o valor da linha: "))
while linha <0:
    print("Digite um valor maior que zero")
    linha = int(input("Digite o valor da linha: "))

coluna = int(input("Digite o valor da coluna: "))
while coluna <0:
    print("Digite um número maior que zero!")
    coluna = int(input("Digite o valor da coluna: "))


for i in range(1, linha+1):
    print(i)
    for j in range(1, coluna+1):
        produto = i * j
        print(f"{i} * {j} = {produto}")
        if produto %2==0:
            print(f"{produto} O número é par")
        elif produto %2 !=0:
            print(f"{produto} O número é ímpar!")
        if produto > 20:
            print(f"{produto}É grande!")
        elif produto >=11 and produto <=20:
            print(f"{produto} É médio!")
        else:
            print(f"{produto}É pequeno!")
```

#### Padrões de lógica que se repetem:

#### Laços aninhados: um for externo para linhas e um interno para colunas.

#### Cálculo dentro do laço: cada par (i, j) gera um produto.

#### Classificação com condições: cada produto é analisado em par/ímpar e pequeno/médio/grande.

#### **Dependência entre laços:** *o produto depende dos dois índices (i e j).*

#### <u>**Exercício 4 Boletim de turmas e alunos**</u>

#### Uma escola deseja registrar notas de várias turmas.

#### O programa deve ler:

- a quantidade de turmas;
- para cada turma, a quantidade de alunos;
- para cada aluno, 2 notas.

#### Para cada aluno, o programa deve calcular a média e informar:

- Aprovado, se a média for maior ou igual a 7;
- Recuperação, se a média for maior ou igual a 5 e menor que 7;
- Reprovado, caso contrário.

#### Ao final de cada turma, mostrar:

- média geral da turma;
- quantidade de aprovados;
- quantidade de alunos em recuperação;
- quantidade de reprovados.

#### Ao final, identifique quais padrões de lógica se repetem ao longo da solução.

```python
turmas = int(input("Digite a quantidade de turmas para se calcular as notas: "))

soma_turmas_notas =0
soma_alunos_notas = 0
media_turma = 0
qtd_aprovados = 0
qtd_reprovados = 0
qtd_recuperacao = 0
qtd_alunes = 0
soma_media_alunos = 0

for i in range(turmas):
    alunes = int(input(f"Digite a quantidade de alunos da turma {i}: "))
    soma_media_alunos = 0
    qtd_aprovados = 0
    qtd_reprovados = 0
    qtd_recuperacao = 0
    qtd_alunes += alunes
    
    for j in range(alunes):
        qtd_notas =2
        soma_alunos_notas = 0
        while qtd_notas > 0:
            notas = float(input("Digite a nota do aluno: "))
            soma_alunos_notas += notas
            qtd_notas -=1
        media_aluno = soma_alunos_notas / 2
        soma_media_alunos +=media_aluno
        if media_aluno >=7:
            print(f" {j} Aprovado!")
            qtd_aprovados +=1
        elif media_aluno >=5  and media_aluno < 7:
            print("Recuperação!")
            qtd_recuperacao +=1

        else:
            print("Reprodo!")
            qtd_reprovados +=1
    media_turma = soma_media_alunos/alunes

    print("A média da turma é:", media_turma)
    print("Possui", qtd_aprovados)
    print("Possui", qtd_recuperacao)
    print("Possui", qtd_reprovados)



```


#### <u> **Exercício 5** Sistema de pedidos por mesas em uma lanchonete </u>

#### Uma lanchonete deseja simular um sistema simples de pedidos.

#### O programa deve ler a quantidade de mesas atendidas.

#### Para cada mesa:

- ler a quantidade de pedidos feitos;
- para cada pedido, ler:
    - o código do produto;
    - a quantidade pedida.
    
- Considere os produtos:
   - 1 = Hambúrguer, R$ 18,00
   - 2 = Refrigerante, R$ 6,00
   - 3 = Sobremesa, R$ 9,00

#### Regras:

- se o código do produto for inválido, o pedido deve ser ignorado e uma mensagem deve ser exibida;
- se a quantidade for menor ou igual a zero, o pedido também deve ser ignorado;
#### ao final de cada mesa, mostrar:
- valor total da mesa;
- quantidade total de itens válidos.
- ao final do programa, mostrar:
  -  faturamento total;
   - quantidade vendida de cada tipo de produto.

#### Ao final, identifique quais padrões de lógica se repetem ao longo da solução.

```python
mesas = int(input("Informe a quantidade de mesas atendidas"))
pedidos_totais = 0
valor_hamburguer = 18
valor_refri = 6
valor_sobremesa = 9
faturamento_total= 0
qtd_hamburguer = 0
qtd_refri = 0
qtd_sobremesa= 0

for i in range(mesas):

    num_pedidos = int(input("Digite a quantidade de pedidos feitos: "))
    pedidos_totais += num_pedidos
    valor_total_mesa= 0
    valor_pedido = 0
    qtd_invalido = 0
    qtd_valido = 0

    for j in range(num_pedidos):

        cod_produto = int(input("Digite o código do produto: 1- Hambúrguer| 2- Refrigerante | 3- Sobremesa"))
        qtd_pedidos = int(input(f"Digite a quantidade de pedidos de {cod_produto} para {j} ")) 
        if qtd_pedidos <=0:
            print("Quantidade Inválida")
            qtd_invalido+=1
            continue
        if cod_produto not in(1,2,3):
            print("Código digitado inválido!")
            qtd_invalido+=1
        else: 
            qtd_valido+=1
            if cod_produto ==1 and qtd_pedidos >0 :
                print("Hambúrguer")
                qtd_hamburguer+= qtd_pedidos
                valor_pedido = valor_hamburguer* qtd_pedidos
            if cod_produto ==2 and qtd_pedidos > 0:
                print("Refrigerante!")
                qtd_refri+=qtd_pedidos
            if cod_produto ==3 and qtd_pedidos > 0:
                print("Sobremesa")
                qtd_sobremesa += qtd_pedidos
                valor_pedido =valor_sobremesa * qtd_pedidos
            ### Contabilizando o valor da mesa total para cada pedido e incrementando:
            valor_total_mesa += valor_pedido
    ### Após a cada loop interno é incrementado o valor total da mesa para o faturamento total:
    faturamento_total += valor_total_mesa
    print("O valor dessa mesa foi de: ", valor_total_mesa)
    print("A quantidade de pedidos válidos foi de:", qtd_valido)
print("O faturamento total após processamento foi de: ", faturamento_total)
print(f" A quantidade  total de Hamburguer vendida foi de  {qtd_hamburguer}")
print(f" A quantidade  total de Refri vendida foi de  {qtd_refri}")
print(f" A quantidade  total de Spbremesa vendida foi de  {qtd_sobremesa}")
```

#### Padrões de lógica que se repetem

#### **Laços aninhados:** um for externo para mesas, outro para pedidos.

#### **Validação de entrada:** checar se o código e a quantidade são válidos antes de processar.

#### **Acúmulo de valores:** somar quantidades e valores para calcular totais por mesa e totais gerais.

#### **Classificação com condições:** decidir se o pedido é válido ou inválido.

#### **Separação de escopo:** variáveis da mesa zeradas dentro do loop da mesa, variáveis globais fora

```python
entrada = int(input("Digite um número para se calcular a sequência: "))
maior_sequencia = 0
anterior = entrada
while entrada !=0:
    entrada = int(input("Digite um número para se calcular a sequência: "))
    if anterior < entrada:
        maior_sequencia +=1
    anterior = entrada
print(maior_sequencia)
```
