#### **Exercício 1** - *Melhorando a legibilidade de uma função*

#### Um desenvolvedor escreveu uma função para realizar uma operação matemática utilizando repetições, mas escolheu nomes pouco representativos.

- def f(x, y):
    - r = 0

    - for i in range(x):
        - r += y

    - return r

#### Considere que x e y são números inteiros não negativos.

#### Reescreva essa função para melhorar sua legibilidade, sem modificar o algoritmo utilizado para calcular o resultado.

#### Na nova versão:

- utilize um nome de função que descreva sua finalidade;
- substitua os nomes genéricos dos parâmetros;
- utilize nomes de variáveis mais significativos;
- mantenha o processamento por somas sucessivas.

#### Acrescente chamadas à função para verificar os seguintes resultados:

- 3 e 4  - resultado esperado: 12
- 0 e 5  - resultado esperado: 0
- 5 e 2  - resultado esperado: 10

#### Entrega: arquivo exercicio_01.py contendo a função reescrita e as três chamadas para verificação.

```python
def multiplifcacao_por_soma(num_repeticoes, num_escolhido):
    valor_multiplicacao_soma = 0
    for i in range(num_repeticoes):
        valor_multiplicacao_soma += num_escolhido
    return valor_multiplicacao_soma
```

```python
numero_repeticoes = int(input(" Multiplicar o número por quantas vezes (x)? "))
while numero_repeticoes <0:
    print(f"O número digitado:{numero_repeticoes} deverá ser maior que zero")
    numero_repeticoes = int(input(" Multiplicar o número por quantas vezes (x)? "))
numero  = int(input("Digite o número ser multiplicado: "))
while numero <0:
    print("Não pode ser menor que zero!")
    numero  = int(input("Digite o número ser multiplicado: "))
```

```python
resultado_multiplicacao = multiplifcacao_por_soma(numero_repeticoes,numero )
print(f"{numero_repeticoes} e {numero} - resultado esperado: {resultado_multiplicacao}")
```

```python
resultado_multiplicacao = multiplifcacao_por_soma(numero_repeticoes,numero )
print(f"{numero_repeticoes} e {numero} - resultado esperado: {resultado_multiplicacao}")
```

#### **Exercício 3** - *Criando e consultando uma docstring*

#### Retome a função contar_pares() desenvolvida na questão anterior.

#### Acrescente uma docstring que informe:

- o objetivo da função;
- o nome, o tipo e a finalidade do parâmetro;
- o tipo e o significado do valor retornado;
- o comportamento para listas vazias, zero e números negativos.

#### Depois de documentar a função, acrescente ao programa os comandos:

- *help(contar_pares)*

#### *print(contar_pares.__doc__)*

#### Execute o programa e observe como o Python apresenta as informações documentadas.

#### Entrega: arquivo exercicio_03.py contendo a função documentada e os dois comandos de consulta.



```python
def contar_pares(lista_numeros):
    """Função que conta a quantidade de números pares
    
    Parâmetros: 
    Uma lista (list ) com x números do tipo INT, negativos ou positivos, incluindo o zero

    Retorno:
    Número inteiro que representa a quantidade de números pares dentro da lista, 

    Para lista (*lista*) vazias, será retornado 0
    Para apenas um número na lista e este for o zero será retornado 1
    Para números negativos dentro da lista, será retornando a quantidade de números negativos que forem múltiplos de 2
    
    """
    quantidade = 0
    for numero in lista_numeros:
        if numero %2 ==0:
            quantidade +=1
      
    return quantidade
```

```python
help(contar_pares)
```

```python
print(contar_pares.__doc__)
```
