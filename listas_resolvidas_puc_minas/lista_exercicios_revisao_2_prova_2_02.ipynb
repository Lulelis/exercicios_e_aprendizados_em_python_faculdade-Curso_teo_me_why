#### **Letra-b-)**

#### 1. Estrutura inicial

- A matriz começa com valores inteiros e é processada pela função ajustar. Na letra B, o fator passado é 3, então cada elemento da diagonal principal será somado com 3.

#### 2. Função ajustar

- Dentro da função:

- Diagonal principal (i == j): cada elemento recebe +3.

- Valores pares fora da diagonal: são divididos por 2.

- A variável alterados conta quantos elementos da diagonal foram modificados.


- Resultado após ajustar:[[4, 1, 6], [2, 8, 3], [7, 4, 12]]

    - E alterados = 3, porque três elementos da diagonal foram ajustados.

#### 3. Referências

#### Como referencia = matriz e primeira_linha = matriz[0], qualquer modificação feita dentro da função afeta diretamente a matriz original e também a variável primeira_linha.
#### Isso garante que, ao alterar a matriz, a primeira linha também seja atualizada.

#### 4. Atualização da primeira linha:

- Depois da função, o código faz:

- primeira_linha[-1] = primeira_linha[-1] + qtd
    - O último elemento da primeira linha (que era 3) recebe +3, tornando-se 6.
Assim, a primeira linha final é [4, 1, 6].

#### 5. Impressões finais

- O código imprime Cada linha da matriz:

    - Linha [4, 1, 6]
    - Linha [2, 8, 3]
    - Linha [7, 4, 12]

- A soma da primeira coluna: 4 + 2 + 7 = 13.

- A quantidade de elementos alterados: 3.

#### Saída Final:

- Linha [4, 1, 6]
- Linha [2, 8, 3]
- Linha [7, 4, 12]
- Coluna 13
- Alterados 3

#### Resumo conceitual

- O for duplo percorre toda a matriz e aplica as regras.

- As referências garantem que primeira_linha seja afetada junto com a matriz.

- O incremento final ajusta o último elemento da primeira linha.

- A soma da coluna e o contador de alterações confirmam o resultado


```python
def ajustar(matriz, fator):
    alterados = 0

    for i in range(len(matriz)):
        for j in range(len(matriz[i])):
            if i == j:
                matriz[i][j] = matriz[i][j] + fator
                alterados = alterados + 1
            elif matriz[i][j] % 2 == 0:
                matriz[i][j] = matriz[i][j] // 2

    return alterados


def soma_coluna(matriz, coluna):
    total = 0

    for linha in matriz:
        total = total + linha[coluna]

    return total


n = int(input())
matriz = []

for i in range(n):
    linha = []
    for j in range(n):
        linha.append(int(input()))

    matriz.append(linha)

referencia = matriz
primeira_linha = matriz[0]

qtd = ajustar(referencia, n)
primeira_linha[-1] = primeira_linha[-1] + qtd

for linha in matriz:
    print("Linha", linha)

print("Coluna", soma_coluna(matriz, 0))
print("Alterados", qtd)
```

#### Por exemplo:
#### - [4, 4, 4, 2, 2, 7, 4, 4, 9]

#### deve produzir:
- [4, 2, 7, 4, 9]

#### Observe que o valor 4 aparece novamente na lista final porque sua última ocorrêncianão é consecutiva às ocorrências iniciais. Não utilize conjuntos para resolver a questão.

```python
def remover_consecutivos(lista_num):
    nova_lista = []
    for i in range(len(lista_num)-1):
        if lista_num[i] != lista_num[i+1]:
            nova_lista.append(lista_num[i])
    nova_lista.append(lista_num[-1])
    
    return nova_lista
```

#### Por exemplo, considere:

#### Nesse caso, os valores acima da diagonal são 2, 3 e 6, enquanto os valores abaixo são 4, 7 e 8.

#### Ao final, informe as duas somas e indique qual região possui o maior valor acumulado ou se existe empate.

```python
def soma_elementos_acima_baixo_diagonal(matriz):
    soma_acima = 0
    soma_abaixo = 0
    for i in range(len(matriz)):
        for j in range(len(matriz[i])):
            if i!=j:
                if  i< j:
                    soma_acima+= matriz[i][j]
                elif i>j:
                    soma_abaixo+= matriz[i][j]
    if soma_abaixo > soma_acima:
        print(f"Soma dos elementos Abaixo da Diagonal Principal é maior!\nSoma abaixo:{soma_abaixo}/Soma Acima:{soma_acima}")
    elif soma_acima > soma_abaixo:
        print(f"Soma dos elementos Acima da Diagonal Principal é maior!\nSoma acima:{soma_acima}/Soma Abaixo:{soma_abaixo}")
    else:
        print(f"Soma Acima e Abaixo possuem o mesmo valor!\nSoma Abaixo:{soma_abaixo}/Soma Acima:{soma_acima}")
    return soma_acima, soma_abaixo
```

#### O estacionamento está dividido em fileiras, e algumas regiões são mais utilizadas que outras. Faça um programa que receba uma matriz M × N representando a ocupação atual.
#### Crie uma função que calcule a taxa de ocupação de cada linha e outra função que determine a quantidade de vagas livres em cada coluna.
#### O programa deve identificar a fileira com maior taxa de ocupação e a coluna com maior quantidade de vagas livres. Em caso de empate, considere o menor índice.
#### Além disso, construa um dicionário no qual cada veículo presente na matriz esteja associado a uma tupla contendo sua linha e sua coluna.
#### Em seguida, leia o identificador de um veículo e informe sua posição. Caso o veículo não esteja estacionado, exiba uma mensagem adequada.
#### Considere que um mesmo identificador não aparece em mais de uma posição da matriz.

```python
def verifica_posicao_matriz(id, matriz_estacionamento):
    tem_na_matriz = False
    for i in range(len(matriz_estacionamento[0])):
        for j in range(len(matriz_estacionamento)):
            if matriz_estacionamento[j][i] == id:
                print(f"A posição informada:{id} já existe na matriz de estacionamento!")
                tem_na_matriz= True
    return tem_na_matriz
```

```python
def taxa_ocupacao_matriz(matriz_estacionamento):
    taxa_ocupacao_lista = []
    for i in range(len(matriz_estacionamento)):
        soma_ocupacao = 0
        qtd_termos_linhas = 0
        for j in range(len(matriz_estacionamento[i])):
            qtd_termos_linhas +=1
            if matriz_estacionamento[i][j] != 0:
                soma_ocupacao +=1
        taxa_ocupacao = (soma_ocupacao/qtd_termos_linhas) * 100
        print(f"Taxa ocupação linha:{i}ª:{taxa_ocupacao:.2f}")
        taxa_ocupacao_lista.append(taxa_ocupacao)
    return taxa_ocupacao_lista
```
