#### 11. Escreva uma função chamada remover_consecutivos(valores) que receba uma lista de números inteiros e retorne uma nova lista eliminando apenas repetições consecutivas.

#### Por exemplo:
#### - [4, 4, 4, 2, 2, 7, 4, 4, 9]

#### deve produzir:
- [4, 2, 7, 4, 9]

#### Observe que o valor 4 aparece novamente na lista final porque sua última ocorrêncianão é consecutiva às ocorrências iniciais. Não utilize conjuntos para resolver a questão.

```python
def remover_consecutivos(lista_num):
    nova_lista = []
    for i in range(len(lista_num)-1):
        if lista_num[i] != lista_num[i+1]:
            nova_lista.append(lista_num[i])
    nova_lista.append(lista_num[-1])
    
    return nova_lista
```

```python
quantidade_numeros  = int(input("Quantos números a lista irá receber?"))
while quantidade_numeros < 0:
    print("Digite uma quantidade maior que 0 de números para compor a lista!")
    quantidade_numeros  = int(input("Quantos números a lista irá receber?"))

lista_num = []
for i in range(1, quantidade_numeros+1):
    numeros = int(input(f"Digite o :{i}º número da lista: "))
    lista_num.append(numeros)
print(lista_num)
```

```python
nova_lista = remover_consecutivos(lista_num)
print("Nova lista:",nova_lista)
```

#### **14.** Uma lista de números inteiros deve ser reorganizada de forma que todos os valores pares apareçam antes dos valores ímpares.

#### Crie uma função chamada separar_pares_impares(valores) que retorne uma nova lista com essa organização, preservando a ordem relativa em que os números pares aparecem entre si e a ordem relativa dos números ímpares entre si.

#### Por exemplo:
#### [7, 4, 2, 9, 6, 3, 8]
#### deve produzir:
#### [4, 2, 6, 8, 7, 9, 3]
#### *Não utilize sorted() para resolver a questão.*

```python
def separar_pares_impares(lista_num):
    nova_lista =[]
    lista_pares = []
    lista_impar =[]
    for i in range(len(lista_num)):
        if lista_num[i]%2 ==0:
            lista_pares.append(lista_num[i])
        elif lista_num[i] %2 !=0:
            lista_impar.append(lista_num[i])
    nova_lista = lista_pares + lista_impar
    return nova_lista
```

```python
quantidade_numeros  = int(input("Quantos números a lista irá receber?"))
while quantidade_numeros < 0:
    print("Digite uma quantidade maior que 0 de números para compor a lista!")
    quantidade_numeros  = int(input("Quantos números a lista irá receber?"))

lista_num = []
for i in range(1, quantidade_numeros+1):
    numeros = int(input(f"Digite o :{i}º número da lista: "))
    lista_num.append(numeros)
print(lista_num)
lista_organizada = separar_pares_impares(lista_num)
print(f"Lista Pares e Ímpar Organizada:{lista_organizada}")
```

#### as somas das linhas são 7, 14 e 10. Portanto, a linha de índice 1 deve ser identificada como a de maior soma.

```python
def montar_matriz():
    matriz_retangular = []
    num_linhas = int(input("Digite o número de Linhas que a matriz terá!"))
    while num_linhas < 0:
        print("A matriz deverá ter um número de Linhas maior que zero!")
        num_linhas = int(input("Digite o número de Linhas que a matriz terá!"))
    num_colunas = int(input("Digite o número de colunas que terá na matriz!"))
    while num_colunas < 0:
        print("Deverá ser maior que zero!")
        num_colunas = int(input("Digite o número de colunas que terá na matriz!"))
    for i in range(num_linhas):
        colunas = []
        for j in range(num_colunas):
            numero = int(input(f"Digite o número: {j+1} da linha:{i+1}"))
            colunas.append(numero)
        matriz_retangular.append(colunas)
    return matriz_retangular

```

```python
def linha_matriz_maior_valor(matriz):
    linha_maior = ""
    maior_valor = float("-inf")
    for i in range(len(matriz)):
        soma = 0
        for j in range(len(matriz[0])):
            soma += matriz[i][j]
        if soma > maior_valor:
            maior_valor = soma
            linha_maior = i
    return linha_maior, maior_valor
```
