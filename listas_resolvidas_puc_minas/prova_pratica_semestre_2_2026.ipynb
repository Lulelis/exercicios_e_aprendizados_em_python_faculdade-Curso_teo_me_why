#### **Questão_1-** *Leitura de Arquivos-Leitura em dicionários+Escrita em outro arquivo para salvar os resultados*

```python


def estoque_quantidade(qtd_linhas_arquivo):
    with open ("dados1.txt", "w", encoding="utf-8") as arquivo_estoque:
        for i in range(1,qtd_linhas_arquivo+1):
            cod = input("Digite o cÃ³digo do arquivo: ").lower()
            tipo = input("Digite o tipo no arquivo: ").upper()
            while tipo not in ("E", "S"):
                print("Apenas o tipo E ou S")
                tipo = input("Digite o tipo no arquivo: ").upper()
            qtd = int(input(f"Digite a quantidade de: {cod}"))
            while qtd <0:
                print("Deve ser um inteiro positivo!")
                qtd = int(input(f"Digite a quantidade de: {cod}"))
            arquivo_estoque.write(f"{cod};{tipo};{qtd}\n")
    print("Arquivo salvo com sucesso")


qtd_escrita_arquivo = int(input("Qual a quantidade mÃ¡xima de escrita no arquivo? "))
while qtd_escrita_arquivo <0:
    print("Deve ser maior que zero!")
    qtd_escrita_arquivo = int(input("Qual a quantidade mÃ¡xima de escrita no arquivo? "))
estoque_quantidade(qtd_escrita_arquivo)

nome_arquivo = "dados1.txt"

def dicio_arquivos(nome_arquivo):
    dicio_estoque = {}
    maior_pico = float("-inf")
    with open(nome_arquivo, "r", encoding="utf-8") as arquivo_estoque:
        for dados in arquivo_estoque:
            parte_dados = dados.strip().split(";")
            codigo = parte_dados[0]
            tipo = parte_dados[1]
            qtd_estoque = int(parte_dados[2])
            if codigo not in dicio_estoque:
                dicio_estoque[codigo] = {}
                if tipo =="E":
                    dicio_estoque[codigo]["maior_pico"] = qtd_estoque
                    dicio_estoque[codigo]["estoque_atual"] = qtd_estoque
            
            elif codigo in dicio_estoque:
                if tipo == "E":
                    dicio_estoque[codigo]["estoque_atual"] += qtd_estoque
                    if dicio_estoque[codigo]["maior_pico"] < qtd_estoque:
                        dicio_estoque[codigo]["maior_pico"] = qtd_estoque
                if tipo == "S":
                    dicio_estoque[codigo]["estoque_atual"] = dicio_estoque[codigo]["estoque_atual"] - qtd_estoque

                    
        return dicio_estoque
dicionario_estoque = dicio_arquivos(nome_arquivo)

def criar_arquivos_alertas(dicio_estoque):
    with open("alertas.txt", "w", encoding="utf-8") as arquivo_alertas:
        for chaves in dicio_estoque:
            nome_codigo = chaves
            pico = 0
            estoque_final = 0
            for chave, valor in dicio_estoque[nome_codigo].items():
                if chave == "maior_pico" and valor >=10:
                    pico = valor
                if chave ==  "estoque_atual" and  valor <=3:
                    estoque_final = valor
                
            if pico != 0 and estoque_final !=0:
                arquivo_alertas.write(f"{nome_codigo};{pico};{estoque_final}\n")
                    
    print("Arquivo salvo com sucesso!")

criar_arquivos_alertas(dicionario_estoque)
```
