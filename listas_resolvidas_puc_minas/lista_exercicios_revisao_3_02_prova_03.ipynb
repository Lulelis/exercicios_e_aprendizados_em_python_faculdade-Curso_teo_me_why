#### **26.**- *Simulação Bancária com Transferência entre Instâncias*

#### •	Conceitos Abordados: Encapsulamento de regras de negócios (saque limitado ao saldo, depósito positivo), validação de operações e método transferir com recebimento de objeto destino.


#### Crie uma classe chamada Conta para representar uma conta bancária simples.

#### A classe deve possuir os atributos:
  - • numero;
  - • titular;
  - • saldo.

#### Implemente os seguintes métodos:
  - • depositar(valor), que acrescenta ao saldo apenas valores positivos;
  - • sacar(valor), que realiza o saque somente quando o valor for positivo e houver saldo suficiente;
  - • consultar_saldo(), que retorna o saldo atual;
  - • exibir(), que apresenta os dados da conta.

#### Depois, acrescente o método:
#### > **transferir(destino, valor)**

#### O parâmetro destino deve receber outro objeto da classe Conta. A transferência somente deve ser realizada se o saque puder ser efetuado na conta atual. Nesse caso, o valor deverá ser depositado na conta de destino utilizando os métodos já definidos nas classes.
#### No programa principal, crie duas contas e realize uma transferência entre elas.

```python
class Conta_Bancaria:
    def __init__(self, numero, titular, saldo):
        self.numero = numero
        self.titular = titular
        self.saldo = saldo

    ### Métodos:
    def consultar_saldo(self):
        return self.saldo
    
    def depositar(self, valor):
        while valor <0:
            print("Não existe valor negativo!")
            valor = float(input("Digite o valor novamente: "))
        self.saldo += valor
        print(f"Valor de R${valor} depositado na conta: {self.numero} do titular: {self.titular}")
        return self.saldo

    def sacar(self, valor):
        while valor <0:
            print("Não existe valor negativo")
            valor = float(input("Digite o valor novamente: "))
        saque_realizado_com_sucesso = True
        if valor > self.consultar_saldo():
            print(f"Valor de saldo insuficiente para prosseguir com operação!\nSaldo altuar:{self.consultar_saldo()}")
            saque_realizado_com_sucesso = False
            return saque_realizado_com_sucesso
        elif valor <= self.consultar_saldo():
            self.saldo -= valor
            print(f"Saque de: {valor} realizado com sucesso!\nSaldo atual:{self.consultar_saldo()}")
            return saque_realizado_com_sucesso

    def exibir(self):
        print(f"Dados da conta Bancária!\nNúmero:{self.numero}\nTitular:{self.titular}\nSaldo Atual R$:{self.saldo}")

    def transferir(self, destino, valor):
        dinheiro_saque = self.sacar(valor)
        if dinheiro_saque:
            depositar = destino.depositar(valor)
            print(f"Saldo da conta que recebeu o saque: {depositar}\nSaldo Atual: {destino.consultar_saldo()}")
        elif not dinheiro_saque:
            print(f"Imposível prosseguir com operação de transferência bancária uma vez que:{self.consultar_saldo()} é insuficiente para o valor: {valor}")

        
```

```python
def criar_conta_bancarias():
    #nome_conta = input("Qual será o nome da conta Bancária? ").lower
    for i in range(1):
        numero_conta = input(f"Qual o {i+1}º número da conta do titular: ").lower()
        titular = input(f"Qual o nome do titular da {i+1}º conta pertencente ao {numero_conta}").lower()
        saldo = float(input(f"Qual o {i+1}º saldo da conta do titular: {titular} de número: {numero_conta}"))
        while saldo <0:
            print("Não existe saldo negativo neste exemplo")
            saldo = float(input(f"Qual o {i+1}º saldo da conta do titular: {titular} de número: {numero_conta}"))
        nome_conta = Conta_Bancaria(numero_conta, titular, saldo)
    return  nome_conta
```

```python
class Cliente:
    def __init__(self, nome, telefone):
        self.nome = nome
        self.telefone = telefone

    def exibir(self):
        print(f"Nome Cliente:{self.nome}\nTelefone:{self.telefone}")
```

```python
class Veiculo:
    def __init__(self, placa, modelo, proprietario):
        self.placa = placa
        self.modelo = modelo
        self.proprietario = proprietario

    def exibir(self):
        print(f"Veículo Proprietário:{self.proprietario.nome}-Telefone dele: {self.proprietario.telefone}\nPlaca:{self.placa}-Modelo:{self.modelo}")
```
