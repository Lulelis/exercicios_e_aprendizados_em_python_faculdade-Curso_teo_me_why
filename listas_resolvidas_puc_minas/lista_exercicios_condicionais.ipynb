Uma residência deseja estimar o consumo mensal de energia de um equipamento.

Leia:

- potência do aparelho (Watts)

Após ler os dados, calcule o consumo diário, consumo mensal e custo mensal (em reais), e os imprima na tela.

 > Para calcular o consumo diário, devemos calcular: (potencia * horas) / 1000

```python

potencia = float(input("A potência do equipamento informada é: "))
print(potencia)
horas = int(input("Informe o total de horas: "))
print(horas)
dias_uso_por_mes = int(input("Dias de uso do equipamento por mês: "))

preco_energia = potencia * horas
consumo_diario = preco_energia/ 1000
print(f"Consumo diário, é:{consumo_diario} ")
consumo_mensal = consumo_diario * 30
print(f"O consumo mensal é: "{consumo_mensal})


```

#### Exercício_2

Planejamento de Viagem
Um motorista deseja estimar o custo de combustível de uma viagem.

Leia:

- distância da viagem (km)
- consumo do carro (km/l)
- preço do combustível (R$/l)

> Calcule o quantidade de litros necessários para a viagem, e o custo total da viagem. Imprima os resultados na tela.

```python
distancia = float(input("A distância da viagem em km:"))
print("A distância é: ", distancia)
consumo_carro = float(input("O consumo do carro é: "))
print(consumo_carro)
preco_combustivel = float(input("Informe o preço do combustível em R$: "))
print(preco_combustivel)

#Consumo está em Km por litro
# Distancia  está em Km
# Quantos litros, é só dividir a distância pelo consumo : distância / consumo
quantidade_litros = distancia / consumo_carro
print(f"A quantidade de listro necessárias para a viagem, é: {quantidade_litros:.2f}L")
custo_total_viagem = preco_combustivel * quantidade_litros
print(f"O custo total da viagem, é: {custo_total_viagem}")

```

#### Exercício_3

Boletim Simplificado
Leia três notas e calcule as seguintes informações:

- Soma das notas
- Média das notas
- Pontos que faltam para o aluno ficar com total (máximo de pontos possíveis)
Observação: Cada prova vale 10 pontos.

```python
nota_maxima = 30
nota_1 = float(input("A primeira nota do aluno é: "))
nota_2 = float(input("A segunda nota do aluno é: "))
nota_3 = float(input("A terceira nota do aluno é: "))

soma_notas_aluno = nota_1 + nota_2 + nota_3
perda_pts = nota_maxima - soma_notas_aluno
media_notas = soma_notas_aluno / 3

print("Somatório nota do aluno nas três provas: ", soma_notas_aluno)
print(f"O Aluno perdeu um total de : {perda_pts}")
print(f"A média das notas do aluno é: {media_notas:.2f}")
```

Classificação de temperatura
Escreva um programa que leia a temperatura atual em graus Celsius.

O programa deve verificar as seguintes situações:

se a temperatura é menor que 10
se a temperatura está entre 10 e 25
se a temperatura é maior que 25
Para cada situação, o programa deve mostrar:

Temperatura baixa
Temperatura agradável
Temperatura alta
Desafio adicional

Versão A: utilizando vários if independentes.
Versão B: utilizando operadores lógicos (and) para verificar intervalos, e exclusão mútua

```python
temperatura = float(input("Digite a temperatura em graus celcius:"))
if temperatura < 10:
    print("Temperatura baixa")
elif temperatura >= 10 and temperatura <=25:
    print("Temperatura Agradável!")
else:
    print("Temperatura Alta")
```

#### Exercício 4:

#### Exerício 5:

Sistema de aprovação em disciplina
Uma disciplina utiliza dois critérios para aprovação:

média final
frequência
O programa deve ler:

média do aluno
percentual de frequência
As regras são:

Se a frequência for menor que 75%, o aluno está reprovado por falta.
Caso contrário, analisar a média:
média ≥ 60 → aprovado
média entre 40 e 59 → recuperação
média abaixo de 40 → reprovado por nota
O programa deve mostrar o resultado final do aluno.

Desafio adicional

Versão A: utilizando condicionais aninhadas.
Versão B: tentando reduzir o número de níveis de aninhamento usando operadores lógicos (and).

```python
media_final = float(input("Ensira a média final do Aluno:"))
percentual_frequencia =  int(input("Ensina o percentual de frequência do aluno (%):"))
if percentual_frequencia <75:
    print("Reprovado por falta")
elif media_final >=60:
    print("Aprovado")
elif media_final >=40 and media_final <=59:
    print("Recuperação!")
elif media_final < 40:
    print("Reprovado por nota")
```

Um laboratório de informática possui regras específicas para acesso durante a noite.

O programa deve ler:

- idade do usuário
- se possui matrícula ativa (1 = sim, 0 = não)
- se possui autorização especial (1 = sim, 0 = não)

Regras de acesso:

> Estudantes com matrícula ativa e idade ≥ 18 podem entrar.

> Estudantes com matrícula ativa menores de 18 só entram com autorização.

> Pessoas sem matrícula ativa só entram com autorização

```python
idade_usuario = int(input("Informe a idade do usuario"))
matricula = int(input("Informe a matricula para as seguintes condições: especificamemente matricula_ativa: 1 e matricula inativa: 0"))

matricula_ativa = 1
matricula_nao_ativa = 0

if matricula == matricula_ativa and idade_usuario >=18:
    print("Pode Entrar")
elif matricula  == matricula_ativa and idade_usuario < 18:
    print("Só entram com autorização")
elif matricula == matricula_nao_ativa:
    print("Só entram com autorização")

```

Sistema de classificação em uma competição
Uma competição escolar de programação classifica participantes com base em:

pontuação obtida
tempo gasto na prova
O programa deve ler:

pontuação total do participante
tempo total gasto (em minutos)
Classificação:

pontuação ≥ 90 → Ouro
pontuação ≥ 70 → Prata
pontuação ≥ 50 → Bronze
caso contrário → sem medalha
Regra adicional:

Se o participante obtiver Ouro e terminar a prova em menos de 120 minutos, ele recebe:

Participante destaque da competição
O programa deve mostrar:

a classificação obtida
e, se aplicável, o título de destaque







```python
pontuacao_total = int(input("Informe a pontuação total dos participantes"))
tempo_total_gasto = int(input("Informe o tempo total gasto"))

if pontuacao_total >=90 and tempo_total_gasto < 120:
    print("Participante Destaque da competição!")
elif pontuacao_total >=90:
    print("Ouro!")
elif pontuacao_total >=70 and pontuacao_total <90:
    print("Prata!")
elif pontuacao_total >=50 and pontuacao_total <=70:
    print("Bronze")
else:
    print("Sem medalha")
```
