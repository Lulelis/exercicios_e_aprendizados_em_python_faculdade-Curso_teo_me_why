```python
### Entrada:

#- Pedir ao usuário uma string com o número binário

entrada = input("Digite um número binário: ") ### guardar como string pois é necessário acessar cada caracterer individualmente
posicao = 0 ### Contando a posição do dígito (da direita para a esquerda)

decimal = 0 ### variável para acumular o valor convertido

while len(entrada) > 0:
    ultimo_digito = int(entrada[-1]) ### pegando o último dígito e convertendo para número
    decimal = decimal + (ultimo_digito * (2 ** posicao)) ### soma o valor convertido
    posicao +=1
    entrada = entrada[:-1]  ### Removendo o último caractere da string
print("O valor decimal: ", decimal)

```

```python

#### Entrada

# - 5 Etapas de chacagem: em cada etapa (consumo fixo de 1.7% da bateria total inicial) 
## -  se velocidade de vento > 30 km/h o drone deve cancelar a decolagem imediantamente 
### Baterua abaixo de 10%: processo interrompido

### Saída:

#-  Informar se a decolagem foi "Autorizada", "Cancelada por Vento Forte" ou "Abortada por Bateria"

##- Indicar em qual etapa o problema ocorreu e o nível final de carga



count =0 
bateria_drone = float(input("Informe a carga de bateria do drone: "))
while bateria_drone <0 or bateria_drone > 100.00:
    print("Bateria informada incorreta. Informe de 0 à 100")
    bateria_drone = float(input("Informe a carga de bateria do drone: "))
consumo_fixo = (1.7/100 * bateria_drone)

for bateria in range(1,5+1):
    bateria_drone -= consumo_fixo
    velocidade_vento= float(input("Informe a condição do vento: "))
    if velocidade_vento > 30.00:
        print(f"Condição climática desfavorável, ventos {velocidade_vento} acimas de 30km/h. Interrompendo!")
        break
    if bateria_drone < 10.00:
        print(f"Operação interrompida. Bateria abaixo de 10%. ")
        break
    count +=1

if count ==0:
    print("Não ocorreu nenhuma etapa!")
else:
    if bateria_drone >= 10.00 and velocidade_vento <30:
        print(f"Decolagem com sucesso! Nível de bateria em {bateria_drone}")
    elif velocidade_vento > 30.00:
        print("Operação cancelada por ventos fortes!")

    elif bateria_drone < 10.00:
        print(f"Abortada por Bateria! {bateria_drone}, na etapa {count}")

```
