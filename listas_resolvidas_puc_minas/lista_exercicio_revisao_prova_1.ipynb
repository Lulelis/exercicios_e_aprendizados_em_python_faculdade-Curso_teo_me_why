```python
### Monitorar o estoque ao final do inventário
#### Entradas
# -nome do item
# - quantidade disponível em estoque
# - quantidade minima ideal para funcionamento

#### - processo continua até que o usuário informe (fim) como nome do produto

### Requisitos:

#- Quantidade negativa será considerado inválida e deverá ser ignorado

### Produtos válidos:
#  - classificar em: 
#       - estoque crítico: “estoque crítico” quando a quantidade disponível for menor que 50% da quantidade mínima ideal;
#       - estoque baixo: quando a quantidade disponível for menor que a quantidade mínima ideal, mas bçai estiver em situação crítica
#       - estoque adequado: quando a quantidade disponível for maior ou igual à quantidade mínima

### Pós processamento:
# informar:
#     - quantos produtos ficarma em situação crítica
#     - quantos produtos ficaram com estoque baixo e 
#     - quantos ficaram  com estoque adequado
#  informar o nome do produto com maior diferença positiva entre estoque disponível e estoque mínimo, considerando apenas produtos válidos. 

count_critico = 0
count_estoque_baixo = 0
count_estoque_adequado = 0
maior_diferenca = 0
maior_produto = None
count_geral = 0


item = input("Qual o nome do produto?")
while item !="fim":
    estoque = int(input("Digite a quantidade em estoque do produto:"))
    if estoque <0:
        print("Quantidade negativa,será ignorado")
        estoque = int(input("Digite a quantidade em estoque do produto:"))
        continue
    quantidade_minima = int(input("Digite a quantidade mínima ideal para funcionamento deste produto digitado"))
    if quantidade_minima <0:
        print("Quantidade negativa informada, será ignorado!")
        quantidade_minima = int(input("Digite a quantidade mínima ideal para funcionamento deste produto digitado"))
        continue
    diferenca = estoque - quantidade_minima
    if diferenca > 0 and diferenca > maior_diferenca:
        maior_diferenca = diferenca
        maior_produto = item
    percentual_produto_disponivel = 0.50 * quantidade_minima
    if estoque < percentual_produto_disponivel:
        print("Estoque crítico!")
        count_critico +=1
    elif estoque < quantidade_minima:
        print("Estoque baixo!")
        count_estoque_baixo+=1
    else:
        print("Estoque adequado!")
        count_estoque_adequado +=1
    count_geral +=1
    item = input("Qual o nome do produto. Digite: fim para encerrar o programa!")

if count_geral > 0:
    print(f"Contagem de produtos em situação crítica: {count_critico}")
    print(f"Contagem de produtos em estoque baixo: {count_estoque_baixo}")
    print(f"Contagem de produtos em situação de estoque adequado {count_estoque_adequado}")
    if maior_produto == None:
        print("Nenhum produto com diferença positiva foi encontrado!")
    else:
        print(f"O produto que possui a maior quantidade válida é: {maior_produto}, com diferença de {maior_diferenca}")
else:
    print("Nenhum produto válido foi processado!")



```

```python

### Especificando a entrada apenas para números positivos e que não sejam float
count= 0
soma = 0
media = 0
entrada = int(input("Digite um valor: "))
maior = None
menor = None


while entrada !=0:
    entrada = int(input("Digite um número válido: "))
    if entrada  <0:
        print("Valor inválido! Digite apenas positivos!")
    else:
        count +=1
        soma  += entrada
        if maior is None or entrada > maior:
            maior = entrada
        if menor is None or entrada < menor:
            menor = entrada
    entrada = int(input("Digite um valor positivo (0) será encerrado: "))

if count > 0:
    media = soma /count
    print(f"Quantidade:{count} ")
    print(f"Média:{media} ")
    print(f"Maior valor: {maior}")
    print(f"Menor valor: {menor}")

else:
    print("Nenhum valor válido foi informado!")



```

#### 28. Considere o código a seguir:

```python
contador = 0
soma = 0

for i in range(5):
    n = int(input())
    print(i, n)
    if n >=0:
        soma += n
        contador +=1
media = soma/contador
print(media)

### Aparentemente o programa está realizando sim a soma apenas para valores que sejam maiores que 0 e dividindo apenas quando se digita também um valor maior que zero
```

Falha do programa acima:

- Se todos os 5 valores digitados forem negativos, o contador nunca é incrementado.
- Resultado: na linha media = soma / contador, você teria uma divisão por zero.
- Esse é o erro: *ZeroDivisionError*.

---

Correção:



```python
contador = 0
soma = 0

for i in range(5):
    n = int(input())
    print(i,n)
    if n>=0:
        soma+=n
        contador +=1
if contador ==0:
    print("Sem números digitados válidos para cálculo de média!")
else:
    media = soma/contador
    print(media)
```

#### 29. Você está programando a lógica de segurança de um drone de entregas. Enquanto o equipamento está em trânsito para entrega, o sistema receberá todos os dados do usuário e exibirá as mensagens de alerta na tela de saída. O drone só tem permissão para decolar se passar por uma sequência de verificações de sensores que consomem bateria. O drone inicia com uma carga de bateria informada pelo operador. O processo de decolagem consiste em 5 etapas de checagem. Em cada etapa, o programa gera um consumo fixo de 1,7% da bateria total inicial. Além disso, em cada etapa, o programa recebe do sensor de vento a velocidade do vento no momento atual (em km/h). Se, em qualquer uma das 5 etapas, a velocidade do vento for superior a 30 km/h, o drone deve cancelar a decolagem imediatamente por segurança. Caso a bateria caia abaixo de 10% durante qualquer momento das checagens, o processo também deve ser interrompido com o alerta "Bateria Insuficiente". O programa deve informar ao final se a decolagem foi "Autorizada", "Cancelada por Vento Forte"ou "Abortada por Bateria", indicando em qual etapa o problema ocorreu e o nível final de carga.

```python

### Processo de decolagem precisa de 5 etapas
count = 0
bateria = 100
consumo = 1.7
while count <=5:
    vento = int(input("Informe a velocidade do vento:"))
    if vento >30:
        print("Cancelada por vento forte")
        break
    else:
        bateria -= consumo
        count +=1
if bateria <10:
    print("Bateria insuficiente!")
    print(f"Abortada por bateria! {bateria}")
else:
    print("Decolagem do Drone Autorizada!")
```
