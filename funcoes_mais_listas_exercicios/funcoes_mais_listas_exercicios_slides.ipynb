- ○ [3, 7, 10, 15, 20] → True
- ○ [3, 7, 6, 15, 20] → False

```python


sequencia = []
for i in range(1,9+1):
    numero = float(input("Digite uma sequência numérica: "))
    sequencia.append(numero)

def sequencia_positiva(sequencia):
    for i in range(len(sequencia)):
        if sequencia[i] > sequencia[i+1]:
            return False
        else:
            return True
```

#### Desafios: **Medições de Tempo**

#### Um sistema de análise de desempenho recebe uma lista com os tempos, em segundos, que um atleta levou para completar diferentes tentativas de um mesmo percurso. Para auxiliar na análise, crie uma função chamada contar_abaixo_media que receba uma lista de tempos como parâmetro e retorne a quantidade de tentativas cujo tempo ficou abaixo da média.

```python
tempos_segundos = []
while True:
    print("Digite um tempo para acrescentar à lista")
    tempos_medidos = float(input("Digite o tempo em segundos:(-1 para encerrar)  "))
    if tempos_medidos <0:
        break
    tempos_segundos.append(tempos_medidos)
```

```python
tempos_segundos
```

```python
def contar_abaixo_medio(tempos_segundos):
    soma = 0
    count = 0
    for i in tempos_segundos:
        soma +=i
    media = soma/ len(tempos_segundos)

    for i in tempos_segundos:
        if i < media:
            count +=1
    return f"A média foi de{media}",  count,"Tentativas abaixo da média"
    
```

```python
contar_abaixo_medio(tempos_segundos)
```

#### Modularizando o problema:


```python
def media_tempo(lista):
    soma = 0
    for i in lista:
        soma +=i
    media =  soma / len(lista)
    return media
```

```python
def  contar_abaixo_medio_2(lista):
    count = 0
    media = media_tempo(lista)
    for i in lista:
        if i < media:
            count +=1
    return count
```

```python
lista = [12.5, 11.8, 13.2, 10.9, 12.0]
resultado_media =  media_tempo(lista)
print("O resultado da média é: ", resultado_media)
contagem_abaixo_media = contar_abaixo_medio_2(lista)
print("Contagem de valores abaixo da média", contagem_abaixo_media)
```

#### ***Desafio Final***: *Conversor de String*

#### Um sistema de análise textual precisa separar uma palavra em partes menores para processar cada caractere individualmente. Crie uma função chamada converter_para_lista que receba uma string como parâmetro e retorne uma lista contendo cada caractere da string em uma posição diferente.

#### converter_para_lista("python") → ["p", "y", "t", "h", "o", "n"]

```python

### Entrada: Uma função que receberá como parâmetro um caractere

### Processamento: Pegar cada caractere da palavra digitada e adicinar letra por letra em uma posição da lista

### Saída: Uma lista com cada letra da palavra digita em uma posição crescente no index
def converter_para_lista(entrada):
    lista_vazia = []
    for i in range(len(entrada)):
        lista_vazia.append(entrada[i])
    return lista_vazia
```

```python
entrada = input("Digite uma palavra:")
print(converter_para_lista(entrada))
```

#### **11.** Escreva uma função que receba uma lista de números e retorne uma nova lista com cada valor multiplicado por 2.

```python
### Entrada: Função que recebe uma lista de números

### Processamento: Dentro da Função, multiplicar cada número por 2 e adicionar em uma nova lista

### Saída: A nova lista com os números da lista original multiplicado por 2


def dobro(lista_numeros):
    nova_lista = []
    for i in lista_numeros: ### dar erro se usar o len(lista_numeros) Um número sozinho não é iterável → por isso dá erro.
        nova_lista.append(i*2)
    return nova_lista
```

```python
def dobro_2(lista):
    nova_lista = []
    for i in range(len(lista)):
        nova_lista.append(lista[i] * 2)
    return nova_lista
```

```python
lista = [1,2,3,-3,10,20]
lista_dobro = dobro_2(lista)
print(lista_dobro)
```
