### Comandos  IF

#### ***Sintaxe***

- Operadores relacionais
- Operadores Lógicos
        -Experimentação comandos IF


#### ***Características:***

- Estrutura de decisão/condição
    - Faz desvio de Fluxo

- Define-se uma *condição* através de uma **expressão lógica-relacional**

    - Se a condição for verdadeira, executa, caso contrário, ignora o bloco de comandos
    
- Indentação é obrigatória

- Maior: (>): >
- Menor:(<) : <
- Maior ou Igual (>=): >=
- Menor ou Igual (<= ): >=
- Igual(=): ==
- Diferente: (!=): !=





#### Operadores Lógicos

- ***Negação/inversão:*** inverte o valor lógico| **not**

- ***Conjunção:*** Resulta em True quando tudo é True, False caso contrário | **and** --> Só será verdadeiro quando todo mundo for verdadeiroi

- ***Disjunção:*** resulta em True quando ao menos um é True, False caso contrário | **or** --> Quando disjunção só será falso quando todas as condições forem falsas ou verdadeiro quando ao menos uma condição for verdadeira







#### **Operador in:** Usado mais em listas e dicionários

- Verificar o pertecencimento de um item em um conjunto de itens
  - Sintaxe:
  if <elemento> in <iterável>:
      <bloco de comandos>

- Pode ser utilizado com <iteráveis> : listas, tuplas, sets & Dicionários

#### >  Se o elemento pertence a um conjunto: se executa, se não ignora

#### Demonstração de exemplo de código:

1-) Crie um programa que peça para o usuário informar o valor da sua nota. Caso a nota seja maior ou igual a 7.0, então emprima "Aprovado(a)", na tela

```python

nota = float(input("Informe a nota:"))
while nota < 0 or nota >10:
    print("Informe uma nota válida:")
    nota = float(input("Informe uma nota válida"))
if nota >= 7.0:
    print("Aprovado!")
else:
    print("Não foi desta vez")
```

2-) Crie um programa que peça para o que o usuáruo informe o valor de sua nota. Caso a nota informada seja menor que 7.0, mas simultaneamente maior ou igual a 4.0, então imprimir "Tem direito a exame!"

```python

nota = float(input("Informe o valor da nota: "))
while nota <0 or nota >10:
    print(f"Dever ser inserido uma nota válida, a  nota {nota}, não está dentro de 0 À 10 pts")
    nota = float(input("Ensira a nota: "))

if nota >=7:
    print("Aprovado!")

elif nota < 7.0 and nota >=4.0:
    print("Tem direito a refazer o exame!")
else:
    print("Não tem direito a refazer o exame :(")
```

#### 3--) Utilizando só IF´s para realizar a validação:

```python

nota = float(input("Informe o valor da nota: "))
while nota <0 or nota >10:
    print(f"Dever ser inserido uma nota válida, a  nota {nota}, não está dentro de 0 À 10 pts")
    nota = float(input("Ensira a nota: "))

if nota >=7:
    print("Aprovado!")

if nota < 7.0 and nota >=4.0:
    print("Tem direito a refazer o exame!")
if nota <4:
    print("Reprovado")
```

### ***Comandos IF, ELIF, E ELSE***

#### Os comandos if e else:

- Sintaxe

- Experimentos usando o IF e ELSE

#### Basicamente, em resumo:

- Estrutura de decisão composta: Dois Blocos de comandos, com base na condição bloco relacional:

    - Condição **verdadeira**: executa o bloco **if** apenas
    - Condição **Falsa**: executa o bloco de ***else*** apenas

--> Lembrar que não existe ELSE sem IF e não se pode criar vários ELSE para um mesmo IF:

--> O Else não precisa de uma condição explícita, a condição explícita é apenas no IF, fica implícito que a condição do Else é contrária a do IF

- Programa que receve a nota de uma pessoa. Imprimir "Aprovado(a)" na tela, caso a nota seja maior ou igual a 7.0. Caso contrário, imprimir "Reprovado(a)" na tela:

```python
nota = float(input("Informe a  nota é: "))

if nota >=7.0:
    print("Aprovado(a)!")
else: ### nota < 7.0

    print("Reprovado(a)!")
```

## **Início das estruturas de Repetição**

### **Comando WHILE:**


- O comando while 
   -Sintaxe:
- Exemplos

- Inicializa-se uma ou mais variáveis de controle: Antes de bater no While

- Defini-se uma condição de parada
  - Enquanto True: repete mais 1x

- Atualiza-se a(s) variáveis de entrada, fazem que em algum momento essa condição deixe de ser verdadeira

- Cuidado: loop infinito: As variáveis de controles precisam ser inicializadas



#### --> Loop: Executa a condição do início  (várias vezes) enquanto a condição CONTINUAR A SER VERDADEIRA. Caso seja falsa, ignora o bloco de comandos e a condição segue.


--> Antes de entrar no while se **inicializa a ou as variáveis de controle**

```python
count = 0

while count <=10:
    print(f"{count}" )
    count+=1
```

```python
count = 0

while count <=10:
    print(f"{count}","",end="") ### não quebrar a linha
    count+=1
```

### Estrutura de repetição com for:

- Sintaxe
- Iteração ---> O que é uma iteração dentro do contexto de loops laço dentro do contexto de repetição
- Range() 

- Similar ao comando While
    - Permite Repetição: Tantas vezes quanto a lógica necessitar
- *Diferença*
    - Necessita de algo iterável para poder repetir o bloco de comandos
    - Série numérica ou conjunto de elementos (ex: Lista)


- for <variável> in <iterável>:
        - <linha de código 1>
        - <linha de código 2>
        - <linha de código 3>
        - <linha de código n>


#### Semântica: Executa um laço no qual... A variável assume o conteúdo de cada item do <iterável> um a cada iteração

#### A cada iteração a variálvel temporária assume o conteúdo daquele elemento

```python
for i in [1,2,3,4,5]:
    print(i)
```

#### Função Range()

- Sintaxe range(início, fim, passo)

- Semântica: produz uma sequência numérica (iterável)
    - De um <início>(incluindo o elemento inicial)
    - Até um <fim> (excluindo o elemento final)
    - Passo a <passo>
- Comumente utilizada com o comando for

```python
for dados in range(0,21,1):
    print(dados,"",end="")

### Para cada vez que o i mudou de valor, se executou o bloco de comandos
```
