## ***For***

> Percorrer Elemento, sequência, estruturas de dados no Python que são iteráveis - percorrendo de elemento a elemento

--> `Utilizado normalmente para percorrer sequências, estrutura de dados que são iteráveis, ou seja pode-se passar de elemento a elemento`

```python
nome = "LuElias"
for letra in nome: ### letra: variável temporária ### Nome variável declarada fora do for fica trocando de valor a cada rodada....Passando pelos
                   ###  i de índice: Elementos
    print(letra)
```

***Lógica:*** *Uma variável temporária declarada dentro do for que percorrerar outra variável de interesse*

> Percorre os elementos de um objeto

--> O próprio for já realiza essa troca

```python
nome = "LuElias"
for i in nome: ### letra: variável temporária ### Nome variável declarada fora do for
    print(i)
```

#### Quais numeros sao diviseis por 4 no intervalor [4-100]?


```python
numero = 4
max_valor = 100
for i in range(1, max_valor):
    resto =  (i+4)%4
    if resto  == 0:
        print(i)
```

> Escreva um programa que receba uma lista de números do usuário e conte quantas vezes um número específico aparece na lista. Solicite ao usuário um número e exiba a contagem

```python
lista = [1,2,3,5,6,2,4,5,90,90,7,8,9,10,90,1933423,19384,143,33444,555335]
count=0
numero = float(input("Digite um número para compar a lista: "))
for elemento in lista:
    if elemento ==numero:
        count+=1

print(f"O número aparece {count} vezes na lista")
```

> Listas são objetos mutáveis != de strings que são fixas

```python
#### Lista vazia:

idades_2 = []

#### Como não sabe exatamente quantos dados serão inputados, se usa o while
while True:
    entrada = input("Entre com a idade: ")
    if entrada =="":
        break
    idades_2.append(int(entrada))
print(idades_2)
```

```python
media_idades = sum(idades_2)/len(idades_2)
quantidade_de_dados_lista = len(idades_2)
menor_idade = min(idades_2)
maior_idade = max(idades_2)
```

```python
print("Média: ",media_idades)
print("Minimo: ", menor_idade)
print("Maior:", maior_idade)
print("Quantidade", quantidade_de_dados_lista)
```
