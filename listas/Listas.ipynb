#### No final, mostre qual foi o **maior** e o **menor** valor digitado e as suas respectivas **posições** na lista.


```python
def lista_maior_menor():
    lista_num = []
    for i in range(5):
        valores = int(input("Digite os valores: "))
        lista_num.append(valores)
    pos_maior = ""
    pos_menor = ""
    maior = float("-inf")
    menor = float("inf")
    
    for pos, valor in enumerate(lista_num):
        if valor > maior:
            maior = valor
            pos_maior = pos
        if valor < menor:
            menor = valor
            pos_menor = pos
    return f'Lista:{lista_num}\nMaior valor = {maior} Posição= {pos_maior}\nMenor valor = {menor} Posição= {pos_menor}'
```

```python
def funcao_lista_pares_impar(lista):
    lista_pares = []
    lista_impar = []
    for i in range(len(lista)):
        if lista[i]%2 ==0:
            lista_pares.append(lista[i])
        if lista[i]%2!=0:
            lista_impar.append(lista[i])
    return f'Lista: {lista}\nLista Pares:{lista_pares}\nlista impar {lista_impar}'
```

```python
### Resolvendo com dicionários so para ter uma ideia:

dict_pessoas = dict()
while True:
    nome_pessoa = input("Digite o nome da pessoa: ")
    peso = float(input("Digite o peso da pessoa: "))
    dict_pessoas[nome_pessoa] = peso
    print(f"Pessoa Cadastrada: {dict_pessoas}")
    continuar = input("Deseja continuar o programa? sim ou nao").lower()
    while continuar !="sim" and continuar != "nao":
        print("Digite apenas sim ou não")
        continuar = input("Deseja continuar o programa? sim ou nao").lower()
    if continuar =="nao":
        print("Encerrando!")
        break
    else:
        continue
```

```python
def analisa_pessoas(dicio):
    lista_pesadas = []
    lista_leves = []
    maior = float("-inf")
    menor = float("inf")
    count = len(dicio)
    for chave, valor in dicio.items():
        if valor > maior:
            maior = valor
        if valor < menor:
            menor = valor
    for chave, valor in dicio.items():
        if valor == maior or valor > maior:
            lista_pesadas.append(chave)
        if valor < menor or valor == menor:
            lista_leves.append(chave)
    return f'Quantidade de pessoas :{count}\nListagem Pessoas Pesadas: {lista_pesadas}\nListagem pessoas Leves: {lista_leves}'
```


### ✅ Para acessar valores de uma coluna:
### Você percorre as linhas da matriz.
### Então usa range(len(matriz)), porque len(matriz) retorna o número de linhas.

#### Exemplo:


- for i in range(len(matriz)):
    - print(matriz[i][coluna])

### ✅ Para acessar valores de uma linha:
### Você percorre as colunas dessa linha.
### Então usa range(len(matriz[linha])), porque len(matriz[linha]) retorna o número de colunas daquela linha específica.
#### Exemplo:

- for j in range(len(matriz[0])):  # aqui peguei a primeira linha
    - print(matriz[0][j])

### 🔍 Ou seja:

### len(matriz) → número de linhas.

### len(matriz[linha]) → número de colunas daquela linha.

#### Isso funciona tanto para matrizes quadradas quanto para não quadradas, porque cada linha pode ter um tamanho diferente.
