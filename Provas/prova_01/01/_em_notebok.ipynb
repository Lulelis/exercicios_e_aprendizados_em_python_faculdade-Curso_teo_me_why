#### (1,5 ponto) Em sistemas de autenticação é comum monitorar sequências consecutivas de tentativas de acesso malsucedidas para identificar comportamentos suspeitos. Em um ambiente corporativo, uma equipe de desenvolvimento implementou um programa em Python para ler o resultado de várias tentativas de login, registradas uma a uma. Cada tentativa pode ser classificada como:

#### S → sucesso no login;

#### F → falha no login;

#### X → fim da entrada.

#### O objetivo do programa é determinar a maior sequência consecutiva de falhas (F) ocorrida antes do encerramento da entrada. Considere o seguinte código:

```python
tentativa = input("Digite o resultado da tentativa de login:")
atual = 0
maior = 0

while tentativa != "X":
    if tentativa == "F":
        atual = atual + 1
    else:
        if atual > maior:
            maior = atual
        atual = 0

    if atual > maior and tentativa == "S":
        maior = atual

    tentativa = input("Digite o resultado da tentativa de login:")

print(maior)
```

#### Considere que o comportamento esperado do programa é imprimir a quantidade correspondente à maior sequência consecutiva de falhas (F). Para validar o programa, foram elaborados os seguintes casos de teste:

| Caso de Teste | Entradas       | Saída esperada |
|---------------|----------------|----------------|
| Caso 1        | F, F, S, F, X  | 2              |
| Caso 2        | S, F, F, X     | 2              |
| Caso 3        | F, F, F, X     | 3              |
| Caso 4        | S, S, F, S, X  | 1              |

#### Com base na análise do código e dos casos de teste, avalie as afirmações a seguir.

#### I. No Caso 1, o programa produz a saída esperada porque a maior sequência de falhas é finalizada antes do encerramento da entrada, permitindo que o valor acumulado seja comparado com a variável maior.
