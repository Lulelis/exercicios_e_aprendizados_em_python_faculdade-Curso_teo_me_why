#### (1,5 ponto) Em sistemas de autenticação é comum monitorar sequências consecutivas de tentativas de acesso malsucedidas para identificar comportamentos suspeitos. Em um ambiente corporativo, uma equipe de desenvolvimento implementou um programa em Python para ler o resultado de várias tentativas de login, registradas uma a uma. Cada tentativa pode ser classificada como:

#### S → sucesso no login;

#### F → falha no login;

#### X → fim da entrada.

#### O objetivo do programa é determinar a maior sequência consecutiva de falhas (F) ocorrida antes do encerramento da entrada. Considere o seguinte código:

```python
tentativa = input("Digite o resultado da tentativa de login:")
atual = 0
maior = 0

while tentativa != "X":
    if tentativa == "F":
        atual = atual + 1
    else:
        if atual > maior:
            maior = atual
        atual = 0

    if atual > maior and tentativa == "S":
        maior = atual

    tentativa = input("Digite o resultado da tentativa de login:")

print(maior)
```

| Caso 2        | S, F, F, X     | 2              |
| Caso 3        | F, F, F, X     | 3              |
| Caso 4        | S, S, F, S, X  | 1              |

#### (1,5 ponto) Em um sistema de monitoramento de rede, uma equipe de desenvolvimento precisa calcular quantos pares distintos de dispositivos podem trocar dados diretamente entre si em um conjunto com  n dispositivos. Duas soluções em Python foram propostas para realizar esse cálculo.

#### -Código 1:

```python
n = int(input("Digite o valor de n: "))
total = 0
restantes = n - 1

for i in range(1, n):
    total += restantes
    restantes -= 1

print(total)
```

#### Código 2:

```python
n = int(input("Digite o valor de n: "))
total = 0

for i in range(1, n + 1):
    for j in range(i + 1, n + 1):
        total += 1

print(total)
```
