```python
lista_7 = []

for i in range(7):
    numeros = int(input("Insira 7 números inteiros: "))
    lista_7.append(numeros)
print(lista_7)
```

```python
def verifica_se_palindromo(lista):
    lista_normal = []
    lista_reverse = []
    
    for i in lista:
        lista_normal.append(i)

    for j in range(len(lista)-1,-1,-1):
        lista_reverse.append(lista[j])
    if lista_normal == lista_reverse:
        return True
    else:
        return False, f"Não é palíndromo"
```

#### Crie um programa em Python que leia 12 números inteiros e armazene os valores em uma lista.

#### Depois, encontre o tamanho da maior sequência consecutiva em que os valores estão crescendo.

#### Uma sequência crescente acontece quando cada valor é maior que o valor anterior.

- Exemplo de entrada

- 3
- 5
- 7
- 2
- 4
- 6
- 8
- 1
- 2
- 10
- 9
- 11

#### Explicação

- Nesse caso, algumas sequências crescentes são:

- 3, 5, 7
- 2, 4, 6, 8
- 1, 2, 10
- 9, 11

#### A maior sequência crescente tem tamanho 4.

**Saída esperada**
***Maior sequência crescente: 4***

```python
lista_sequencia = []

for i in range(12):
    numeros = int(input("Insira 12 números para compor uma sequência"))
    lista_sequencia.append(numeros)
print(lista_sequencia)

```

```python
def maior_em_sequencia(lista):
    count_passos = 1
    maior_sequencia = 0
    for i in range(len(lista)-1):
        if lista[i] < lista[i+1] :
            count_passos +=1
            if count_passos > maior_sequencia:
                maior_sequencia = count_passos
        else:
            count_passos = 1
    return maior_sequencia
```

#### > Dúvida do por que iniciar com count_passos =1 é:  como é uma *lista de números*, cada número isolado já é uma sequência de tamanho 1. Se começasse com 0, mesmo uma sequência mínima não seria contada. Uma sequência crescente de 3 números: [3,5,7] , se começa com count_passos = 1 (o 3 já conta, começa com count_passos = 1 o 3 já conta, depois cresce para 2 (5), depois 3 (7).). Se começasse com 0, só se chegaria na 2 no final do tamanho da sequência; **O 1 é para garantir que cada elemento já iniciei como uma sequência válida**

```python
print("O Tamanho da sequência será", maior_em_sequencia(lista_sequencia))
```

```python
def analise_palavra(lista):
    #menor = len(lista[0]) --> estou guardando números aqui
    #maior = len(lista[0]) --> Guardando números
    maior = lista[0]
    menor = lista[0]
    count_palavras_a = 0
    mais_de_5 = []

    for i in lista:
        if i[0] =="a":
            count_palavras_a +=1

    for i in range(len(lista)):
        if len(lista[i]) > len(maior):
            maior = lista[i]
           
        elif len(lista[i]) < len(menor):
            menor = lista[i]
            
          
        if len(lista[i]) >  5:
            mais_de_5.append(lista[i])

    return maior, menor, count_palavras_a, mais_de_5
```

```python
maior_palavra, menor_palavra, countador_palavras_a, mais_de_5 = analise_palavra(lista_palavras)
print(f"A maior palavra é: {maior_palavra}")
print(f"A menor palavra é: {menor_palavra}")
print(f"Quantidade de palavras que começam com a {countador_palavras_a}")
print(f"Palavras com mais de 5 caracteres {mais_de_5}")
```
