#### > <u>**Exercício 7 Aplicando reajuste em preços** </u>:

```python
def reajuste_precos(lista_preco):
    preco_reajustado = []
    for i in range(len(lista_preco)):
        novo_preco = 0
        if lista_preco[i] < 100.00:
            novo_preco = lista_preco[i] + (10/100 * lista_preco[i])
            preco_reajustado.append(novo_preco)
        elif lista_preco[i] >= 100.00:
            novo_preco = lista_preco[i] + (5/100 * lista_preco[i])
            preco_reajustado.append(novo_preco)
    print(f"Lista Preços Originais: {lista_preco}\nPreço Reajustado: {preco_reajustado}")
```

```python
lista_precos = []
for i in range(1, 8+1):
    precos = float(input(f"Digite o {i}º preço do produto: "))
    lista_precos.append(precos)
print(lista_precos)
```

```python
reajuste_precos(lista_precos)
```

#### Crie um programa em Python que leia o nome de 8 pessoas e armazene os nomes em uma lista.

#### Depois, leia um nome a ser procurado.

#### O programa deve exibir:

- se o nome foi encontrado ou não;
- quantas vezes esse nome aparece na lista.
- A comparação deve considerar o nome exatamente como foi digitado.

#### Exemplo de entrada

- Ana
- Bruno
- Carlos
- Ana
- João
- Maria
- Ana
- Pedro
- Ana

#### Saída esperada

- Nome encontrado.
- Quantidade de ocorrências: 3


```python
def finding_nomes(lista_nomes, nome_procurado):
    encontrado = True
    count_vezes = 0
    for i in range(len(lista_nomes)):
        if nome_procurado not in lista_nomes:
            encontrado = False
        else:
            if nome_procurado == lista_nomes[i]:
                count_vezes+=1
    if encontrado:
        return f"O nome: {nome_procurado} está na Lista!Quantidade Ocorrências: {count_vezes}"
    elif not encontrado:
        return f"O nome: {nome_procurado} NÃO está na lista: {lista_nomes}"
```

#### Uma empresa registrou a temperatura de uma máquina em 10 momentos diferentes. Crie um programa em Python que leia essas temperaturas e armazene os valores em uma lista.

> Depois, percorra a lista e conte:

- quantas vezes a temperatura aumentou em relação à medição anterior;
- quantas vezes a temperatura diminuiu em relação à medição anterior;
- quantas vezes a temperatura permaneceu igual.

#### Exemplo de entrada
- 30
- 32
- 32
- 31
- 35
- 36
- 34
- 34
- 38
- 37

#### Saída esperada
- Aumentou: 4 vezes
- Diminuiu: 3 vezes
- Permaneceu igual: 2 vezes



```python
def comparacao_temperatura(lista_temp):
    temp_aumentou = 0
    temp_diminuiu = 0
    temp_igual = 0
    for i in range(len(lista_temp)-1):
        if lista_temp[i] < lista_temp[i+1]:
            temp_aumentou +=1
        if lista_temp[i] > lista_temp[i+1]:
            temp_diminuiu +=1
        if lista_temp[i] == lista_temp[i+1]:
            temp_igual +=1
    print(f"Aumentou: {temp_aumentou}\nDiminuiu: {temp_diminuiu}\nPermaneceu Igual: {temp_igual}")
```

```python
def lista_palindromo(lista_num):
    palindromo = True
    for i in range(len(lista_num)-1):
        if lista_num[i] != lista_num[-(i+1)]:
            palindromo = False
    if palindromo:
        return "A Lista É Palíndromo!"
    elif not palindromo:
        return "A lista não é Palíndromo!"
```

### > <u>Explicando a utilização do [-(i+1)] nesse exercício</u>:

--> Acessar os elementos de trás para frente

#### --> Lembrando que em Python, índices negativos comtam a partir do fim da lista:

    - lista_num[-1] --> último elemento
    - lista_num[-2] --> Penúltimo
    - lista_num[-3] --> antepenúltimo

#### --> Portanto, quando se escreve lista_num[-(i+1)], está:

    - Se i=0, pega o lista_num[-1] (último elemento);
    - Se i=1, pega lista_num[-2] (Penúltimo);
    - Se i=2, pega lista_num[-3] (antepenúltimo);
    - E assim por diante
#### Dessa forma po [-(i+1)] não significa "o próximo elemento", mas sim o "elemento espelhado no fim da lista" **E isso garante a comparação simétrica:** primeiro <-> último, segundo <-> penúltimo, etc

#### Então:

- O i+1  funciona como o “o próximo” está correto em *outro contexto* (quando percorremos a lista normalmente).

Mas aqui o sinal de menos muda tudo: ele serve para inverter a direção e pegar do fim.

- Palavras com mais de 5 caracteres: ['banana', 'abacaxi', 'algoritmo', 'janela', 'computador']

#### **Objetivo:** manipular listas de strings, usar len() em strings, filtrar palavras e comparar tamanhos de textos

```python
def analise_palavras(lista_palavras):
    maior_palavra_digitada = float("-inf")
    menor_palavra_digitada = float("inf")
    palavras_letra_a = 0
    lista_5_caracteres = []
    maior_palavra = ""
    menor_palavra = ""
    for i in range(len(lista_palavras)):
        qtd_palavras = 0

        for j in lista_palavras[i]:
            qtd_palavras +=1
        if lista_palavras[i][0]== "a":
            palavras_letra_a +=1
        if qtd_palavras > maior_palavra_digitada:
            maior_palavra_digitada = qtd_palavras
            maior_palavra = lista_palavras[i]
        if qtd_palavras < menor_palavra_digitada:
            menor_palavra_digitada = qtd_palavras
            menor_palavra = lista_palavras[i]
        if qtd_palavras > 5:
            lista_5_caracteres.append(lista_palavras[i])
    print(f"Maior palavra digitada: {maior_palavra}\n Menor Palavra digitada: {menor_palavra}\nQuantidade palavras que começam com a: `{palavras_letra_a}\nPalavra com mais de 5 caracteres: {lista_5_caracteres}")

```
