```python
lista_num_10 = []
for i in range(10):
    numeros = int(input(f"Digite os números!\nNúmero {i+1}: "))
    lista_num_10.append(numeros)
for i in lista_num_10:
    print(i)
```

```python
def segundo_maior_valor(lista_num):
    maior_valor = lista_num[0]
    segundo_maior = lista_num[1]
    for i in range(len(lista_num)):
        if maior_valor < lista_num[i]:
            maior_valor = lista_num[i]
    for i in range(len(lista_num)):
        if lista_num[i] != maior_valor and lista_num[i] > segundo_maior:
            segundo_maior = lista_num[i]
    print(f"Maior valor: {maior_valor}\n Segundo maior valor: {segundo_maior}")
```

```python
segundo_maior_valor(lista_num_10)
```
