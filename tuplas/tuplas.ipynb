#### 💡 Desafio 072 — Números por Extenso

#### Crie um programa que tenha uma **tupla** totalmente preenchida com uma contagem por extenso,   de **zero** até **vinte**.  

#### Seu programa deverá ler um número pelo teclado (**entre 0 e 20**)   e mostrá-lo por **extenso**.


```python
def verifica_tupla( entrada):
    tupla_numeros = ("zero", "um", "dois", "tres", "quatro", "cinco", "seis", "sete", "oito","nove", "dez", "onze", "doze", "treze", "quatorze", "quinze", "dezesseis", "dezessete", "dezoito", "dezenove", "vinte")

    for pos, numero in enumerate(tupla_numeros):
        if entrada == pos:
            print(f"Você digitou {numero}")
        
```

```python
inicializador = input("Deseja inicializar o programa?!").lower()
while inicializador != "sim" and inicializador != "nao":
    print("Digite apenas sim e não: ")
    inicializador = input("Deseja inicializar o programa?!").lower()
while inicializador != "nao":
    print("Verifica se o número está na tupla ou não")
    num = int(input("Digite o número para verificar se ele está contido: "))
    while num <0 or num >20:
        print("Um número entre zero e vinte")
        num = int(input("Digite o número para verificar se ele está contido: "))
    print(verifica_tupla(num))
    inicializador = input("Deseja inicializar o programa?!").lower()
    while inicializador != "sim" and inicializador != "nao":
        print("Digite apenas sim e não: ")
        inicializador = input("Deseja inicializar o programa?!").lower()
```

#### 💡 Desafio 073 — Tabela do Campeonato Brasileiro

#### Crie uma **tupla** preenchida com os **20 primeiros colocados** da Tabela do **Campeonato Brasileiro de Futebol**, na ordem de colocação.  

#### Depois, mostre:


- A) Apenas os **5 primeiros colocados**.  
- B) Os **últimos 4 colocados** da tabela.  
- C) Uma lista com os times em **ordem alfabética**.  
- D) Em que **posição** na tabela está o time da **Chapecoense**.


```python
def adiocinar_elementos_lista():
    lista_times = []
    entrada = int(input("Quantos elementos adicionar na lista? "))
    for i in range(entrada):
        times = input("Digite o time a ser adicionado: ").lower()
        lista_times.append(times)
    tuplas_times = tuple(lista_times)
    return tuplas_times
```

#### Exbindo os 5 primeiros com Slicing:

```python

tupla_times = adiocinar_elementos_lista()
```

```python
for pos, valor in enumerate(tupla_times):
    print(f'Posição:{pos} Valor: {valor}')
```

```python
cinco_primeiros = tupla_times[0:5]
print(cinco_primeiros)
```

```python
### ultimos colocados
ultimos_quatro = tupla_times[-4:]
print(ultimos_quatro)
```

#### --> Posicao Chapecoense


```python
posicao = ""

for pos, valor in enumerate(tupla_times):
    if valor =="chapecoense":
        posicao = pos
print("A posição de Chapecoense é ",posicao)
```

```python
time_ordenado = sorted(tupla_times)
for i in range(len(time_ordenado)):
    print(time_ordenado[i])
            
```

#### 💡 Desafio 074 — Números Aleatórios em uma Tupla

#### Crie um programa que vai gerar **cinco números aleatórios** e colocar em uma **tupla**.  

#### Depois disso, mostre a **listagem de números gerados** e também indique o **menor** e o **maior** valor que estão na tupla.


```python
import random
```

```python
def gerar_numeros_aleatorios():
    lista = []
    for i in range(1,5+1):
         numero = random.randint(1,1000)
         lista.append(numero)
    tupla_aleatoria = tuple(lista)

    maior = float("-inf")
    menor = float("inf")
    for i in range(len(tupla_aleatoria)):
        if tupla_aleatoria[i] > maior:
              maior = tupla_aleatoria[i]
        if tupla_aleatoria[i] < menor:
             menor = tupla_aleatoria[i]
    return tupla_aleatoria, maior, menor
```

```python
gerar_numeros_aleatorios()
```


#### 💡 Desafio 075 — Análise de Valores em uma Tupla

#### Desenvolva um programa que leia **quatro valores** pelo **teclado** e guarde-os em uma **tupla**.  
No final, mostre:

- A) Quantas vezes apareceu o valor **9**.  
- B) Em que posição foi digitado o primeiro valor **3**.  
- C) Quais foram os números **pares**.



```python

def lendo_valores_tupla():
    lista_valores = []
    for i in range(4):
        print("Digite um número: ")
        num = float(input("Digite o valor: "))
        print(f"Número digitado: {num}")
        lista_valores.append(num)
    tupla_valores = tuple(lista_valores)
    count_valor_9 = 0
    posicao_valor_3 = 0
    num_pares = 0 
    for pos, valor in enumerate(tupla_valores):
        if valor == 9:
            count_valor_9 +=1
        if valor == 3:
            posicao_valor_3 = pos
        if valor%2 ==0:
            num_pares +=1
    for i in tupla_valores:
        print(f"Valores digitados: {i}: ", end="")
    print(f"Valor 9 foi digitado: {count_valor_9} vezes")
    if posicao_valor_3 ==0:
        print("Valor 3 foi digitado 0 vezes")
    else:
        print(f"Valor 3 foi digitado: {posicao_valor_3}")
    print(f"Quantidade de valores pares: {num_pares}")
    return f"Tupla valores: {tupla_valores}\n Quantidade de valors iguais a 9: {count_valor_9}\nPosição Valor 3 {posicao_valor_3}\n Tupla valores Pares: {tupla_pares}"
```

```python
print(lendo_valores_tupla())
```

#### 💡 Desafio 076 — Listagem de Produtos e Preços

#### Crie um programa que tenha uma **tupla única** com **nomes de produtos** e seus respectivos **preços**, na sequência.  

#### No final, mostre uma **listagem de preços**, organizando os dados em forma **tabular**.


```python

def tuplas_produtos():
    lista_produtos_precos = []
    entrada = input("Deseja iniciar o progama?| Sim ou nao").lower()
    while entrada != "sim" and entrada  != "nao":
        print("Digitar apneas sim ou não")
        entrada = input("Deseja iniciar o progama?| Sim ou nao").lower()
    while entrada != "nao":
        print("Tupla de Produtos e Preços")
        produto = input("Digite o nome do produto: ")
        preco = float(input(f"Digite o preço do produto: {produto}"))
        lista_produtos_precos.append(produto)
        lista_produtos_precos.append(preco)
        entrada = input("Deseja iniciar o progama?| Sim ou nao").lower()
        while entrada != "sim" and entrada  != "nao":
            print("Digitar apneas sim ou não")
            entrada = input("Deseja iniciar o progama?| Sim ou nao").lower()
    tupla_produtos = tuple(lista_produtos_precos)
    print("-"*30)
    print(f'{"Listagem de Preços":^40}')
    for i in range(len(tupla_produtos)):
        if i %2 ==0:
            print(f'"Produto:{tupla_produtos[i]:.<30}"', end="")
        else:
            print(f'"Valor:R${tupla_produtos[i]:>10.2f}"')
```

#### 💡 Desafio 077 — Vogais em Palavras

#### Crie um programa que tenha uma tupla com várias palavras (não usar acentos).
#### Depois disso, mostre, para cada palavra, quais são as suas vogais.


```python

def tupla_palavras():
    lista_palavras = []
    entrada = input("Deseja iniciar o programa que adiciona palavras?: Digite sim ou nao ").lower()
    while entrada != "sim" and entrada != "nao":
        print("Apenas sim ou não!")
        entrada = input("Deseja iniciar o programa que adiciona palavras?: Digite sim ou nao ").lower()
    while entrada !="nao":
        print("Digite a palavra: ")
        palavra = str(input("QUAL A PALAVRA SEM ACENTO")).lower()
        lista_palavras.append(palavra)
        entrada = input("Deseja continuar o pograma que adiciona palavras?: Digite sim ou nao ").lower()
        while entrada != "sim" and entrada != "nao":
            print("Apenas sim ou nao!")
            entrada = input("Deseja iniciar o programa que adiciona palavras?: Digite sim ou nao ").lower()
    tupla_palavras = tuple(lista_palavras)
    for i in range(len(tupla_palavras)):
        count_vogais = ""
        for j in tupla_palavras[i]:
            if j  in ("aêáeêéiíoóúuûô"):
                count_vogais +=j
        print(f"Na palavra: {tupla_palavras[i].upper()} tem as vogais: {count_vogais}")
    
```

```python
print(tupla_palavras())
```
